**대안 가설: 지금 잡히는 건 P 형태가 아니라 PR 분절 / QRS 개시 형태(이상전도)다.**


⚠️ **이건 상한이다** — 주입은 모든 S 비트에 **동일한 파형**을 얹으므로 두 템플릿

⚠️ **null 을 다시 안 재는 근거**: 라벨을 섞으면 주입된 파형이 섞인 라벨과 무관해지므로
`medS`·`medN` 양쪽에 고루 섞인다 → 영분포는 기저와 같다. **셔플을 3배로 늘리지 않는다.**

````python
# CELL 0 — 공용 사전점검
import numpy as np
from scipy import stats

def decide(lo, hi, thr, direction):
    if direction not in (">", "<"):
        raise ValueError("direction 은 '>' 또는 '<'")
    if direction == ">":
        if lo > thr: return "✅ 지지"
        if hi < thr: return "❌ 기각"
    else:
        if hi < thr: return "✅ 지지"
        if lo > thr: return "❌ 기각"
    return "⚠️ 미결"

def equiv(lo, hi, margin):
    """**등가성 판정**(R29 ①). CI **전체**가 ±margin 안이어야 등가."""
    if lo > -margin and hi < margin:
        return "✅ 등가"
    if lo > margin or hi < -margin:
        return "❌ 차이 있음"
    return "⚠️ 미결"

def need_n(n, lo, hi, mean, margin):
    """등가 판정에 필요한 개체 수(R30 ①). 점추정이 여유 밖이면 **None(불가능)**."""
    if not (np.isfinite(lo) and np.isfinite(hi) and np.isfinite(mean)) or n < 1:
        return float("nan")
    slack = margin - abs(mean)
    if slack <= 0:
        return None
    return float(n) * (((hi - lo) / 2.0) / slack) ** 2

def judge(mean, lo, hi, n, margin):
    """★ **등가와 우월성을 한 관문에서 둘 다**(R31 ①) + 필요표본(R30 ①)."""
    eq = equiv(lo, hi, margin)
    sup = decide(lo, hi, 0.0, ">")
    nn = need_n(n, lo, hi, mean, margin)
    if nn is None:
        frame = (f"⛔ **등가 판정 불가** — 점추정 {mean:+.4f} 이 여유 ±{margin} 밖이라 "
                 "어떤 표본으로도 불가능 → **우월성 + 효과크기 프레임**(R31 ①)")
    elif not np.isfinite(nn):
        frame = "⚠️ 필요 개체 계산 불가"
    else:
        frame = f"등가에 필요한 개체 ≈ **{nn:.0f}** (현재 {n})"
    return eq, sup, nn, frame

def ladder_read(lad):
    """★★ **`raw` 앵커로 교란을 판정한다**(R32 ② — Q7-P 가 여기서 틀렸다).

    교란이 차이를 만든다면 **무통제(`raw`)가 최댓값**이어야 한다. 아니면 사다리 내부의
    단조 감쇠는 「교란 제거」가 아니라 **중간 추정량(잔차화)의 인공물**이다.
    `lad` 는 {'raw':…, 'lin':…, …, 'strat':…} 순서 보장 dict."""
    if "raw" not in lad:
        return "⛔ `raw` 앵커가 없다 — 교란 판정 불가(R32 ②)"
    ks = [k for k in lad if np.isfinite(lad[k])]
    if len(ks) < 3:
        return "⚠️ 사다리가 짧다"
    vals = [lad[k] for k in ks]
    amax = ks[int(np.argmax(np.abs(vals)))]
    total = lad[ks[-1]] - lad["raw"]
    if amax != "raw":
        return (f"▸ **`raw` 가 최댓값이 아니다**(최대 `{amax}`) → 사다리 내부 감쇠를 "
                f"**교란으로 읽지 않는다**(잔차화 인공물). 실제 통제 효과 = "
                f"raw→{ks[-1]} **{total:+.4f}**")
    if abs(lad[ks[-1]]) <= 0.7 * abs(lad["raw"]):
        return (f"⚠️ **`raw` 가 최댓값이고 단조 감쇠 → 잔여 교란**(R31 ③). "
                f"실제 통제 효과 raw→{ks[-1]} **{total:+.4f}**")
    return f"▸ `raw` 가 최댓값이나 감쇠가 약하다. raw→{ks[-1]} **{total:+.4f}**"

class AssetError(RuntimeError): pass
print("CELL 0 ✅")
````

````python
# CELL 3 — 【Q-A】 특징 · 층 키 · **QRS 개시 실측** · 점수
from sklearn.metrics import roc_auc_score
from sklearn.linear_model import LogisticRegression

# ── 특징 (Q7-N/P 승계 — 여기서 새로 고르지 않는다)
def local_base(pre_v, k):
    n = len(pre_v); out = np.empty(n); med = float(np.median(pre_v))
    for i in range(n):
        a = max(0, i - k)
        out[i] = med if i - a < 3 else float(np.median(pre_v[a:i]))
    return out

def trend(pre_v, w):
    n = len(pre_v); out = np.zeros(n); x = np.arange(w, dtype=float)
    xc = x - x.mean(); den = float((xc * xc).sum())
    for i in range(n):
        a = i - w
        if a < 0:
            continue
        y = pre_v[a:i]
        out[i] = float(((y - y.mean()) * xc).sum() / den)
    return out

def all_feats(pre_v, post_v):
    """**`post_rr` 은 기저에 안 들어간다**(R28 ②) — 보고용 `f3` 에만 쓴다."""
    med = float(np.median(pre_v)); n = len(pre_v)
    F = {"f1": med - pre_v}
    for k in sorted(set(BASIS_K) | set(PROBE_K) | {HIST_K}):
        F[f"f2_{k}"] = 1.0 - pre_v / np.maximum(local_base(pre_v, k), 1e-9)
    b16 = local_base(pre_v, LB_K); b_first = local_base(pre_v, BASIS_K[0])
    F["f6"] = 1.0 - b16 / max(med, 1e-9)
    cv = np.empty(n)
    for i in range(n):
        a = max(0, i - TREND_W); w_ = pre_v[a:i] if i - a >= 3 else pre_v[:3]
        cv[i] = float(np.std(w_) / max(np.mean(w_), 1e-9))
    F["f4"] = cv
    F["trend"] = trend(pre_v, TREND_W)
    F["f5"] = np.r_[0.0, F[f"f2_{BASIS_K[0]}"][:-1]]
    F["f1_rank"] = stats.rankdata(F["f1"]) / n
    F["f3"] = 1.0 - (pre_v + post_v) / np.maximum(2.0 * b_first, 1e-9)
    F["f3_glob"] = 1.0 - (pre_v + post_v) / (2.0 * max(med, 1e-9))
    return F

def basis_mat(F, kind):
    cols = ["f1"] + [f"f2_{k}" for k in BASIS_K] + ["f6"]
    Z = np.stack([F[c] for c in cols], axis=1)
    if kind == "quad":
        Z = np.c_[Z, Z ** 2]
    elif kind == "hist":
        Z = np.c_[Z, F["f4"], F["trend"], F[f"f2_{HIST_K}"]]
    elif kind == "rank":
        Z = np.stack([stats.rankdata(F[c]) / len(F[c]) for c in cols], axis=1)
    elif kind != "lin":
        raise AssetError(f"기저 종류 {kind} 를 모른다")
    Z = (Z - Z.mean(0)) / (Z.std(0) + 1e-12)
    return np.c_[np.ones(len(Z)), Z]

def prep(s, kind):
    s = np.asarray(s, float)
    return stats.rankdata(s) / len(s) if kind == "rank" else s

RESID_EPS = 1e-9
def residualize(s, Z):
    s = np.asarray(s, float)
    if not np.isfinite(s).all():
        return None
    beta, *_ = np.linalg.lstsq(Z, s, rcond=None)
    e = s - Z @ beta
    if float(np.std(e)) <= RESID_EPS * (float(np.std(s)) + 1e-12):
        return np.zeros_like(e)
    return e

def strat_key(F):
    """★ 주 분석의 층 — `f1` **정확값** × `f2_16` 구간.
    `f1` 정확 일치 ⇔ `pre_rr` 정확 일치 ⇒ **직전 T 위치도 정확히 통제**된다(R32 ①).
    그래서 이 노트북은 마스크를 **안 만든다**."""
    pre_key = np.unique(np.round(F["f1"], 6), return_inverse=True)[1].astype(np.int64)
    f2b = np.floor(F[f"f2_{LB_K}"] / F2_BIN).astype(np.int64)
    return pre_key * (int(f2b.max() - f2b.min()) + 1) + (f2b - f2b.min())

def strat_auc(sc, tt, key, min_s=1, min_n=1):
    sc = np.asarray(sc, float)
    if not np.isfinite(sc).all():
        return float("nan")
    uq, inv = np.unique(key, return_inverse=True)
    num = den = 0.0
    for j in range(len(uq)):
        m = inv == j
        s_, n_ = sc[m & tt], sc[m & ~tt]
        if len(s_) < min_s or len(n_) < min_n:
            continue
        gt = float((s_[:, None] > n_[None, :]).sum())
        eq = float((s_[:, None] == n_[None, :]).sum())
        num += gt + 0.5 * eq; den += float(len(s_) * len(n_))
    return num / den if den >= 1 else float("nan")

# ── ★ 비트별 QRS 개시 실측 (민감도 · Q4)
def qrs_onset(B1):
    """리드 0 기울기로 QRS 개시를 찾는다. R 근처 최대 |기울기| 의 `ONSET_FRAC` 아래로
    `ONSET_RUN` 연속 떨어지는 첫 지점을 **뒤에서 앞으로** 찾는다.

    ★ **기울기를 3탭 평활한 뒤** 문턱을 건다 — 안 하면 표본 잡음의 |diff| 가 QRS 기울기와
    같은 규모라 검출기가 탐색 경계에서 **포화**한다(픽스처가 잡았다).
    ⚠️ 검출기 파라미터가 있다 — **고정 창이 1차이고 이건 민감도**다.
    반환: (개시 index, **포화 여부**) — 포화가 많으면 적응형 팔을 쓰지 않는다."""
    lo, hi = ONSET_SEARCH
    d = np.abs(np.diff(B1, axis=-1))                      # (n, L-1)
    k = np.ones(3) / 3.0
    d = np.apply_along_axis(lambda v: np.convolve(v, k, mode="same"), 1, d)
    peak = d[:, QRS_LO - 1:QRS_HI].max(axis=1) + 1e-12
    thr = ONSET_FRAC * peak
    out = np.full(len(B1), hi, dtype=np.int64)
    sat = np.ones(len(B1), dtype=bool)
    for i in range(len(B1)):
        cnt = 0
        for j in range(hi - 1, lo - 1, -1):
            if d[i, j] < thr[i]:
                cnt += 1
                if cnt >= ONSET_RUN:
                    out[i] = j + ONSET_RUN; sat[i] = False
                    break
            else:
                cnt = 0
    return np.clip(out, lo, hi), sat

def dist(Bw, ref):
    dd = Bw - ref[None]
    return np.sqrt((dd * dd).sum(axis=(1, 2)))

def two_template_cv(Bw, tt, K, seed, n_rep):
    accs = []
    for rep in range(max(n_rep, 1)):
        rng = np.random.RandomState(seed + 1000 * rep)
        fold = rng.permutation(len(tt)) % K
        sc = np.full(len(tt), np.nan)
        for f in range(K):
            te = fold == f; tr = ~te
            if not te.any() or int(tt[tr].sum()) < 2 or int((~tt[tr]).sum()) < 2:
                return None
            medN = np.median(Bw[tr & ~tt], axis=0); medS = np.median(Bw[tr & tt], axis=0)
            sc[te] = dist(Bw[te], medN) - dist(Bw[te], medS)
        accs.append(sc)
    return np.mean(np.stack(accs), axis=0)

def cv_logit(X, tt, K, seed, n_rep):
    accs = []
    for rep in range(max(n_rep, 1)):
        rng = np.random.RandomState(seed + 1000 * rep)
        fold = rng.permutation(len(tt)) % K
        sc = np.full(len(tt), np.nan)
        for f in range(K):
            te = fold == f; tr = ~te
            if not te.any() or int(tt[tr].sum()) < 2 or int((~tt[tr]).sum()) < 2:
                return None
            mu = X[tr].mean(0); sd = X[tr].std(0) + 1e-9
            lr = LogisticRegression(max_iter=2000, C=1.0)
            lr.fit((X[tr] - mu) / sd, tt[tr].astype(int))
            sc[te] = lr.decision_function((X[te] - mu) / sd)
        accs.append(sc)
    return np.mean(np.stack(accs), axis=0)

RHY_COLS = ["f1", "f2_8", "f2_16", "f3", "f4", "f5", "f6"]   # ★ 전부 RR 파생(창 특징 0개)

def run_struct(t_):
    prev_ = np.r_[False, t_[:-1]]; next_ = np.r_[t_[1:], False]
    iso = t_ & ~prev_ & ~next_
    rl = mx = 0
    for v in t_:
        rl = rl + 1 if v else 0
        mx = max(mx, rl)
    return float(iso.sum() / max(t_.sum(), 1)), int(mx)

run.log("\n" + "=" * 100)
run.log("【Q-A】 특징 · 층 키 · QRS 개시 · 점수")
run.log("=" * 100)
run.log("  ▸ **마스크를 만들지 않는다**(R32 ① ③) — `f1` 정확 층화가 `pre` 를, 따라서")
run.log("    직전 T 위치를 이미 정확히 통제한다. Q7-P 의 SAFE 컷은 모집단을 바꿨다")
T0 = time.time()
FEAT, META, SKIP, ONS, SATF = {}, {}, [], {}, {}
for r in ALLR:
    mm = np.where(REC == r)[0]
    tt = (Y[mm] == IDX_S)
    if int(tt.sum()) < MIN_S_TPL or int((~tt).sum()) < MIN_S_TPL:
        SKIP.append((int(r), f"S {int(tt.sum())} · N {int((~tt).sum())}")); continue
    F = all_feats(PRE[mm], POST[mm])
    Bm = BEAT[mm]
    on, sat = qrs_onset(Bm[:, 0, :])
    amp = float(np.median(Bm[:, :, QRS_LO:QRS_HI].max(axis=2)
                          - Bm[:, :, QRS_LO:QRS_HI].min(axis=2)))
    iso_f, mx_run = run_struct(tt)
    FEAT[int(r)] = F; ONS[int(r)] = on; SATF[int(r)] = float(sat.mean())
    META[int(r)] = dict(tt=tt, idx=mm, key=strat_key(F), amp=amp,
                        pos=int(tt.sum()), prev=float(tt.mean()),
                        iso_frac=iso_f, max_run=mx_run, n=int(len(mm)))
RS = sorted(META)
run.log(f"  채점 **{len(RS)}개체** · 제외 {len(SKIP)}개체 · {time.time()-T0:.0f}초")
_on = np.concatenate([ONS[r] for r in RS])
run.log(f"  QRS 개시 실측(민감도) — 중앙 index **{np.median(_on):.0f}** "
        f"(= R{(np.median(_on)-RPRE)/360*1000:+.0f}ms) · IQR "
        f"[{np.percentile(_on,25):.0f}, {np.percentile(_on,75):.0f}]")
run.log(f"  ▸ `p_late_strict` 끝 index {SEGS22['p_late_strict'][1]} "
        f"(R{(SEGS22['p_late_strict'][1]-RPRE)/360*1000:+.0f}ms) 가 그보다 앞이면 "
        "**고정 창이 QRS 개시를 안 문다**")
_sat = float(np.mean([SATF[r] for r in RS]))
ONSET_OK = _sat < 0.5
run.log(f"  검출기 **포화율 {_sat:.3f}** (탐색 경계에서 못 찾은 비율)")
if not ONSET_OK:
    run.log("    ⛔ **포화가 절반을 넘는다 — 적응형 창(Q4)을 쓰지 않는다.** 잡음 대비")
    run.log("       QRS 기울기가 부족하다는 뜻이고, 그 사실 자체를 결과로 적는다")
CONFIG["qrs_onset_median"] = float(np.median(_on))
CONFIG["qrs_onset_saturation"] = _sat
run.save_json("config", CONFIG)
````

````python
# CELL 4 — 【Q-B】 점수 · 사다리 채점 (raw 앵커 포함)
ARMS_WIN = list(SEGS)
PURE = ["f1", "f2_16", "f3", "f3_glob", "f4", "f5", "f6"]
PROBES = [f"f2_{k}" for k in PROBE_K] + ["f1_rank"]
LR_ARMS = ["lr_rhy", "lr_f1"]
ARMS = PURE + PROBES + ARMS_WIN + LR_ARMS
LABEL_ARMS = ARMS_WIN + LR_ARMS

def win_of(r, name, inject=None):
    """개체 `r` 의 창 `name` 파형. `inject=(win, amp)` 면 **S 비트에** 합성 P 섭동을 얹는다."""
    a, b = SEGS[name]
    Bw = BEAT[META[r]["idx"]][:, :, a:b].astype("float64").copy()
    if inject is not None:
        iw, amp = inject
        if iw == name:                      # 주입 창에서만 파형이 바뀐다
            tt = META[r]["tt"]; L = b - a
            x = np.arange(L, dtype=float)
            bump = np.exp(-((x - (L - 1) / 2.0) ** 2) / (2 * (L / 6.0) ** 2))
            Bw[tt] += amp * META[r]["amp"] * bump[None, None, :]
    return Bw

def build_scores(r, tt, K, seed, inject=None):
    """라벨을 쓰는 팔(템플릿 · 로지스틱). **셔플 null 도 같은 함수.**"""
    F = FEAT[r]; S = {}
    X = np.stack([F[c] for c in RHY_COLS], axis=1)
    for nm_, XX in (("lr_rhy", X), ("lr_f1", X[:, [RHY_COLS.index("f1")]])):
        sc_ = cv_logit(XX, tt, K, seed, N_REPEAT)
        if sc_ is None:
            return None
        S[nm_] = sc_
    for nm_ in ARMS_WIN:
        st = two_template_cv(win_of(r, nm_, inject), tt, K, seed, N_REPEAT)
        S[nm_] = st if st is not None else np.full(len(tt), np.nan)
    return S

run.log("\n" + "=" * 100)
run.log("【Q-B】 사다리 채점 — " + " → ".join(LADDER) + "  (★ raw 앵커 · R32 ②)")
run.log("=" * 100)
BASES = ("lin", "rank", "quad", "hist")
MET = {m: {a: np.full(len(RS), np.nan) for a in ARMS} for m in LADDER}
SCORES = {}
T0 = time.time()
for i, r in enumerate(RS):
    tt = META[r]["tt"]; key = META[r]["key"]; F = FEAT[r]
    S = build_scores(r, tt, K_FOLD, SEED0)
    if S is None:
        continue
    SCORES[r] = S
    Z = {b_: basis_mat(F, b_) for b_ in BASES}
    for a in ARMS:
        s = S[a] if a in S else F[a]
        if not np.isfinite(s).all():
            continue
        MET["raw"][a][i] = roc_auc_score(tt.astype(int), s)
        for b_ in BASES:
            e = residualize(prep(s, b_), Z[b_])
            if e is not None:
                MET[b_][a][i] = roc_auc_score(tt.astype(int), e)
        MET["strat"][a][i] = strat_auc(s, tt, key)
run.log(f"  ({time.time()-T0:.0f}초) 팔별 — " + " · ".join(LADDER))
for a in ARMS:
    star = "  ★" if a in SEGS22 else ("  ▸" if a in ARMS_WIN else "")
    run.log(f"    {a:<14} " + " · ".join(f"{np.nanmean(MET[m][a]):.4f}" for m in LADDER) + star)
run.log("\n  ▸ 잔차 열의 **팔 간 수준 비교는 하지 않는다**(R31 ⑤). 주 지표는 `strat`,")
run.log(f"    공동 1차는 `{CO_PRIMARY}` 이고 효율 표를 §Q-E 에 병기한다(R32 ⑥)")
CONFIG["metrics"] = {m: {a: float(np.nanmean(MET[m][a])) for a in ARMS} for m in LADDER}
run.save_json("config", CONFIG)
````

````python
# CELL 5 — 【Q-C】 라벨셔플 null (전 팔 실측 · R26 ②)
run.log("\n" + "=" * 100)
run.log(f"【Q-C】 라벨셔플 null (셔플 {N_SHUF}회 · {len(RS)}개체 · 전 팔)")
run.log("=" * 100)
T1 = time.time()
NULL = {m: {a: np.full(len(RS), np.nan) for a in ARMS} for m in LADDER}
NSE  = {m: {a: np.full(len(RS), np.nan) for a in ARMS} for m in LADDER}
for i, r in enumerate(RS):
    if r not in SCORES:
        continue
    tt = META[r]["tt"]; key = META[r]["key"]; F = FEAT[r]
    Z = {b_: basis_mat(F, b_) for b_ in BASES}
    acc = {m: {a: [] for a in ARMS} for m in LADDER}
    for s_ in range(N_SHUF):
        rng = np.random.RandomState(SEED0 + 7919 * (s_ + 1) + int(r))
        ts = rng.permutation(tt)                       # 유병률 보존
        Ss = build_scores(r, ts, K_FOLD, SEED0 + 31 * (s_ + 1))
        if Ss is None:
            continue
        for a in ARMS:
            s = Ss[a] if a in Ss else F[a]
            if not np.isfinite(s).all():
                continue
            acc["raw"][a].append(roc_auc_score(ts.astype(int), s))
            for b_ in BASES:
                e = residualize(prep(s, b_), Z[b_])
                if e is not None:
                    acc[b_][a].append(roc_auc_score(ts.astype(int), e))
            acc["strat"][a].append(strat_auc(s, ts, key))
    for m in LADDER:
        for a in ARMS:
            v_ = np.asarray([x for x in acc[m][a] if np.isfinite(x)], float)
            if len(v_) >= 2:
                NULL[m][a][i] = float(v_.mean())
                NSE[m][a][i] = float(v_.std(ddof=1) / np.sqrt(len(v_)))
run.log(f"  ({time.time()-T1:.0f}초)  **{PRIMARY}** — 실측 vs null(±SE) vs 초과분")
for a in ARMS:
    m_ = np.nanmean(MET[PRIMARY][a]); n_ = np.nanmean(NULL[PRIMARY][a])
    se_ = np.nanmean(NSE[PRIMARY][a])
    star = "  ★" if a in SEGS22 else ("  ▸" if a in ARMS_WIN else "")
    run.log(f"    {a:<14} {m_:.4f}  null {n_:.4f} ±{se_:.4f}  **초과 {m_-n_:+.4f}**{star}")
CONFIG["null"] = {a: float(np.nanmean(NULL[PRIMARY][a])) for a in ARMS}
run.save_json("config", CONFIG)
````

````python
# CELL 6 — 【Q-D】 ★★ 양성 대조 — **검출 바닥을 실측한다**(Q2)
# 이게 없으면 Q1 의 음성 결과가 「QRS 개시였다」인지 「폭을 줄여 바닥 아래로 갔다」인지
# 구분되지 않는다. **Q1 과 완전히 같은 추정량·같은 폭·같은 null** 로 잰다.
#
# ⚠️ null 을 다시 안 잰다: 라벨을 섞으면 주입 파형이 섞인 라벨과 무관해져 `medS`·`medN`
#    양쪽에 고루 섞인다 → 영분포는 기저와 같다. (셔플을 3배로 늘리지 않는 근거)
run.log("\n" + "=" * 100)
run.log(f"【Q-D】 ★★ 양성 대조 — 진폭 {AMPS} × 개체 중앙 R 진폭")
run.log("=" * 100)
run.log("  ⚠️ **이건 검출력의 상한이다** — 모든 S 비트에 **동일 파형**을 얹으므로 두")
run.log("     템플릿 추정량에 가장 유리하다. 여기서 안 잡히는 크기는 실데이터에서 못 잡는다")
T2 = time.time()
INJ = {w: {a_: np.full(len(RS), np.nan) for a_ in AMPS} for w in INJ_WINS}
for i, r in enumerate(RS):
    if r not in SCORES:
        continue
    tt = META[r]["tt"]; key = META[r]["key"]
    for w in INJ_WINS:
        for a_ in AMPS:
            st = two_template_cv(win_of(r, w, inject=(w, a_)), tt, K_FOLD, SEED0, N_REPEAT)
            if st is not None:
                INJ[w][a_][i] = strat_auc(st, tt, key)
run.log(f"  ({time.time()-T2:.0f}초)")
CONFIG["inject"] = {w: {str(a_): float(np.nanmean(INJ[w][a_])) for a_ in AMPS}
                    for w in INJ_WINS}
run.save_json("config", CONFIG)
````

````python
# CELL 7 — 【Q-E】 관문 Q1 · Q1b · Q1c + 검출 바닥 + 누출 바닥(양수) + 효율 표
def boot_pair(a1, a2, meth, seed, nb=NB_BOOT, q=2.5, arr=None):
    """짝의 **초과분 차이** 개체 부트스트랩 + null 오차 전파(R26 ②)."""
    m1 = MET[meth][a1] if arr is None else arr
    d = (m1 - NULL[meth][a1]) - (MET[meth][a2] - NULL[meth][a2])
    se = np.sqrt(np.nan_to_num(NSE[meth][a1]) ** 2 + np.nan_to_num(NSE[meth][a2]) ** 2)
    ok = np.isfinite(d); d = d[ok]; se = se[ok]
    if len(d) < 3:
        return float("nan"), float("nan"), float("nan"), 0
    rng = np.random.RandomState(seed)
    v = np.empty(nb)
    for b in range(nb):
        ix = rng.randint(0, len(d), len(d))
        v[b] = (d[ix] - rng.normal(0.0, 1.0, len(ix)) * se[ix]).mean()
    return float(d.mean()), float(np.percentile(v, q)), float(np.percentile(v, 100 - q)), len(d)

def boot_excess(a, meth, seed, nb=NB_BOOT, q=2.5):
    d = MET[meth][a] - NULL[meth][a]; se = np.nan_to_num(NSE[meth][a])
    ok = np.isfinite(d); d = d[ok]; se = se[ok]
    if len(d) < 3:
        return float("nan"), float("nan"), float("nan"), 0
    rng = np.random.RandomState(seed)
    v = np.empty(nb)
    for b in range(nb):
        ix = rng.randint(0, len(d), len(d))
        v[b] = (d[ix] - rng.normal(0.0, 1.0, len(ix)) * se[ix]).mean()
    return float(d.mean()), float(np.percentile(v, q)), float(np.percentile(v, 100 - q)), len(d)

run.log("\n" + "=" * 100)
run.log(f"【Q-E】 관문 (주 {PRIMARY} · 공동 1차 {CO_PRIMARY} · {len(RS)}개체)")
run.log("=" * 100)
VERD, DIFF = {}, {}
def g_(k, v, d):
    VERD[k] = v; run.log(f"  {k:<4}{v}  {d}")

for gname, pa, pb in GATE_PAIRS:
    wa = SEGS[pa][1] - SEGS[pa][0]; wb = SEGS[pb][1] - SEGS[pb][0]
    if wa != wb:
        raise AssetError(f"폭 불일치 {pa}({wa}) vs {pb}({wb}) — R27 ③ 위반")
    m_, lo_, hi_, n_ = boot_pair(pa, pb, PRIMARY, SEED0 + 11, q=BONF3 * 100)
    if n_ < 3:
        DIFF[gname] = dict(mean=m_, lo=lo_, hi=hi_, n=n_, pair=[pa, pb], width=wa)
        g_(gname, "⛔ 측정 불가", f"{pa}−{pb} — 개체 {n_}"); continue
    eq, sup, nn, frame = judge(m_, lo_, hi_, n_, EQUIV_MARGIN)
    lad = {}
    for meth in LADDER:
        lad[meth] = boot_pair(pa, pb, meth, SEED0 + 13)[0]
    cm, clo, chi, _ = boot_pair(pa, pb, CO_PRIMARY, SEED0 + 14, q=BONF3 * 100)
    DIFF[gname] = dict(mean=m_, lo=lo_, hi=hi_, n=n_, pair=[pa, pb], width=wa,
                       equiv=eq, sup=sup, need_n=(None if nn is None else float(nn)),
                       ladder={k: float(v) for k, v in lad.items()},
                       co=dict(mean=cm, lo=clo, hi=chi))
    g_(gname, sup, f"★ **폭 {wa}** ({pa} 초과) − ({pb} 초과) **{m_:+.4f}** "
                   f"[{lo_:+.4f}, {hi_:+.4f}] · Bonf 3 · n={n_}")
    run.log(f"       등가 {eq} · {frame}")
    run.log(f"       [공동 1차 {CO_PRIMARY}] {cm:+.4f} [{clo:+.4f}, {chi:+.4f}] "
            f"· 우월성 {decide(clo, chi, 0.0, '>')}")
    run.log("       사다리 " + " · ".join(f"{k} {v:+.4f}" for k, v in lad.items()))
    run.log(f"       {ladder_read(lad)}")

# ── ★★ Q2 검출 바닥 — 관문 해석의 분모
run.log("\n  ★★ **Q2 검출 바닥** (회수 = 주입 후 초과 − 기저 초과 · 같은 null)")
FLOOR = {}
for w in INJ_WINS:
    base_m = boot_excess(w, PRIMARY, SEED0 + 20)[0]
    row = []
    hit = None
    for a_ in AMPS:
        rm, rlo, rhi, rn = boot_pair(w, NEG22, PRIMARY, SEED0 + 21, arr=INJ[w][a_])
        bm = DIFF.get("Q1", {}).get("mean")
        gm, glo, ghi, _ = boot_pair(w, NEG22, PRIMARY, SEED0 + 22)
        rec = rm - gm                                  # 회수량(짝 기준)
        ok = "✅" if rlo > ghi else ("▸" if rm > gm else "❌")
        row.append(f"a={a_:.2f} 회수 {rec:+.4f} (주입 {rm:+.4f} vs 기저 {gm:+.4f}) {ok}")
        if hit is None and rlo > 0:
            hit = a_
    FLOOR[w] = hit
    run.log(f"    {w:<14} " + " | ".join(row))
    run.log(f"      → **검출 바닥 = {'a ≥ ' + format(hit, '.2f') if hit else '⛔ 20%에서도 미검출'}**")
CONFIG["detect_floor"] = {w: (None if v is None else float(v)) for w, v in FLOOR.items()}

# ── ★ 누출 바닥은 **양의 초과분 최댓값** · 음의 초과는 따로 (R32 ④)
run.log("\n  누출 바닥 (★ **양의 초과분 최댓값** · 음의 초과는 과잉보정 진단 · R32 ④)")
LEAK_ARMS = PROBES + ["f4", "f5", "f3", "f3_glob"]
pos, neg = {}, {}
for a in LEAK_ARMS:
    mm_, ll_, hh_, _ = boot_excess(a, PRIMARY, SEED0 + 15)
    (pos if mm_ > 0 else neg)[a] = (mm_, ll_, hh_)
    run.log(f"    {a:<9} 초과 {mm_:+.4f} [{ll_:+.4f}, {hh_:+.4f}]")
LEAK_MAX = max((v[0] for v in pos.values()), default=0.0)
LEAK_ARM = max(pos, key=lambda k: pos[k][0]) if pos else "(없음)"
run.log(f"    ★ **누출 바닥 = {LEAK_MAX:.4f}** (`{LEAK_ARM}`) — **수준 주장에만** 뺀다(R32 ⑤)")
if neg:
    run.log("    ⚠️ **과잉보정·부호반전 진단**(바닥 아님) — " +
            " · ".join(f"{k} {v[0]:+.4f}" for k, v in neg.items()))
run.log("    ▸ 짝의 차이(Q1·Q1b·Q1c)에서는 공통 누출이 상쇄되므로 **빼지 않는다**(R32 ⑤)")
CONFIG["leak_max"] = float(LEAK_MAX); CONFIG["leak_arm"] = LEAK_ARM

# ── ★ 효율 표 (R32 ⑥ — 편향을 줄이려 분산을 얼마나 샀나)
run.log(f"\n  효율 표 — `{PRIMARY}` vs `{CO_PRIMARY}` CI 폭 (R32 ⑥ · AIPW 전 단계)")
for gname, pa, pb in GATE_PAIRS:
    d_ = DIFF.get(gname, {})
    if "co" not in d_:
        continue
    wp = d_["hi"] - d_["lo"]; wc = d_["co"]["hi"] - d_["co"]["lo"]
    run.log(f"    {gname:<4} {PRIMARY} 폭 {wp:.4f} · {CO_PRIMARY} 폭 {wc:.4f} "
            f"→ **{wp/max(wc,1e-9):.2f}배**  (부호 "
            f"{'일치' if d_['mean']*d_['co']['mean'] >= 0 else '**불일치**'})")
CONFIG["gates"] = VERD; CONFIG["diffs"] = DIFF
run.save_json("config", CONFIG)
````

````python
# CELL 8 — 【Q-F】 Q3 추정 대상 교체 시연 · Q4 적응형 창 · Q5 층 분해 (관문 아님)
run.log("\n" + "=" * 100)
run.log("【Q-F】 Q3 ΔAUROC · Q4 적응형 창 · Q5 층 분해 (관문 아님)")
run.log("=" * 100)

# ── Q3 ★ 추정 대상 교체 — 리듬 전용 대비 **창 특징 추가**의 ΔAUROC
# ⚠️ **개체 내**다. 교차환자가 아니다(그건 Q7-O/Q7-R). 그렇게 정확히 적는다.
# ⚠️ Q7-P 의 `lr_all`−`lr_norr` 는 **둘 다 창 특징이 0개**라 이 물음의 근거가 아니었다.
run.log("  Q3 — 리듬 전용(`lr_rhy`, RR 파생 7개) + **창 점수** 추가의 ΔAUROC")
run.log("     ⚠️ **개체 내 교차적합**이다 — 교차환자 일반화를 말하지 않는다")
D3 = {w: np.full(len(RS), np.nan) for w in SEGS22}
for i, r in enumerate(RS):
    if r not in SCORES:
        continue
    tt = META[r]["tt"]; F = FEAT[r]
    Xr = np.stack([F[c] for c in RHY_COLS], axis=1)
    base = cv_logit(Xr, tt, K_FOLD, SEED0 + 5, N_REPEAT)
    if base is None:
        continue
    b_auc = roc_auc_score(tt.astype(int), base)
    for w in SEGS22:
        sc = SCORES[r].get(w)
        if sc is None or not np.isfinite(sc).all():
            continue
        Xw = np.c_[Xr, sc]
        full = cv_logit(Xw, tt, K_FOLD, SEED0 + 5, N_REPEAT)
        if full is not None:
            D3[w][i] = roc_auc_score(tt.astype(int), full) - b_auc
for w in SEGS22:
    d = D3[w][np.isfinite(D3[w])]
    if len(d) < 3:
        run.log(f"    {w:<14} 개체 부족"); continue
    rng = np.random.RandomState(SEED0 + 31)
    v = [d[rng.randint(0, len(d), len(d))].mean() for _ in range(1500)]
    run.log(f"    {w:<14} ΔAUROC **{d.mean():+.4f}** "
            f"[{np.percentile(v,2.5):+.4f}, {np.percentile(v,97.5):+.4f}] · n={len(d)}")
CONFIG["delta_auroc"] = {w: float(np.nanmean(D3[w])) for w in SEGS22}

# ── Q4 (민감도) 비트별 QRS 개시 적응형 창 vs 고정 창
run.log("\n  Q4 (민감도) 적응형 창 [개시−22, 개시) vs 고정 `p_late_strict`")
ADAP = np.full(len(RS), np.nan)
for i, r in enumerate(RS) if ONSET_OK else []:
    if r not in SCORES:
        continue
    tt = META[r]["tt"]; on = ONS[r]; Bm = BEAT[META[r]["idx"]]
    Bw = np.stack([Bm[j, :, max(on[j] - W22, 0):max(on[j], W22)] for j in range(len(on))])
    if Bw.shape[-1] != W22:
        continue
    st = two_template_cv(Bw.astype("float64"), tt, K_FOLD, SEED0, N_REPEAT)
    if st is not None:
        ADAP[i] = strat_auc(st, tt, META[r]["key"])
fx = np.nanmean(MET[PRIMARY]["p_late_strict"])
if ONSET_OK:
    run.log(f"    고정 {fx:.4f}  vs  적응형 {np.nanmean(ADAP):.4f}  "
            f"차이 {np.nanmean(ADAP)-fx:+.4f}   ← **고정 창이 1차**, 이건 민감도")
else:
    run.log(f"    ⛔ 검출기 포화로 **미실행**. 고정 {fx:.4f} 만 쓴다(R29 ②)")
CONFIG["adaptive"] = (float(np.nanmean(ADAP)) if ONSET_OK and np.isfinite(ADAP).any()
                      else None)

# ── Q5 층 분해 — ★ 폭 정합 짝으로만 (R31 ④)
run.log("\n  Q5 층 분해 — ★ **폭 22 짝으로만**(R31 ④)")
ISOF = np.array([META[r]["iso_frac"] for r in RS])
for nm_, msk in (("고립 S", ISOF >= 0.7), ("혼합", (ISOF < 0.7) & (ISOF > 0.3)),
                 ("런 우세", ISOF <= 0.3)):
    n_ = int(msk.sum())
    if n_ < 3:
        run.log(f"    {nm_:<8} {n_}개체 — 3 미만이라 CI 를 내지 않는다(R17)"); continue
    parts = []
    for gname, pa, pb in GATE_PAIRS:
        d = ((MET[PRIMARY][pa] - NULL[PRIMARY][pa])
             - (MET[PRIMARY][pb] - NULL[PRIMARY][pb]))[msk]
        d = d[np.isfinite(d)]
        if len(d) < 3:
            parts.append(f"{gname} n/a"); continue
        rng = np.random.RandomState(SEED0 + 22)
        v = [d[rng.randint(0, len(d), len(d))].mean() for _ in range(800)]
        parts.append(f"{gname} {d.mean():+.4f} "
                     f"[{np.percentile(v,2.5):+.4f},{np.percentile(v,97.5):+.4f}]")
        CONFIG.setdefault("layers", {}).setdefault(nm_, {})[gname] = float(d.mean())
    run.log(f"    {nm_:<8} {n_:>2}개체 · " + " · ".join(parts))
run.save_json("config", CONFIG)
````
