### 가설 두 개 — 서로 **다른 예측**을 한다

**(A) 오염 끌림.** 좌표별 중앙값은 스칼라 중앙값처럼 강건하지 않다. 유병률 π 의
혼합에서 각 좌표의 중앙값은 N 분포의 **(0.5−π)/(1−π) 분위**로 끌려간다. π=0.28 이면
N 의 **31 분위**다. 좌표 170개가 **같은 방향으로** 끌리면 누적된다.
→ 예측: **끌림 `‖med(전체) − med(N)‖ / σ_N` 이 (오라클 − 다수결) 격차를 만든다.**
→ 이건 **오라클로 고쳐진다.**

**(B) 산포 비대칭.** `‖b − ref‖` 는 「기저와 다름」이 아니라 **「비전형성」**을 잰다.
기저 리듬이 잡다하고(`σ_N` 큼) 이소성이 정형적이면(`σ_S` 작음) N 이 더 멀다.
→ 예측: **`σ_N/σ_S` 가 클수록 오라클 AUROC 가 낮다.**
→ 이건 **오라클로 안 고쳐진다.** 기준을 어디에 놓든 마찬가지다.

`#865` 에서 이미 무늬를 봤다 — S 는 CV 0.063(규칙적), N 은 CV 0.224(불규칙).

## 처방 — **두 템플릿 점수**

단일 기준까지의 거리 대신 **두 방향의 차**를 쓴다.

```
g(b) = ‖b − N템플릿‖ − ‖b − S템플릿‖          (크면 S)
```

이게 왜 두 가설 **모두**에 면역인지는 산수로 나온다. `sep = ‖medS − medN‖` 라 하면

| 비트 | `‖b−N템‖` | `‖b−S템‖` | `g` |
|---|---|---|---|
| N | ≈ σ_N | ≈ √(σ_N² + sep²) | **음수** |
| S | ≈ √(σ_S² + sep²) | ≈ σ_S | **양수** |

**σ_N 이 아무리 커도 부호가 안 바뀐다** — (B) 면역. 그리고 템플릿을 혼합 중앙이
아니라 각 군에서 뽑으므로 **(A) 도 면역**이다.

> ⚠️ **대가: S 템플릿은 라벨을 쓴다.** 그래서 이 점수는 **성능 주장이 아니라 상한**
> 이다 — Q7-E 의 오라클과 같은 지위다. 배포판은 템플릿을 **다른 환자들**에서 만드는
> Q7-G 다. 그 전에 **상한이 존재하는지**부터 확인하는 게 이 실험이다.
> 무감독 근사(짧은RR 군 = S̃ · 긴RR 군 = Ñ)도 같이 재지만 **Q7-E 결과로 보아 실패가
> 기본 예측**이다 — 그래도 사전등록한다(H6).

## 사전등록

| 관문 | 내용 | 문턱 |
|---|---|---|
| **H1** | **두 템플릿이 단일 기준을 이긴다** — 짝지은 (두템플릿 − 오라클단일) | CI 하한 **> 0** |
| **H2** | **반전 0** — 두 템플릿에서 유의하게 뒤집힌 개체 (개체별 SE + 본페로니) | **= 0** |
| **H3** | **RR 위치형 대비 비열등** — 짝지은 (두템플릿 − RR위치형) | CI 하한 **> −0.05** |
| **H4** | **(A) 오염 끌림** — `rho(끌림, 오라클−다수결)` | CI 하한 **> 0** |
| **H5** | **(B) 산포 비대칭** — `rho(σ_N/σ_S, 오라클 AUROC)` | CI 상한 **< 0** |
| **H6** | **무감독판이 쓸모 있나** — 짝지은 (무감독 두템플릿 − 다수결단일) | CI 하한 **> 0** |

**H1·H2 가 「상한이 존재하는가」**, **H3 이 「형태가 리듬을 따라잡는가」**(Q7-E 에서
−0.1148 로 졌던 그 비교), **H4·H5 가 기전**, **H6 이 배포 가능성**이다.

### H5 는 부분적으로 **구조적**이다 — 부분상관을 병기한다

오라클 점수는 `‖b − med(N)‖` 이므로 σ_N 이 분모처럼 들어간다. σ_N 이 크면 오라클이
낮은 건 어느 정도 정의상 그렇다. 그래서 **`SNR = sep/σ_N` 을 통제한 부분상관**을
같이 낸다. SNR 을 고정하고도 `σ_N/σ_S` 가 남으면 그때 (B) 가 독립적으로 성립한다.

### 픽스처가 먼저 잡은 것 — **템플릿 인-샘플 누수**

처음 구현은 레코드 전체 비트로 `medS`·`medN` 을 만들고 **그 비트들을 그대로 채점**했다.
합성 **null 코호트**(S 와 N 의 P 파를 아예 같게 만든 것)를 돌렸더니 두 템플릿 AUROC 가
**0.8916** 이었다. **형태 신호가 0인데도.**

원인은 단순하다. S 템플릿이 채점 대상 S 비트들의 중앙값이라 **S 는 자기 템플릿에
가까울 수밖에 없다.** S 개수가 적을수록 심한데, SVDB 유병률 중앙이 **0.035** 라 정확히
그 조건이다. 그대로 돌렸으면 **가짜 상한**을 보고 "형태 축이 리듬을 이겼다" 고 썼을 것이다.

→ 모든 템플릿을 **`K_FOLD`겹 교차적합**한다. 단일 기준(`orc1`)도 같은 문제를 약하게
갖는다(같은 null 에서 **0.545**) — 공정하게 같이 교차적합한다. 산포 `σ_N`·`σ_S` 도
**겹 밖 거리**로 잰다(인-샘플이면 과소추정된다).

⚠️ 이 누수는 **Q7-E 의 오라클 값에도 있었다**(`orc1` 과 같은 구성). 크기는 null 에서
약 +0.045 이고 SVDB 는 N 이 훨씬 많아 더 작을 것이다 — 결론(오라클≈다수결)을 바꿀
크기는 아니지만, Q7-E 오라클을 **정확한 상한으로 인용하지 않는다**.

### 그리고 두 번째로 잡은 것 — **겹 배정 잡음이 SE 에서 빠져 있었다**

교차적합을 넣었더니 null 매크로는 0.505 로 내려왔는데, **개체 하나(`#907`, 0.3847)가
여전히 '유의 반전' 으로 찍혔다.** 신호 0인 코호트에서.

원인: 겹 배정이 무작위라 점수 자체에 잡음이 얹히는데, **비트 부트스트랩은 점수를
고정으로 보므로 그 몫을 못 센다.** SE 가 과소추정되면 본페로니를 걸어도 샌다.
→ 교차적합을 `N_REPEAT`회 되풀이하고 **`SE = √(부트스트랩SE² + 되풀이간 분산)`** 으로
합친다. 점수는 되풀이 평균을 쓴다.

### 제외 · 발동 개체 수를 먼저 센다 (R21 ③-b)

- **S < `MIN_S_TPL`(=20)** 이면 교차적합 훈련 겹의 S 가 16개 밑으로 떨어진다 → 제외하고
  **이름을 남긴다**. Q7-E 의 70개체 중 S≥20 은 **58개** → 그 정도가 예상 코호트다.
- Q7-E 는 처방이 **11/70 에서만 발동**했는데 매크로만 보면 「차이 없음」으로 읽혔다.
  여기서는 `|두템플릿 − 오라클단일| > 0.01` 인 개체 수를 **관문보다 먼저** 출력한다.

````python
# CELL 4 — 【H-B】 ★ 전수 채점 — 템플릿은 전부 **K겹 교차적합**
#
#   점수 5종 (전부 P영역, 방향은 「크면 S」):
#     maj      ‖b − med(전체)‖              단일 · 무감독 · 현행 `_medref`
#     orc1     ‖b − med(N)‖                 단일 · **라벨** = Q7-E 의 오라클
#     two_orc  ‖b−med(N)‖ − ‖b−med(S)‖      ★ 두 템플릿 · **라벨** = 이 실험의 상한
#     two_uns  ‖b−med(Ñ)‖ − ‖b−med(S̃)‖      두 템플릿 · **무감독**(긴RR군=Ñ, 짧은RR군=S̃)
#     rr       med(pre) − pre               위치형 · 기준 불변 대조군 (교차적합 불필요 —
#                                            상수 이동이라 순위가 안 바뀐다)
#
#   ★★ **교차적합이 이 셀의 핵심이다.** 템플릿을 채점 대상 비트로 만들면 안 된다.
#      픽스처 실측: 형태 신호가 **0인** 합성 코호트에서 인-샘플 두 템플릿이 **0.8916**.
#      S 가 적을수록 심하다 — SVDB 유병률 중앙이 0.035 라 정확히 그 조건이다.
from sklearn.metrics import roc_auc_score
from sklearn.cluster import KMeans

def dist(B, ref):
    d = B - ref[None]
    return np.sqrt((d * d).sum(axis=(1, 2)))

def boot_auroc(tt, sc, seed, nb):
    """레코드 **내부**(비트) 부트스트랩 → (lo, hi, SE). 관문은 본페로니 보정한 SE 를 쓴다(R20)."""
    rng = np.random.RandomState(seed)
    v = []
    for _ in range(nb):
        j = rng.randint(0, len(tt), len(tt)); tj = tt[j]
        if 0 < tj.sum() < len(tj):
            v.append(roc_auc_score(tj.astype(int), sc[j]))
    if len(v) < 20:
        return float("nan"), float("nan"), float("nan")
    return (float(np.percentile(v, 2.5)), float(np.percentile(v, 97.5)),
            float(np.std(v, ddof=1)))

def unsup_templates(Btr, pre_tr, Bte, seed, min_clus):
    """무감독 2군 템플릿 — **훈련 겹에서만** 만든다. 라벨을 인자로 받지 않는다.

    쓰는 사전지식은 「이소성은 이르다」 하나(Q7-E 와 같은 리듬 사전지식).
    Q7-E 에서 이게 84% 개체에서 다수결로 퇴화한다는 걸 이미 봤다 — 그래서 H6 의
    기본 예측은 실패다. 그래도 사전등록해서 잰다.
    """
    km = KMeans(2, n_init=10, random_state=seed).fit(Btr.reshape(len(Btr), -1))
    lab = km.labels_
    n0, n1 = int((lab == 0).sum()), int((lab == 1).sum())
    if min(n0, n1) < min_clus:
        return None
    m0, m1 = float(np.median(pre_tr[lab == 0])), float(np.median(pre_tr[lab == 1]))
    if m0 == m1:
        return None
    long_ = 0 if m0 > m1 else 1
    medL = np.median(Btr[lab == long_], axis=0)
    medS_ = np.median(Btr[lab == 1 - long_], axis=0)
    return dist(Bte, medL) - dist(Bte, medS_), float((lab == 1 - long_).mean()), n0, n1

def _one_pass(B, tt, pre_m, K, seed, min_clus, do_unsup):
    """K겹 교차적합 1회. 반환: (점수 dict, 겹밖거리, 통계) 또는 (None, 사유, None)."""
    rng = np.random.RandomState(seed)
    fold = rng.permutation(len(tt)) % K
    SC = {k: np.full(len(tt), np.nan) for k in ("maj", "orc1", "two_orc", "two_uns")}
    dN_oof = np.full(len(tt), np.nan); dS_oof = np.full(len(tt), np.nan)
    seps, drags, short_frac, sizes = [], [], [], []
    uns_ok = do_unsup
    for f in range(K):
        te = fold == f; tr = ~te
        if not te.any():
            continue
        if int(tt[tr].sum()) < 2 or int((~tt[tr]).sum()) < 2:
            return None, f"{K}겹 중 한 겹의 훈련쪽 클래스가 2개 미만", None
        medA = np.median(B[tr], axis=0)
        medN = np.median(B[tr & ~tt], axis=0)
        medS = np.median(B[tr & tt], axis=0)
        dN, dS = dist(B[te], medN), dist(B[te], medS)
        SC["maj"][te] = dist(B[te], medA)
        SC["orc1"][te] = dN
        SC["two_orc"][te] = dN - dS
        dN_oof[te], dS_oof[te] = dN, dS
        seps.append(float(np.sqrt(((medS - medN) ** 2).sum())))
        drags.append(float(np.sqrt(((medA - medN) ** 2).sum())))
        if uns_ok:
            u = unsup_templates(B[tr], pre_m[tr], B[te], seed, min_clus)
            if u is None:
                uns_ok = False
            else:
                SC["two_uns"][te], sf, n0, n1 = u
                short_frac.append(sf); sizes.append((n0, n1))
    if not uns_ok:
        SC["two_uns"][:] = np.nan
    # ── 산포는 **겹 밖(out-of-fold)** 거리로 잰다 — 인-샘플이면 과소추정된다
    sig_n = float(np.median(dN_oof[~tt])); sig_s = float(np.median(dS_oof[tt]))
    sep = float(np.mean(seps)); drag_raw = float(np.mean(drags))
    ST = dict(sigma_n=sig_n, sigma_s=sig_s,
              sig_ratio=float(sig_n / max(sig_s, 1e-9)),
              sep=sep, snr=float(sep / max(sig_n, 1e-9)),
              drag=float(drag_raw / max(sig_n, 1e-9)),
              uns_degenerate=bool(do_unsup and not uns_ok))
    if sizes:
        ST["n0"], ST["n1"] = int(np.mean([a for a, _ in sizes])), int(np.mean([b for _, b in sizes]))
        ST["s_frac_short"] = float(np.mean(short_frac))
    return SC, None, ST


def score_record(B, tt, pre_m, K, seed, min_clus, n_rep):
    """**되풀이** K겹 교차적합. 점수는 되풀이 평균, 되풀이간 AUROC 분산을 함께 돌려준다.

    ★ 겹 배정이 무작위라 점수에 잡음이 얹힌다. 비트 부트스트랩 SE 는 점수를 고정으로
      보므로 그 몫을 못 센다 — 실제로 null 코호트에서 가짜 '유의 반전' 이 떴다.
      → SE 는 `√(부트스트랩SE² + 되풀이간 분산)` 으로 합친다(아래 채점 루프).
    """
    acc = {k: [] for k in ("maj", "orc1", "two_orc")}
    au = {k: [] for k in ("maj", "orc1", "two_orc")}
    ST0, uns = None, None
    for rep in range(max(n_rep, 1)):
        SC, why, ST = _one_pass(B, tt, pre_m, K, seed + 1000 * rep, min_clus, rep == 0)
        if SC is None:
            return None, why, None
        if rep == 0:
            ST0, uns = ST, SC["two_uns"]
        for k in acc:
            acc[k].append(SC[k])
            au[k].append(roc_auc_score(tt.astype(int), SC[k]))
    OUT = {k: np.mean(np.stack(v), axis=0) for k, v in acc.items()}
    OUT["two_uns"] = uns
    OUT["rr"] = np.median(pre_m) - pre_m
    REPSD = {k: (float(np.std(v, ddof=1)) if len(v) > 1 else 0.0) for k, v in au.items()}
    REPSD["two_uns"] = 0.0; REPSD["rr"] = 0.0
    ST0["rep_sd"] = REPSD
    return OUT, ST0, REPSD

run.log("\n" + "=" * 100)
run.log("【H-B】 전수 채점 (템플릿 {}겹 교차적합 × {}회 되풀이)".format(K_FOLD, N_REPEAT))
run.log("=" * 100)
PER, SKIP = {}, []
for r in ALLR:
    mm = np.where(REC == r)[0]
    tt = (Y[mm] == IDX_S)
    ns, nn = int(tt.sum()), int((~tt).sum())
    if len(mm) < MIN_BEATS:
        SKIP.append((int(r), f"비트 {len(mm)} < {MIN_BEATS}")); continue
    if ns < MIN_S_TPL:
        SKIP.append((int(r), f"S {ns} < {MIN_S_TPL} — S 템플릿을 교차적합할 수 없다")); continue
    if nn < MIN_S_TPL:
        SKIP.append((int(r), f"N {nn} < {MIN_S_TPL} — N 템플릿을 교차적합할 수 없다")); continue
    B, pre_m = BP[mm], PRE[mm]
    SC, ST, REPSD = score_record(B, tt, pre_m, K_FOLD, SEED0, MIN_CLUS, N_REPEAT)
    if SC is None:
        SKIP.append((int(r), ST)); continue
    row = dict(n=int(len(mm)), pos=ns, prev=float(tt.mean()))
    row.update(ST)
    for k_, sc_ in SC.items():
        if not np.isfinite(sc_).all():
            row[k_] = float("nan"); continue
        row[k_] = float(roc_auc_score(tt.astype(int), sc_))
        if k_ in ("maj", "orc1", "two_orc", "two_uns"):
            lo_, hi_, se_ = boot_auroc(tt, sc_, SEED0 + int(r), NB_REC)
            row[k_ + "_lo"], row[k_ + "_hi"] = lo_, hi_
            row[k_ + "_se_boot"] = se_
            # ★ 겹 배정 잡음을 더한다 — 부트스트랩만 쓰면 null 에서 가짜 반전이 뜬다
            row[k_ + "_se"] = (float(np.sqrt(se_ ** 2 + REPSD.get(k_, 0.0) ** 2))
                               if np.isfinite(se_) else float("nan"))
    # ── Q7-E′ 재료(관문 아님): 라벨 없는 이소성 부담 대리치
    row["burden_proxy"] = float((pre_m < BURDEN_Q * np.median(pre_m)).mean())
    PER[int(r)] = row

run.log(f"  채점 {len(PER)}개체 · **제외 {len(SKIP)}개체** (조용히 빼지 않는다 — 전부 아래에)")
for r, why in SKIP:
    run.log(f"    #{r}  {why}")
if len(PER) < 10:
    raise AssetError("채점된 개체가 너무 적다 — 자산·문턱을 확인할 것")

RS = sorted(PER)
A = {k: np.array([PER[r].get(k, np.nan) for r in RS], float) for k in
     ("maj", "orc1", "two_orc", "two_uns", "rr", "prev", "pos",
      "sigma_n", "sigma_s", "sig_ratio", "sep", "snr", "drag", "burden_proxy")}

# ── ★ R21 ③-b: **처방이 몇 개체에서 실제로 발동했는지를 관문보다 먼저 센다.**
#     Q7-E 는 84% 개체에서 앵커가 다수결과 같은 물건이었는데 매크로만 보면 안 보였다.
act = np.abs(A["two_orc"] - A["orc1"]) > ACT_EPS
run.log(f"\n  ★ 처방 발동 — |두템플릿 − 오라클단일| > {ACT_EPS} 인 개체 "
        f"**{int(act.sum())}/{len(RS)}** ({act.sum()/len(RS)*100:.0f}%)")
run.log(f"    발동 개체 Δ 평균 {np.nanmean((A['two_orc']-A['orc1'])[act]):+.4f}"
        f" · 미발동 {np.nanmean((A['two_orc']-A['orc1'])[~act]):+.4f}")

run.log(f"\n  {'점수':<34}{'매크로':>9}{'SD':>9}{'중앙':>9}{'최소':>9}{'<0.5':>7}")
NM = {"rr": "RR·위치형 (기준 불변)", "maj": "단일·다수결 (_medref, 무감독)",
      "orc1": "단일·오라클 ‖b−medN‖ (라벨)", "two_orc": "★두 템플릿 (라벨=상한)",
      "two_uns": "두 템플릿·무감독 (긴RR/짧은RR)"}
for k in ("rr", "maj", "orc1", "two_orc", "two_uns"):
    v = A[k]; f = np.isfinite(v)
    run.log(f"  {NM[k]:<34}{np.nanmean(v):>9.4f}{np.nanstd(v[f], ddof=1):>9.4f}"
            f"{np.nanmedian(v):>9.4f}{np.nanmin(v):>9.4f}{int((v[f] < 0.5).sum()):>7}")
run.log("\n  ⚠️ `orc1`·`two_orc` 는 **라벨을 쓴다** — 성능이 아니라 **상한**이다.")
run.log("     그리고 이 표는 무감독/오라클 거리 특징이다. 학습 모델 0.8842 와 비교 금지.")
run.log(f"     템플릿은 전부 {K_FOLD}겹 교차적합 × {N_REPEAT}회 — 인-샘플이면 신호 0에서도 0.89 다")
CONFIG["per_record"] = {str(r): PER[r] for r in RS}
CONFIG["skipped"] = [{"rec": r, "why": w} for r, w in SKIP]
CONFIG["activation"] = dict(eps=ACT_EPS, n_active=int(act.sum()), n=len(RS))
run.save_json("config", CONFIG)
````

````python
# CELL 5 — 【H-C】 관문 H1·H2·H3 — 상한이 존재하는가
def boot_diff(a, b, seed, nb=NB_BOOT):
    d = np.asarray(a, float) - np.asarray(b, float)
    d = d[np.isfinite(d)]
    if len(d) < 3:
        return float("nan"), float("nan"), float("nan"), 0
    rng = np.random.RandomState(seed)
    v = np.array([d[rng.randint(0, len(d), len(d))].mean() for _ in range(nb)])
    return float(d.mean()), float(np.percentile(v, 2.5)), float(np.percentile(v, 97.5)), len(d)

run.log("\n" + "=" * 100)
run.log("【H-C】 관문 — 성능")
run.log("=" * 100)
VERD, DIFF = {}, {}
def g_(k, v, d):
    VERD[k] = v; run.log(f"  {k:<4}{v}  {d}")

# ── 유의 반전: 개체별 SE + 본페로니 (R20)
Z_BONF = float(stats.norm.ppf(1 - 0.05 / (2 * max(len(RS), 1))))
run.log(f"  다중검정 보정 — 개체 {len(RS)}개 · 본페로니 z = {Z_BONF:.3f} (개별 95% 는 1.960)")
def sig_inv(r, k):
    a_, s_ = PER[r].get(k, np.nan), PER[r].get(k + "_se", np.nan)
    if not np.isfinite(a_) or not np.isfinite(s_):
        return False
    if s_ <= 0:
        return a_ < 0.5
    return a_ + Z_BONF * s_ < 0.5

# ── H1 두 템플릿 > 단일 오라클
m1, lo1, hi1, n1_ = boot_diff(A["two_orc"], A["orc1"], SEED0 + 1)
DIFF["H1"] = dict(mean=m1, lo=lo1, hi=hi1, n=n1_)
g_("H1", decide(lo1, hi1, 0.0, ">"),
   f"짝지은 (두템플릿 − 오라클단일) **{m1:+.4f}** [{lo1:+.4f}, {hi1:+.4f}] · n={n1_}"
   f" · 좋아진 개체 {int(np.nansum(A['two_orc'] > A['orc1']))}/{n1_}")

# ── H2 반전 0
for tag, k in (("다수결", "maj"), ("오라클단일", "orc1"), ("두템플릿", "two_orc")):
    iv = [r for r in RS if sig_inv(r, k)]
    run.log(f"  ({tag}) 유의 반전 {len(iv)}개: "
            + (", ".join(f"#{r}({PER[r][k]:.3f} ±{PER[r][k+'_se']:.3f})" for r in iv) or "없음"))
    if k == "two_orc":
        inv2 = iv
g_("H2", "✅ 지지" if len(inv2) == 0 else "❌ 기각",
   f"두 템플릿 유의 반전 **{len(inv2)}개**")

# ── H3 RR 위치형 대비 비열등 (Q7-E 에서 −0.1148 로 졌던 비교)
m3, lo3, hi3, n3_ = boot_diff(A["two_orc"], A["rr"], SEED0 + 3)
DIFF["H3"] = dict(mean=m3, lo=lo3, hi=hi3, n=n3_)
g_("H3", decide(lo3, hi3, -NI_RR, ">"),
   f"짝지은 (두템플릿 − RR위치형) **{m3:+.4f}** [{lo3:+.4f}, {hi3:+.4f}]"
   f" vs 여유 −{NI_RR}   (Q7-E 의 앵커는 여기서 −0.1148 로 졌다)")

# ── H6 무감독판
m6, lo6, hi6, n6_ = boot_diff(A["two_uns"], A["maj"], SEED0 + 6)
DIFF["H6"] = dict(mean=m6, lo=lo6, hi=hi6, n=n6_)
g_("H6", decide(lo6, hi6, 0.0, ">"),
   f"짝지은 (무감독 두템플릿 − 다수결단일) **{m6:+.4f}** [{lo6:+.4f}, {hi6:+.4f}] · n={n6_}"
   f"   ← 배포 가능성. Q7-E 결과로 보아 **실패가 기본 예측**")

# ── 최대 기여 개체 (R11-c)
d1 = A["two_orc"] - A["orc1"]
fin = np.isfinite(d1); tot = float(np.abs(d1[fin]).sum())
top = sorted(np.where(fin)[0], key=lambda i: -abs(d1[i]))[:5]
run.log(f"\n  기여 상위 5 (|Δ| 합 {tot:.3f} 중):")
for i in top:
    run.log(f"    #{RS[i]}  Δ {d1[i]:+.4f} ({abs(d1[i])/max(tot,1e-9)*100:.1f}%)"
            f" · 유병률 {A['prev'][i]:.3f} · σN/σS {A['sig_ratio'][i]:.2f}"
            f" · 끌림 {A['drag'][i]:.3f}")
if top:
    m1b, lo1b, hi1b, _ = boot_diff(np.delete(A["two_orc"], top[0]),
                                   np.delete(A["orc1"], top[0]), SEED0 + 7)
    msg = ("← 애초에 차이가 없다" if tot < 1e-6 else
           "← 개선이 살아남는다" if lo1b > 0 else "← **한 개체 이야기였다**")
    run.log(f"  최대 기여 개체 #{RS[top[0]]} 제외 → **{m1b:+.4f}** [{lo1b:+.4f}, {hi1b:+.4f}]   {msg}")
    DIFF["H1_ex_top"] = dict(mean=m1b, lo=lo1b, hi=hi1b, dropped=int(RS[top[0]]))

sd_a = float(np.nanstd(A["two_orc"], ddof=1)); sd_d = float(np.nanstd(d1[fin], ddof=1))
ratio = ("측정 불가" if min(sd_a, sd_d) < 1e-9 else
         f"짝지으면 {sd_a/sd_d:.1f}배 좁다" if sd_d < sd_a else f"짝지어도 안 좁아진다({sd_a/sd_d:.1f}배)")
run.log(f"\n  개체 간 SD — 두템플릿 {sd_a:.4f} · 짝지은 차 {sd_d:.4f}  ({ratio})")
run.log("    ⚠️ 채점된 부분집합의 산포는 모집단의 **하한**이다 (R17)")
CONFIG["gates"] = VERD; CONFIG["diffs"] = DIFF
CONFIG["inverted_two_orc"] = [int(r) for r in inv2]
run.save_json("config", CONFIG)
````

````python
# CELL 6 — 【H-D】 관문 H4·H5 — 두 가설을 가른다
#
#   (A) 오염 끌림  → **오라클로 고쳐진다**  → 끌림이 (오라클 − 다수결) 을 예측해야 한다
#   (B) 산포 비대칭 → **오라클로 안 고쳐진다** → σN/σS 가 오라클 AUROC 를 낮춰야 한다
def boot_rho(x, y, seed, nb=NB_BOOT):
    x, y = np.asarray(x, float), np.asarray(y, float)
    m = np.isfinite(x) & np.isfinite(y)
    x, y = x[m], y[m]
    if len(x) < 5:
        return float("nan"), float("nan"), float("nan"), 0
    r0 = float(stats.spearmanr(x, y).statistic)
    rng = np.random.RandomState(seed); v = []
    for _ in range(nb // 4):
        j = rng.randint(0, len(x), len(x))
        if len(np.unique(x[j])) > 2 and len(np.unique(y[j])) > 2:
            v.append(stats.spearmanr(x[j], y[j]).statistic)
    if len(v) < 20:
        return r0, float("nan"), float("nan"), len(x)
    return r0, float(np.percentile(v, 2.5)), float(np.percentile(v, 97.5)), len(x)

run.log("\n" + "=" * 100)
run.log("【H-D】 관문 — 기전 (A) 오염 끌림 vs (B) 산포 비대칭")
run.log("=" * 100)
gapOM = A["orc1"] - A["maj"]                 # 기준 교체로 회복되는 폭
r4, l4, h4, n4 = boot_rho(A["drag"], gapOM, SEED0 + 11)
DIFF["H4"] = dict(rho=r4, lo=l4, hi=h4, n=n4)
g_("H4", decide(l4, h4, 0.0, ">"),
   f"rho(끌림, 오라클−다수결) **{r4:+.3f}** [{l4:+.3f}, {h4:+.3f}] · n={n4}  [가설 A]")
r4b, l4b, h4b, _ = boot_rho(A["prev"], A["drag"], SEED0 + 12)
run.log(f"    (보조) rho(유병률, 끌림) {r4b:+.3f} [{l4b:+.3f}, {h4b:+.3f}]"
        "  ← 좌표별 중앙값이 유병률만큼 끌려간다는 예측")

r5, l5, h5, n5 = boot_rho(A["sig_ratio"], A["orc1"], SEED0 + 13)
DIFF["H5"] = dict(rho=r5, lo=l5, hi=h5, n=n5)
g_("H5", decide(l5, h5, 0.0, "<"),
   f"rho(σN/σS, 오라클 AUROC) **{r5:+.3f}** [{l5:+.3f}, {h5:+.3f}] · n={n5}  [가설 B]")

# ── ★ H5 는 부분적으로 구조적이다. SNR 을 통제한 **부분상관**을 병기한다.
#    오라클 점수가 ‖b−med(N)‖ 이므로 σ_N 은 분모처럼 들어간다 — 그만큼은 정의상 그렇다.
def partial_spearman(x, y, z):
    x, y, z = (np.asarray(v, float) for v in (x, y, z))
    m = np.isfinite(x) & np.isfinite(y) & np.isfinite(z)
    if m.sum() < 6:
        return float("nan")
    rx, ry, rz = (stats.rankdata(v[m]) for v in (x, y, z))
    def resid(a, b):
        b1 = np.c_[np.ones(len(b)), b]
        return a - b1 @ np.linalg.lstsq(b1, a, rcond=None)[0]
    return float(stats.spearmanr(resid(rx, rz), resid(ry, rz)).statistic)
pr = partial_spearman(A["sig_ratio"], A["orc1"], A["snr"])
run.log(f"    **SNR(=sep/σN) 통제 부분상관 {pr:+.3f}**"
        f"   {'← SNR 을 고정해도 남는다 = (B) 가 독립적으로 성립' if pr < -0.2 else '← SNR 로 대부분 설명된다 = (B) 는 구조적 잔상일 수 있다'}")
CONFIG["partial_rho_sigratio_orc1_given_snr"] = pr

# ── 두 가설의 상대적 크기 — 어느 쪽이 실제로 성능을 깎았나
run.log("\n  σN/σS 층별 (오라클·두템플릿 · 라벨 사용)")
for lo_, hi_ in ((0, 0.8), (0.8, 1.0), (1.0, 1.3), (1.3, 99)):
    s = (A["sig_ratio"] >= lo_) & (A["sig_ratio"] < hi_)
    if s.sum():
        run.log(f"    [{lo_:.1f},{hi_:.1f}) {int(s.sum()):>2}개체 · 다수결 {np.nanmean(A['maj'][s]):.4f}"
                f" · 오라클 {np.nanmean(A['orc1'][s]):.4f}"
                f" · **두템플릿 {np.nanmean(A['two_orc'][s]):.4f}**"
                f" · RR {np.nanmean(A['rr'][s]):.4f}")
run.log("\n  끌림 층별 (오라클 − 다수결 = 기준 오염의 크기)")
for lo_, hi_ in ((0, 0.05), (0.05, 0.15), (0.15, 0.40), (0.40, 99)):
    s = (A["drag"] >= lo_) & (A["drag"] < hi_)
    if s.sum():
        run.log(f"    [{lo_:.2f},{hi_:.2f}) {int(s.sum()):>2}개체 · 유병률 {np.nanmean(A['prev'][s]):.3f}"
                f" · 오라클−다수결 {np.nanmean(gapOM[s]):+.4f}")
CONFIG["gates"] = VERD; CONFIG["diffs"] = DIFF
run.save_json("config", CONFIG)
````

````python
# CELL 7 — 【H-E】 **Q7-E′ 재료 — 관문 아님.** 여기서 문턱을 고르지 않는다
#   Q7-E 가 남긴 숙제: 배포 조건은 「고칠 오염이 있는가」여야 하고 그건 유병률이 정한다.
#   유병률은 라벨이라 못 쓴다 → 무감독 대리치 `#(pre_rr < 0.85×중앙RR)/n` 가 쓸 만한지 본다.
run.log("\n" + "=" * 100)
run.log("【H-E】 Q7-E′ 재료 — 무감독 이소성 부담 대리치 (관문 아님)")
run.log("=" * 100)
rb, lb, hb, nb_ = boot_rho(A["burden_proxy"], A["prev"], SEED0 + 21)
run.log(f"  rho(부담 대리치, 실제 유병률) **{rb:+.3f}** [{lb:+.3f}, {hb:+.3f}] · n={nb_}")
run.log(f"  대리치 분포 — 중앙 {np.nanmedian(A['burden_proxy']):.4f}"
        f" · 최대 {np.nanmax(A['burden_proxy']):.4f}"
        f"  |  실제 유병률 중앙 {np.nanmedian(A['prev']):.4f} · 최대 {np.nanmax(A['prev']):.4f}")
rb2, lb2, hb2, _ = boot_rho(A["burden_proxy"], gapOM, SEED0 + 22)
run.log(f"  rho(부담 대리치, 오라클−다수결) {rb2:+.3f} [{lb2:+.3f}, {hb2:+.3f}]"
        "   ← 이게 양수여야 Q7-E′ 의 조건으로 쓸 수 있다")
run.log("  ⚠️ **여기서 문턱을 고르면 사후조정이다.** Q7-E′ 는 INCART 나 DEV 반쪽에서 정한다.")
deg = [r for r in RS if PER[r].get("uns_degenerate")]
run.log(f"  무감독 분할 퇴화 개체 {len(deg)}개: {deg or '없음'}")
CONFIG["q7e_prime_material"] = dict(
    rho_proxy_prev=[rb, lb, hb], rho_proxy_gap=[rb2, lb2, hb2],
    burden_q=BURDEN_Q, degenerate=[int(x) for x in deg],
    note="사후 재료 — 관문 아님. 문턱은 다른 코호트에서 사전등록한다")
run.save_json("config", CONFIG)
````
