## 이 실험이 고치는 것 — **I3 의 영가설이 틀렸다**

Q7-I 의 I3 은 「정합 조건에서 `f3`(보상성 휴지)가 **우연 0.5** 를 넘는가」였고
0.5745 [0.5321, 0.6162] 로 ✅ 였다. **그런데 같은 정합 조건의 다른 값을 나란히 놓으면
그림이 뒤집힌다.**

| 정합 조건 | 매크로 | 95% CI | 사전등록이 가정한 값 |
|---|---:|---|---|
| `f1` (`pre_rr` 의 함수) | **0.5779** | [0.5518, 0.6039] | **0.5 여야 했다** |
| `STT` (창 음성 대조) | **0.5833** | [0.5267, 0.6393] | 0.5 근처여야 했다 |
| **`f3`** (I3 의 주장) | **0.5745** | [0.5321, 0.6162] | — |

**`f3` 가 셋 중 최저다.** `f3 − f1 = −0.0034` · `f3 − STT = −0.0088`.

원인은 Q7-I 가 이미 냈다 — **정합 후 잔여 RR 격차 0.080**(정합 전 0.320 에서 −75%).
잔여만으로 `f1` 이 0.578 을 낸다. 즉 **정합 조건의 실제 바닥은 0.5 가 아니라 0.578** 이고,
`f3` 는 그 바닥을 못 넘는다.

> **I3 은 「지지」가 아니라 「적용 불가(영가설 오설정)」다.** 사전등록 시점에 「정합이
> `f1` 을 0.5 로 보낸다」를 가정했는데 **실측이 그 가정을 깼다**(R24-b).

## 사전등록 — 영가설을 **실측 바닥**으로 바꾼다

| 관문 | 내용 | 문턱 | 예상 |
|---|---|---|---|
| **J1** | **I3 재판정** — 짝지은 `f3ₘ − max(f1ₘ, STTₘ)` | CI 하한 **> 0** | **기각**(점추정 −0.009) |
| **J2** | **RR 비의존 조합** — 짝지은 `LR(f3,f4)ₘ − f1ₘ` | CI 하한 **> 0** | 경계선 |
| **J3** ★ | **정합이 실제로 작동하는 대역이 있나** — 가장 좁은 유효 대역의 `f1ₘ` | CI 상한 **< 0.52** | ? |
| **J4** | **`f2` 를 설계 대상 층에서 재평가** — 혼합+런 층의 짝지은 `f2_16 − f1` | CI 하한 **> 0** | ? |
| **J5** | **`f5` 는 런 검출자다** — `f1×f5` 상호작용을 넣은 `LR` vs `LR(전부)` | CI 하한 **> 0** | ? |
| **J6** | (관문 아님) `rho(T중첩, P_early \| RR격차)` 의 **부트스트랩 CI** | — | — |

**J3 이 이 실험의 축이다.** 대역을 좁히면 `f1` 은 0.5 로 가야 한다. **어느 대역에서
그렇게 되는지**를 먼저 찾고, **거기서** `f3` 를 읽는다. 그게 I3 의 진짜 답이다.
대가는 정합 가능 개체가 줄어드는 것이다 — 그 곡선을 한 장으로 낸다.

**유효 대역 선택 규칙(사전등록)**: `정합 가능 개체 ≥ MIN_MATCH_REC` 를 만족하는
**가장 좁은** 대역. 데이터를 보고 고르는 게 아니라 **규칙이 고른다**.

## 리뷰가 짚은 나머지 — 검정 가능한 형태로 바꿔 넣는다

**`f2` 를 전역 평균으로 버리지 마라.** `f1` 0.9367 > `f2_16` 0.9232 > `f2_8` 0.9164 라
전역에서는 진다. 그런데 **`f2` 를 만든 이유는 전역 중앙 RR 이 오염된 환자**(런 우세형)를
구하려던 것이었다. Q7-I 층별 표에 `f2` 가 없었다 — **설계 대상 층에서 이기는지를 안 봤다.**
→ **J4**.

**`f5` 는 S 판별자가 아니라 런 검출자다.** 단독 0.4693 · `<0.5` 33/59 로 역방향인데,
고립 S 는 직전이 정상이라 「직전 조기성」이 낮고 런 안의 S 는 직전도 S 라 높다.
이 코호트는 고립 비율 중앙 0.907 이라 고립이 지배한다. → **상호작용항으로 써야 한다**
(**J5**), 그리고 `rho(f5 개체 AUROC, 고립 S 비율)` 이 **음수**임을 보조로 확인한다.

**I5 미결은 천장 효과지 해악이 아니다.**

| 층 | 개체 | `f1` | `LR(전부)` | Δ | **`f1` 의 여유 공간** |
|---|---:|---:|---:|---:|---:|
| 고립 S | 42 | 0.9421 | 0.9642 | +0.0235 | 0.0579 |
| **혼합** | **12** | 0.8952 | 0.9371 | **+0.0443** | **0.1048** |
| 런 우세 | 5 | **0.9905** | 0.9825 | −0.0067 | **0.0095** |

여유가 가장 큰 **혼합 층에서 Δ 가 가장 크다.** **번들의 표적은 혼합 층**이고
런 우세 층은 천장이다 — Q7-I′ 사전등록에 그대로 적는다.

**`#865` 형은 이 축으로 안 고쳐진다.** 런 우세 층에서 오라클은 0.99 를 내는데
전이 모델은 0.0595 로 뒤집힌다. **특징 부족이 아니라 전이 시 기준선 반전**이고,
**정규화·재보정 문제**다 → 별도 축(**Q7-J**)으로 뗀다.

## 「번들이 성능을 올린다」가 아니라 **「번들이 특정 환자를 구한다」**

I1 +0.0252 는 작고 **편중돼 있다** — 좋아진 개체 45/59(76%)인데 **상위 5개체가 |Δ| 합의
52.8%** 를 차지한다(개체의 8.5%). `#867` 제외 시 +0.0252 → +0.0198(−21%).
CI 가 0 을 제외하므로 **효과는 실재**하지만, 서술은 「특정 환자를 구한다」가 정확하다.
→ 카드 `ailab-2026-0058` 을 그렇게 정정했다.

## 편상관은 **CI 를 안 냈다** — 여기서 낸다

Q7-I 【I-0】는 편상관 **점추정만** 출력했다(`+0.246`, `P_late` 은 `+0.109`).
리뷰가 인용한 구간(`[−0.013, +0.474]`)은 **이 실행의 산출물이 아니다.**
→ **J6** 에서 부트스트랩 CI 를 직접 낸다. 0 을 포함하면 **「T 의 독립 기여는 미결」**이
정확한 기록이고, 배제하면 **「독립 통로 확인」**이다.


```python
# CELL 0 — 공용 사전점검
import numpy as np
from scipy import stats

def decide(lo, hi, thr, direction):
    if direction not in (">", "<"):
        raise ValueError("direction 은 '>' 또는 '<'")
    if direction == ">":
        if lo > thr: return "✅ 지지"
        if hi < thr: return "❌ 기각"
    else:
        if hi < thr: return "✅ 지지"
        if lo > thr: return "❌ 기각"
    return "⚠️ 미결"

class AssetError(RuntimeError): pass
print("CELL 0 ✅")
```

```python
# CELL 3 — 【J-A】 특징·점수 (Q7-I 승계 · 여기서 재설계하지 않는다)
from sklearn.metrics import roc_auc_score
from sklearn.linear_model import LogisticRegression

def local_base(pre_v, k):
    n = len(pre_v); out = np.empty(n); med = float(np.median(pre_v))
    for i in range(n):
        a = max(0, i - k)
        out[i] = med if i - a < 3 else float(np.median(pre_v[a:i]))
    return out

def rhythm_feats(pre_v, post_v, ks):
    med = float(np.median(pre_v)); F, NM = [], []
    F.append(med - pre_v); NM.append("f1")
    b_first = None
    for k in ks:
        b = local_base(pre_v, k)
        if b_first is None:
            b_first = b
        F.append(1.0 - pre_v / np.maximum(b, 1e-9)); NM.append(f"f2_{k}")
    F.append(1.0 - (pre_v + post_v) / np.maximum(2.0 * b_first, 1e-9)); NM.append("f3")
    cv = np.empty(len(pre_v))
    for i in range(len(pre_v)):
        a = max(0, i - ks[0]); w = pre_v[a:i] if i - a >= 3 else pre_v[:3]
        cv[i] = float(np.std(w) / max(np.mean(w), 1e-9))
    F.append(cv); NM.append("f4")
    F.append(np.r_[0.0, F[1][:-1]]); NM.append("f5")
    return np.stack(F, axis=1), NM

def dist(B, ref):
    d = B - ref[None]
    return np.sqrt((d * d).sum(axis=(1, 2)))

def cv_logit(X, tt, K, seed, n_rep):
    accs = []
    for rep in range(max(n_rep, 1)):
        rng = np.random.RandomState(seed + 1000 * rep)
        fold = rng.permutation(len(tt)) % K
        sc = np.full(len(tt), np.nan)
        for f in range(K):
            te = fold == f; tr = ~te
            if not te.any() or int(tt[tr].sum()) < 2 or int((~tt[tr]).sum()) < 2:
                return None
            mu = X[tr].mean(0); sd = X[tr].std(0) + 1e-9
            lr = LogisticRegression(max_iter=2000, C=1.0)
            lr.fit((X[tr] - mu) / sd, tt[tr].astype(int))
            sc[te] = lr.decision_function((X[te] - mu) / sd)
        accs.append(sc)
    return np.mean(np.stack(accs), axis=0)

def two_template_cv(B, tt, K, seed, n_rep):
    accs = []
    for rep in range(max(n_rep, 1)):
        rng = np.random.RandomState(seed + 1000 * rep)
        fold = rng.permutation(len(tt)) % K
        sc = np.full(len(tt), np.nan)
        for f in range(K):
            te = fold == f; tr = ~te
            if not te.any() or int(tt[tr].sum()) < 2 or int((~tt[tr]).sum()) < 2:
                return None
            medN = np.median(B[tr & ~tt], axis=0); medS = np.median(B[tr & tt], axis=0)
            sc[te] = dist(B[te], medN) - dist(B[te], medS)
        accs.append(sc)
    return np.mean(np.stack(accs), axis=0)

def run_struct(t_):
    prev_ = np.r_[False, t_[:-1]]; next_ = np.r_[t_[1:], False]
    iso = t_ & ~prev_ & ~next_
    rl = mx = 0
    for v in t_:
        rl = rl + 1 if v else 0
        mx = max(mx, rl)
    return float(iso.sum() / max(t_.sum(), 1)), int(mx)

run.log("\n" + "=" * 100)
run.log("【J-A】 특징·점수 계산")
run.log("=" * 100)
SCORES, META, SKIP = {}, {}, []
for r in ALLR:
    mm = np.where(REC == r)[0]
    tt = (Y[mm] == IDX_S)
    if int(tt.sum()) < MIN_S_TPL or int((~tt).sum()) < MIN_S_TPL:
        SKIP.append((int(r), f"S {int(tt.sum())} · N {int((~tt).sum())} — 최소 {MIN_S_TPL} 미달")); continue
    pre_m, post_m = PRE[mm], POST[mm]
    X, NAMES = rhythm_feats(pre_m, post_m, KS)
    S = {nm_: X[:, j] for j, nm_ in enumerate(NAMES)}
    i1, i5 = NAMES.index("f1"), NAMES.index("f5")
    XI = np.c_[X, (X[:, i1] - X[:, i1].mean()) * (X[:, i5] - X[:, i5].mean())]   # ★ f1×f5
    arms = {"lr_all": X, "lr_norr": X[:, [NAMES.index("f3"), NAMES.index("f4")]],
            "lr_f1": X[:, [i1]], "lr_inter": XI}
    ok = True
    for nm_, XX in arms.items():
        sc_ = cv_logit(XX, tt, K_FOLD, SEED0, N_REPEAT)
        if sc_ is None:
            ok = False; break
        S[nm_] = sc_
    if not ok:
        SKIP.append((int(r), "겹 안 클래스 부족")); continue
    st = two_template_cv(BSTT[mm], tt, K_FOLD, SEED0, N_REPEAT)
    S["stt"] = st if st is not None else np.full(len(tt), np.nan)
    iso_f, mx_run = run_struct(tt)
    SCORES[int(r)] = S
    META[int(r)] = dict(tt=tt, pre=pre_m, n=int(len(mm)), pos=int(tt.sum()),
                        prev=float(tt.mean()), iso_frac=iso_f, max_run=mx_run,
                        rr_gap_pre=float(abs(np.median(pre_m[~tt]) - np.median(pre_m[tt]))
                                         / max(np.median(pre_m), 1e-9)))
RS = sorted(SCORES)
run.log(f"  채점 {len(RS)}개체 · 제외 {len(SKIP)}개체")
for r, why in SKIP:
    run.log(f"    #{r}  {why}")
if len(RS) < 10:
    raise AssetError("채점된 개체가 너무 적다")
ARMS = ["f1", "f2_8", "f2_16", "f3", "f4", "f5", "lr_f1", "lr_all", "lr_norr", "lr_inter", "stt"]
RAW = {a: np.array([roc_auc_score(META[r]["tt"].astype(int), SCORES[r][a])
                    if np.isfinite(SCORES[r][a]).all() else np.nan for r in RS]) for a in ARMS}
ISO = np.array([META[r]["iso_frac"] for r in RS])
run.log(f"\n  무정합 매크로 — " + " · ".join(f"{a} {np.nanmean(RAW[a]):.4f}" for a in
        ("f1", "f2_16", "f3", "lr_f1", "lr_all", "lr_inter", "stt")))
run.save_json("config", CONFIG)
```

```python
# CELL 4 — 【J-B】 ★ 정합 강도 곡선 — 대역을 좁히면 f1 이 0.5 로 가나
def matched_auc(sc, tt, pre_v, band_w, min_s, min_n):
    """같은 `pre_rr` 대역 안에서만 쌍을 센다. **평가만 제한**한다(Q7-F/I 승계).
    반환 (조건부 AUROC, 남은 S, 쌍, 대역수, **정합 후 잔여 RR 격차**)."""
    b = np.floor(pre_v / max(band_w, 1e-9)).astype(np.int64)
    num = den = 0.0; ks_ = nb_ = 0; ds, dn = [], []
    for bb in np.unique(b):
        m = b == bb
        s_, n_ = sc[m & tt], sc[m & ~tt]
        if len(s_) < min_s or len(n_) < min_n:
            continue
        ks_ += len(s_); nb_ += 1
        ds.append(pre_v[m & tt]); dn.append(pre_v[m & ~tt])
        gt = float((s_[:, None] > n_[None, :]).sum())
        eq = float((s_[:, None] == n_[None, :]).sum())
        num += gt + 0.5 * eq; den += float(len(s_) * len(n_))
    if den < 1:
        return float("nan"), ks_, 0.0, nb_, float("nan")
    a_ = np.concatenate(ds); b_ = np.concatenate(dn)
    resid = float(abs(np.median(b_) - np.median(a_)) / max(np.median(pre_v), 1e-9))
    return num / den, ks_, den, nb_, resid

run.log("\n" + "=" * 100)
run.log("【J-B】 정합 강도 곡선 — 대역폭 vs 잔여 격차 vs 각 팔")
run.log("=" * 100)
CURVE = {}
for bf in BANDS:
    M = {a: np.full(len(RS), np.nan) for a in ARMS}
    okm = np.zeros(len(RS), bool); resid = np.full(len(RS), np.nan)
    ns_ = np.full(len(RS), np.nan)
    for i, r in enumerate(RS):
        tt = META[r]["tt"]; pre_m = META[r]["pre"]
        bw = max(BAND_MIN, bf * np.median(pre_m))
        for a in ARMS:
            sc_ = SCORES[r][a]
            if not np.isfinite(sc_).all():
                continue
            v, ks_, pr_, nb_, rs_ = matched_auc(sc_, tt, pre_m, bw, MIN_BAND_S, MIN_BAND_N)
            M[a][i] = v
            if a == "f1":
                okm[i] = bool(ks_ >= MIN_MATCH_S and pr_ >= MIN_MATCH_PAIR)
                resid[i] = rs_; ns_[i] = ks_ / max(META[r]["pos"], 1)
    CURVE[bf] = dict(M=M, ok=okm, resid=resid, s_frac=ns_)
    run.log(f"  대역 {bf:.3f} — 정합 가능 **{int(okm.sum()):>2}/{len(RS)}** · "
            f"남은 S 비율 중앙 {np.nanmedian(ns_[okm]):.3f} · "
            f"**잔여 RR 격차 {np.nanmedian(resid[okm]):.4f}** | "
            f"f1 {np.nanmean(M['f1'][okm]):.4f} · f3 {np.nanmean(M['f3'][okm]):.4f} · "
            f"STT {np.nanmean(M['stt'][okm]):.4f} · LR(f3,f4) {np.nanmean(M['lr_norr'][okm]):.4f} · "
            f"LR(전부) {np.nanmean(M['lr_all'][okm]):.4f}")

# ── ★ 유효 대역은 **규칙이 고른다** — 데이터를 보고 고르지 않는다
elig = [bf for bf in BANDS if int(CURVE[bf]["ok"].sum()) >= MIN_MATCH_REC]
if not elig:
    raise AssetError(f"정합 가능 개체 ≥ {MIN_MATCH_REC} 인 대역이 없다 — 여기서 멈춘다")
BF = min(elig)                      # 가장 좁은 유효 대역
MOK = CURVE[BF]["ok"]; MM = CURVE[BF]["M"]
run.log(f"\n  ★ 유효 대역(규칙 선택: 정합 가능 ≥ {MIN_MATCH_REC} 중 가장 좁은 것) = **{BF:.3f}**"
        f" · 개체 {int(MOK.sum())} · 잔여 격차 {np.nanmedian(CURVE[BF]['resid'][MOK]):.4f}")
run.log(f"    (Q7-I 는 0.050 을 썼다 — 그때 잔여 0.080 · f1 0.5779)")
CONFIG["curve"] = {f"{bf:.3f}": dict(
    n_ok=int(CURVE[bf]["ok"].sum()),
    resid=float(np.nanmedian(CURVE[bf]["resid"][CURVE[bf]["ok"]])) if CURVE[bf]["ok"].any() else None,
    macro={a: float(np.nanmean(CURVE[bf]["M"][a][CURVE[bf]["ok"]])) for a in ARMS}) for bf in BANDS}
CONFIG["band_selected"] = BF
run.save_json("config", CONFIG)
```

```python
# CELL 5 — 【J-C】 관문 J1·J2·J3 — **영가설을 실측 바닥으로**
def boot_diff(a, b, seed, nb=NB_BOOT, mask=None, const=None):
    a = np.asarray(a, float)
    d = (a - const) if const is not None else (a - np.asarray(b, float))
    if mask is not None:
        d = d[mask]
    d = d[np.isfinite(d)]
    if len(d) < 3:
        return float("nan"), float("nan"), float("nan"), 0
    rng = np.random.RandomState(seed)
    v = np.array([d[rng.randint(0, len(d), len(d))].mean() for _ in range(nb)])
    return float(d.mean()), float(np.percentile(v, 2.5)), float(np.percentile(v, 97.5)), len(d)

run.log("\n" + "=" * 100)
run.log(f"【J-C】 관문 (유효 대역 {BF:.3f})")
run.log("=" * 100)
VERD, DIFF = {}, {}
def g_(k, v, d):
    VERD[k] = v; run.log(f"  {k:<4}{v}  {d}")

# ── J1 I3 재판정 — 영가설은 우연 0.5 가 아니라 **max(정합 f1, 정합 STT)**
FLOOR = np.nanmax(np.stack([MM["f1"], MM["stt"]]), axis=0)
run.log(f"  정합 조건 실측 — f1 {np.nanmean(MM['f1'][MOK]):.4f} · STT {np.nanmean(MM['stt'][MOK]):.4f}"
        f" · **바닥(개체별 max) {np.nanmean(FLOOR[MOK]):.4f}** · f3 {np.nanmean(MM['f3'][MOK]):.4f}")
m1, lo1, hi1, n1_ = boot_diff(MM["f3"], FLOOR, SEED0 + 1, mask=MOK)
DIFF["J1"] = dict(mean=m1, lo=lo1, hi=hi1, n=n1_)
g_("J1", decide(lo1, hi1, 0.0, ">"),
   f"짝지은 f3ₘ − max(f1ₘ, STTₘ) **{m1:+.4f}** [{lo1:+.4f}, {hi1:+.4f}] · n={n1_}"
   f"   ← **I3 재판정.** 우연 0.5 가 아니라 실측 바닥이 영가설이다")
m1b, lo1b, hi1b, _ = boot_diff(MM["f3"], MM["f1"], SEED0 + 2, mask=MOK)
run.log(f"    (참고) f3ₘ − f1ₘ  {m1b:+.4f} [{lo1b:+.4f}, {hi1b:+.4f}]")
m1c, lo1c, hi1c, _ = boot_diff(MM["f3"], MM["stt"], SEED0 + 3, mask=MOK)
run.log(f"    (참고) f3ₘ − STTₘ {m1c:+.4f} [{lo1c:+.4f}, {hi1c:+.4f}]")

# ── J2 RR 비의존 조합
m2, lo2, hi2, n2_ = boot_diff(MM["lr_norr"], MM["f1"], SEED0 + 4, mask=MOK)
DIFF["J2"] = dict(mean=m2, lo=lo2, hi=hi2, n=n2_)
g_("J2", decide(lo2, hi2, 0.0, ">"),
   f"짝지은 LR(f3,f4)ₘ − f1ₘ **{m2:+.4f}** [{lo2:+.4f}, {hi2:+.4f}] · n={n2_}"
   f"   ← f3 단독이 아니라 **RR 비의존 조합**이 바닥을 넘는가")
m2b, lo2b, hi2b, _ = boot_diff(MM["lr_norr"], FLOOR, SEED0 + 5, mask=MOK)
run.log(f"    (참고) LR(f3,f4)ₘ − max(f1ₘ, STTₘ) {m2b:+.4f} [{lo2b:+.4f}, {hi2b:+.4f}]")

# ── J3 ★ 정합이 실제로 작동하나
m3, lo3, hi3, n3_ = boot_diff(MM["f1"], None, SEED0 + 6, mask=MOK, const=0.0)
DIFF["J3"] = dict(mean=m3, lo=lo3, hi=hi3, n=n3_, band=BF)
g_("J3", decide(lo3, hi3, F1_COLLAPSE, "<"),
   f"유효 대역 {BF:.3f} 의 f1ₘ **{m3:.4f}** [{lo3:.4f}, {hi3:.4f}] vs 상한 {F1_COLLAPSE}"
   f"   ← 지지여야 그 대역의 f3 를 읽을 수 있다")
if not VERD["J3"].startswith("✅"):
    run.log("    ⛔ **정합이 f1 을 무력화하지 못했다.** 이 대역의 f3 값은 잔여 조기성을")
    run.log("       담고 있을 수 있으므로 **J1·J2 를 「post_rr 의 기여」로 인용하지 않는다**")
CONFIG["gates"] = VERD; CONFIG["diffs"] = DIFF
run.save_json("config", CONFIG)
```

```python
# CELL 6 — 【J-D】 관문 J4·J5 — 리뷰가 짚은 두 특징을 **설계 의도대로** 재평가
S_ISO = ISO >= ISO_HI; S_MIX = (ISO >= ISO_LO) & (ISO < ISO_HI); S_RUN = ISO < ISO_LO
run.log("\n" + "=" * 100)
run.log("【J-D】 f2 를 설계 대상 층에서 · f5 를 상호작용으로")
run.log("=" * 100)
run.log(f"  층 크기 — 고립 {int(S_ISO.sum())} · 혼합 {int(S_MIX.sum())} · 런 {int(S_RUN.sum())}")

# ── J4 f2 — 전역 평균이 아니라 **설계 대상 층**(전역 중앙 RR 이 오염된 층)에서
TARGET = S_MIX | S_RUN
run.log(f"\n  층별 f1 vs f2 (무정합)")
for nm_, msk in (("고립 S", S_ISO), ("혼합", S_MIX), ("런 우세", S_RUN)):
    if msk.sum() < 2:
        continue
    run.log(f"    {nm_:<8}{int(msk.sum()):>3}개체 · f1 {np.nanmean(RAW['f1'][msk]):.4f}"
            f" · f2_8 {np.nanmean(RAW['f2_8'][msk]):.4f}"
            f" · f2_16 {np.nanmean(RAW['f2_16'][msk]):.4f}"
            f" · 여유(1−f1) {1-np.nanmean(RAW['f1'][msk]):.4f}")
if int(TARGET.sum()) < 3:
    VERD["J4"] = "⚠️ 미결"
    run.log(f"  J4  ⚠️ 미결  대상 층 개체 {int(TARGET.sum())} < 3")
else:
    m4, lo4, hi4, n4_ = boot_diff(RAW["f2_16"], RAW["f1"], SEED0 + 11, mask=TARGET)
    DIFF["J4"] = dict(mean=m4, lo=lo4, hi=hi4, n=n4_)
    g_("J4", decide(lo4, hi4, 0.0, ">"),
       f"**혼합+런 층**({n4_}개체) 짝지은 f2_16 − f1 **{m4:+.4f}** [{lo4:+.4f}, {hi4:+.4f}]"
       f"   ← f2 는 **전역 중앙 RR 이 오염된 환자**를 구하려던 특징이다")
    m4b, lo4b, hi4b, _ = boot_diff(RAW["f2_16"], RAW["f1"], SEED0 + 12, mask=S_ISO)
    run.log(f"    (대조) 고립 S 층에서 f2_16 − f1 {m4b:+.4f} [{lo4b:+.4f}, {hi4b:+.4f}]"
            "  ← 설계 대상이 아닌 층")

# ── J5 f5 — 단독이 아니라 f1 과의 **상호작용**으로
m5, lo5, hi5, n5_ = boot_diff(RAW["lr_inter"], RAW["lr_all"], SEED0 + 13)
DIFF["J5"] = dict(mean=m5, lo=lo5, hi=hi5, n=n5_)
g_("J5", decide(lo5, hi5, 0.0, ">"),
   f"짝지은 LR(전부+f1×f5) − LR(전부) **{m5:+.4f}** [{lo5:+.4f}, {hi5:+.4f}] · n={n5_}")
rho5, lo5r, hi5r = (float(stats.spearmanr(RAW["f5"], ISO).statistic), None, None)
rngr = np.random.RandomState(SEED0 + 14)
vv = [stats.spearmanr(RAW["f5"][j], ISO[j]).statistic
      for j in (rngr.randint(0, len(RS), len(RS)) for _ in range(1000))]
vv = [v for v in vv if np.isfinite(v)]
run.log(f"    보조: rho(f5 개체 AUROC, 고립 S 비율) **{rho5:+.3f}** "
        f"[{np.percentile(vv,2.5):+.3f}, {np.percentile(vv,97.5):+.3f}]")
run.log("      → **음수면 f5 는 S 판별자가 아니라 런 검출자**다(고립이 많을수록 f5 가 낮다)")
CONFIG["gates"] = VERD; CONFIG["diffs"] = DIFF
CONFIG["rho_f5_iso"] = [rho5, float(np.percentile(vv, 2.5)), float(np.percentile(vv, 97.5))]
CONFIG["strata"] = dict(iso=int(S_ISO.sum()), mix=int(S_MIX.sum()), run=int(S_RUN.sum()))
run.save_json("config", CONFIG)
```

```python
# CELL 7 — 【J-E】 J6 — 편상관의 **부트스트랩 CI** (Q7-I 는 점추정만 냈다)
run.log("\n" + "=" * 100)
run.log("【J-E】 편상관 CI — 「T 의 독립 기여」를 구간으로 말한다")
run.log("=" * 100)
cfgs = sorted(glob.glob(os.path.join(PROJECT, "runs", "*q7f_window_confound", "config.json")))
if not cfgs:
    run.log("  ⚠️ Q7-F config.json 을 못 찾았다 — J6 **생략**(맞춰 넣지 않는다)")
else:
    QP = {int(k): v for k, v in json.load(open(cfgs[-1]))["per_record"].items()}
    qr = sorted(QP)
    ov = np.array([QP[r]["ov_gap_early"] for r in qr])
    pe = np.array([QP[r]["P_early"] for r in qr]); pl = np.array([QP[r]["P_late"] for r in qr])
    rg = np.array([QP[r]["rr_gap"] for r in qr])

    def partial_spearman(x, y, z):
        m = np.isfinite(x) & np.isfinite(y) & np.isfinite(z)
        if m.sum() < 6:
            return float("nan")
        rx, ry, rz = (stats.rankdata(v[m]) for v in (x, y, z))
        def resid(a, b):
            b1 = np.c_[np.ones(len(b)), b]
            return a - b1 @ np.linalg.lstsq(b1, a, rcond=None)[0]
        return float(stats.spearmanr(resid(rx, rz), resid(ry, rz)).statistic)

    def boot_partial(x, y, z, seed, nb=1000):
        r0 = partial_spearman(x, y, z)
        rng = np.random.RandomState(seed); v = []
        for _ in range(nb):
            j = rng.randint(0, len(x), len(x))
            if len(np.unique(z[j])) > 3:
                pv = partial_spearman(x[j], y[j], z[j])
                if np.isfinite(pv):
                    v.append(pv)
        if len(v) < 50:
            return r0, float("nan"), float("nan")
        return r0, float(np.percentile(v, 2.5)), float(np.percentile(v, 97.5))

    for tag, yv in (("P_early", pe), ("P_late", pl)):
        r0, lo_, hi_ = boot_partial(ov, yv, rg, SEED0 + 21)
        run.log(f"  rho(T중첩, {tag} | RR격차) = **{r0:+.3f}** [{lo_:+.3f}, {hi_:+.3f}] · n={len(qr)}")
        CONFIG.setdefault("partial_ci", {})[tag] = [r0, lo_, hi_]
    pc = CONFIG["partial_ci"]["P_early"]
    if np.isfinite(pc[1]) and pc[1] > 0:
        run.log("  → CI 가 0 을 배제한다: **직전 T 침입은 RR 격차 위의 독립 통로**다")
    else:
        run.log("  → CI 가 0 을 포함한다: **T 의 독립 기여는 미결**이다.")
        run.log("     「RR 이 주, T 가 부, T 의 독립 기여는 미결」이 정확한 기록이다")
run.save_json("config", CONFIG)
```
