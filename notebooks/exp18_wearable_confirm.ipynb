## 사전등록 (결과 보기 전에 고정)

**모든 관문은 시드 t-CI 로 3분 채점한다.** 점추정 채점은 금지(실험17 P-1 의 실패).

| | 예측 | 성격 |

> ⚠️ **`P-1` 은 6시드로도 아슬아슬할 수 있다.** `ILMI` 의 D 가 +0.0424 인데 문턱이 0.05 다.

> **`ILMI` 가 미결로 남아도 그것이 결과다**: 4전극에서 하후측벽만 경계이고,

````python
# CELL 0 — 공용 사전점검 (pipelines/ecg_preflight.py 인라인)
class LabelVocabError(ValueError):
    pass

# ── pipelines/ecg_preflight.py 인라인 (원본·테스트는 repo)
def assert_label_vocab(requested, available, kind="label", counts=None, min_count=1):
    """요청한 이름이 실제 어휘에 **전부** 있는지 확인한다. 하나라도 없으면 예외.

    0건은 "데이터에 그 소견이 없다"가 아니라 **대개 이름을 잘못 골랐다**는 뜻이다.
    그 둘을 구별하려고 어휘 자체를 대조한다.

    requested : 쓰려는 이름들
    available : 데이터에서 실제로 관측된 이름 집합
    counts    : {이름: 건수} (있으면 min_count 미만도 함께 보고)
    """
    requested, available = list(requested), set(available)
    unknown = [r for r in requested if r not in available]
    if unknown:
        raise LabelVocabError(
            f"{kind} 어휘에 없는 이름 {unknown}.\n"
            f"  → 0건이 나온 이유는 '데이터에 없어서'가 아니라 **이름이 틀려서**다.\n"
            f"  실제 어휘({len(available)}개): {sorted(available)}"
        )
    thin = []
    if counts:
        thin = [(r, counts.get(r, 0)) for r in requested if counts.get(r, 0) < min_count]
    return {"ok": True, "n_requested": len(requested), "thin": thin}

def decide(lo, hi, thr, direction):
    """사전등록 관문의 **유일한** 계약: 지지(True) / 기각(False) / 미결(None).

    CI 가 임계값을 걸치면 **기각이 아니라 미결**이다. 검정력 부족을 반증으로
    위장하지 않기 위해서다. 점추정 2분 채점은 금지한다(실험13b·14 에서 그 실수를 했다).
    """
    if direction not in (">", "<"):
        raise ValueError("direction 은 '>' 또는 '<'")
    if direction == ">":
        if lo > thr:
            return True
        if hi < thr:
            return False
    else:
        if hi < thr:
            return True
        if lo > thr:
            return False
    return None

MARK = {True: "✅ 지지", False: "❌ 기각", None: "⚠️ 미결"}

def collapse_report(scores, floors, names, lift=1.2):
    """붕괴 감시 — 단위(클래스·부위)별로 판정한다.

    scores : {이름: 성능}   floors : {이름: 무작위 수준(유병률 등)}
    성능이 floor 의 `lift` 배도 안 되면 그 단위는 죽은 것으로 본다.

    ★ 하나 죽었다고 실험 전체를 무효화하지 않는다. 무효 판단은 호출자가
      '과반이 죽었는가 / 대조군 한쪽이 통째로 죽었는가'로 따로 한다.
    """
    dead = [n for n in names if scores.get(n, 0.0) < floors.get(n, 0.0) * lift]
    alive = [n for n in names if n not in dead]
    return {"dead": dead, "alive": alive,
            "fatal_majority": len(dead) >= len(names) / 2 if names else True}

def assert_arm_shape(arm, expected_rows, name="arm"):
    """저장된 arm 의 행 수가 **겹 크기**인지 확인한다.

    MedKOSRun.save_arm 은 그 겹의 예측만 저장한다(전체 길이가 아니다).
    겹 순서는 `np.where(CV == k)[0]` 의 오름차순이므로,
      OOF[np.where(CV == k)[0]] = load_arm(...)      ← 이렇게 **넣는다**
      load_arm(...)[전역인덱스]                       ← 이렇게 자르면 IndexError
    실험15d G0 에서 이 혼동으로 터졌다.
    """
    n = arm.shape[0]
    if n != expected_rows:
        raise ValueError(
            f"{name} 행 수 {n} != 기대 {expected_rows}.\n"
            "  → arm 은 **겹 크기**로 저장된다. 전역 인덱스로 자르지 말고 "
            "OOF[np.where(CV==k)[0]] = arm 형태로 넣을 것."
        )
    return {"ok": True, "rows": n}

FRONTAL_IDENTITIES = (
    ("III", 2, lambda I, II: II - I),                 # 아인트호벤
    ("aVR", 3, lambda I, II: -(I + II) / 2.0),        # 골드버거
    ("aVL", 4, lambda I, II: I - II / 2.0),
    ("aVF", 5, lambda I, II: II - I / 2.0),
)

def assert_lead_order(X, tol=0.02, sample=200, seed=0):
    """12유도 캐시의 **채널 순서**를 신호 자체로 검증한다.

    헤더의 유도 이름을 믿지 말고 아인트호벤·골드버거 항등식으로 확인한다:
        III = II − I,  aVR = −(I+II)/2,  aVL = I − II/2,  aVF = II − I/2
    넷이 모두 맞으면 0..5 = I,II,III,aVR,aVL,aVF 이고 표준 순서상 6..11 = V1..V6 이다.
    → `{I,II}` = [0,1], `{II,V1}` = [1,6] 을 쓸 근거가 생긴다.

    유도 순서를 틀리면 **예외 없이 조용히 다른 실험**이 된다. 그래서 잰다.
    ※ 원신호(mV) 전제 — 채널별로 정규화한 배열에는 쓸 수 없다.
    """
    import numpy as np
    if X.ndim != 3 or X.shape[2] != 12:
        raise ValueError(f"X 는 (n, t, 12) 여야 한다 — 받은 모양 {X.shape}")
    rs = np.random.RandomState(seed)
    idx = rs.choice(len(X), size=min(sample, len(X)), replace=False)
    S = X[idx].astype("float64")
    I, II = S[:, :, 0], S[:, :, 1]
    report, bad = {}, []
    for name, j, f in FRONTAL_IDENTITIES:
        want = f(I, II)
        scale = np.abs(want).mean() + 1e-9
        err = float(np.abs(S[:, :, j] - want).mean() / scale)
        report[name] = err
        if err > tol:
            bad.append(f"{name}(ch{j}) 상대오차 {err:.3f}")
    if bad:
        raise ValueError(
            "유도 순서가 표준(I,II,III,aVR,aVL,aVF,V1..V6)이 아니다: " + ", ".join(bad) + "\n"
            "  → 항등식이 깨졌다는 것은 채널 배치가 다르거나 채널별 정규화가 걸렸다는 뜻이다.\n"
            "  마스크 인덱스([0,1] 사지 · [1,6] II+V1)를 그대로 쓰면 조용히 다른 실험이 된다."
        )
    return {"ok": True, "n_checked": len(idx), "rel_err": report}

print("사전점검 적재: assert_label_vocab · decide · collapse_report · "
      "assert_arm_shape · assert_lead_order")
````

````python
# CELL 3 — 학습 (45회) · 실험16·17 arm 은 읽어 쓴다
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.metrics import roc_auc_score as _auc

def build_head(seed):
    """★ 실험15~17 과 완전히 동일."""
    tf.keras.utils.set_random_seed(seed)
    si = layers.Input((X.shape[1], 12))
    x = si
    for f, k in ((32, 9), (64, 7), (128, 5), (128, 3)):
        x = layers.Conv1D(f, k, padding="same", activation="relu")(x)
        x = layers.BatchNormalization()(x)
        x = layers.MaxPooling1D(2)(x)
    h = layers.Dense(64, activation="relu")(layers.GlobalAveragePooling1D()(x))
    h = layers.Dropout(0.3)(h); h = layers.Dense(64, activation="relu")(h)
    m = models.Model(si, layers.Dense(NS, activation="sigmoid")(h))
    m.compile(optimizer=tf.keras.optimizers.Adam(1e-3, clipnorm=1.0),
              loss="binary_crossentropy")
    return m

def split(k):
    te = np.where(CV == k)[0]; rest = np.where(CV != k)[0]
    rs_ = np.random.RandomState(SEED0 + k); rs_.shuffle(rest)
    n_val = max(int(len(rest) * 0.12), 200)
    return te, rest[:n_val], rest[n_val:]

def train_one(c, sd, k, save=True):
    te, va, tr = split(k)
    mk = MASKS[c]
    m = build_head(SEED0 + 100 * k + 15 + sd)      # ★ 15~17 과 같은 시드 공식
    m.fit(X[tr] * mk, Ymul[tr], validation_data=(X[va] * mk, Ymul[va]),
          epochs=EPOCHS, batch_size=128, verbose=0)
    p = m.predict(X[te] * mk, batch_size=512, verbose=0)
    if save and k == 0 and sd == SEEDS_OLD[0]:
        run.save_model(m, f"head_{c}")
    tf.keras.backend.clear_session()
    return te, p

# ── 【G0-c】 재사용 정합성 — {I,II} 시드0 재학습 대조 (5회)
run.log("\n【G0-c】 재사용 정합성 — {I,II} 시드0 재학습 대조")
_ref = np.zeros((len(EID), NS), "float32"); _src = set()
for k in range(K_FOLD):
    a, tag = find_arm("I+II", 0, k)
    if a is None:
        raise RuntimeError(f"I+II_s0_f{k} 를 어디서도 못 찾았습니다")
    _ref[np.where(CV == k)[0]] = a; _src.add(tag)
_chk = np.zeros((len(EID), NS), "float32"); _t = time.time()
for k in range(K_FOLD):
    te, p = train_one("I+II", 0, k, save=False)
    _chk[te] = p
PER = (time.time() - _t) / K_FOLD
drift = {s: float(_auc(Ymul[:, j].astype(bool), _chk[:, j])
                  - _auc(Ymul[:, j].astype(bool), _ref[:, j]))
         for j, s in enumerate(SITES)}
DRIFT_MAX = max(abs(v) for v in drift.values())
run.log(f"  출처 {_src} · ΔAUROC " + " ".join(f"{s}{drift[s]:+.4f}" for s in SITES))
run.log(f"  → 최대 |Δ| = {DRIFT_MAX:.4f} (문턱 {DRIFT_THR}) · 1회 {PER:.0f}s")
REUSE_OK = DRIFT_MAX <= DRIFT_THR
run.log("  " + ("✅ 재사용 확정" if REUSE_OK else f"⚠️ 환경이 다르다 → {ON_DRIFT}"))
if not REUSE_OK and ON_DRIFT == "stop":
    raise RuntimeError(f"드리프트 {DRIFT_MAX:.4f} > {DRIFT_THR}")

# ── 필요한 arm 을 모은다 (없으면 학습)
#    드리프트가 있으면 부모 arm 을 **믿지 않고** 전부 이 세션에서 다시 학습한다.
NO_REUSE = (not REUSE_OK) and ON_DRIFT == "retrain"
TODO = {(c, sd, k) for c, _s in WANT.items() for sd in _s for k in range(K_FOLD)
        if NO_REUSE or not has_parent_arm(c, sd, k)}
run.log(f"\n학습 대상 {len(TODO)}회 (예상 ≈ {len(TODO) * PER / 60:.0f}분)"
        + ("  ※ 드리프트 감지 → 부모 arm 미사용" if NO_REUSE else ""))
OOF = {c: {} for c in WANT}
t0, done = time.time(), 0
for c, _sds in WANT.items():
    for sd in _sds:
        OOF[c][sd] = np.zeros((len(EID), NS), "float32")
        for k in range(K_FOLD):
            te = np.where(CV == k)[0]
            a = run.load_arm(f"{c}_s{sd}_f{k}")          # 이번 실행 체크포인트 우선(재개)
            if a is None and (c, sd, k) not in TODO:
                a, _ = find_arm(c, sd, k)                # 부모에서 재사용
            if a is None:
                te, a = train_one(c, sd, k)
                run.save_arm(f"{c}_s{sd}_f{k}", a)
                done += 1
                if done == 1:
                    run.log(f"  ⏱ 첫 학습 {time.time()-t0:.0f}s")
            assert_arm_shape(a, len(te), name=f"{c}_s{sd}_f{k}")
            OOF[c][sd][te] = a
        run.log(f"  {c:<12} 시드{sd} 준비 ({done} 학습 · {time.time()-t0:.0f}s)")
run.log(f"\n총 {time.time()-t0:.0f}s · 이번 세션 학습 {done}회")
````

````python
# CELL 4 — 지표 (구성 × 부위 × 시드)
from sklearn.metrics import average_precision_score
from scipy import stats

def spec_at_sens(score, pos, target=0.90):
    p, n = score[pos], score[~pos]
    if len(p) == 0 or len(n) == 0:
        return np.nan, np.nan
    thr = float(np.quantile(p, 1.0 - target, method="lower"))
    return float((n < thr).mean()), float((score >= thr).mean())

def t_ci(vals, conf=0.95):
    v = np.asarray([x for x in vals if np.isfinite(x)], float)
    n = len(v); m = float(v.mean()) if n else float("nan")
    if n < 2:
        return m, np.nan, np.nan, 0.0
    sd = float(v.std(ddof=1))
    return m, m - stats.t.ppf(.5 + conf / 2, n - 1) * sd / np.sqrt(n), \
           m + stats.t.ppf(.5 + conf / 2, n - 1) * sd / np.sqrt(n), sd

ORDER = ["II", "I+II", "II+V1", MAIN, "I+II+V2", FIVE_A, FIVE_B, REF]
M = {c: {} for c in ORDER}
for c in ORDER:
    for j, s in enumerate(SITES):
        y = Ymul[:, j].astype(bool)
        M[c][s] = {"auroc": {}, "auprc": {}, "alarm": {}, "spec": {}}
        for sd in WANT[c]:
            sc = OOF[c][sd][:, j]
            M[c][s]["auroc"][sd] = float(_auc(y, sc))
            M[c][s]["auprc"][sd] = float(average_precision_score(y, sc))
            sp, al = spec_at_sens(sc, y)
            M[c][s]["spec"][sd] = sp; M[c][s]["alarm"][sd] = al

def Dvals(c, s):
    """구성 c·부위 s 의 시드별 D. REF 와 **같은 시드**끼리만 짝짓는다."""
    sds = [x for x in WANT[c] if x in WANT[REF]]
    return [M[REF][s]["auroc"][x] - M[c][s]["auroc"][x] for x in sds]

run.log("\n" + "=" * 130)
run.log("【AUROC】 시드 평균 · 괄호는 (전극 수 / 시드 수)")
run.log("=" * 130)
run.log(f"  {'부위':<7}{'평면':<7}{'n':>6}"
        + "".join(f"{c + f'({N_ELEC[c]}/{len(WANT[c])})':>18}" for c in ORDER))
for j, s in enumerate(SITES):
    run.log(f"  {s:<7}{SITE_PLANE[s]:<7}{int(Ymul[:, j].sum()):>6,}"
            + "".join(f"{np.mean(list(M[c][s]['auroc'].values())):>18.4f}" for c in ORDER))

run.log("\n【잔여 결손 D = AUROC(12) − AUROC(구성)】")
run.log(f"  {'부위':<7}" + "".join(f"{c:>18}" for c in ORDER[:-1]))
for s in SITES:
    run.log(f"  {s:<7}" + "".join(f"{np.mean(Dvals(c, s)):>+18.4f}" for c in ORDER[:-1]))

run.log("\n【경보율】 민감도 0.90")
run.log(f"  {'부위':<7}" + "".join(f"{c:>18}" for c in ORDER))
for s in SITES:
    run.log(f"  {s:<7}" + "".join(
        f"{np.nanmean(list(M[c][s]['alarm'].values())):>18.3f}" for c in ORDER))

prev = {s: float(Ymul[:, j].mean()) for j, s in enumerate(SITES)}
FATAL = False
run.log("\n【붕괴 감시】")
for c in ORDER:
    r = collapse_report({s: np.mean(list(M[c][s]["auprc"].values())) for s in SITES},
                        prev, SITES)
    FATAL = FATAL or r["fatal_majority"]
    run.log(f"  {c:<12} 사망 {r['dead'] or '없음'}"
            + ("  ⛔ 과반" if r["fatal_majority"] else ""))
````

````python
# CELL 5 — 【A0】 실험17 P-1 재채점 + 【A1】 부위별 라우팅 천장 (학습 0회)
run.log("\n" + "=" * 100)
run.log("【A0】 실험17 P-1 재채점 — 점추정이 아니라 시드 t-CI 로")
run.log("=" * 100)
run.log("  실험17 은 IMI 초과 +0.0108 > 0.01 을 **점추정으로** 기각했다(규약 위반).")
A0 = {}
for s in SITES:
    sds = [x for x in SEEDS_OLD if x in WANT[MAIN] and x in WANT["I+II"]]
    base = [min(M[REF][s]["auroc"][x] - M["I+II"][s]["auroc"][x],
                M[REF][s]["auroc"][x] - M["II+V1"][s]["auroc"][x]) for x in sds]
    cur = [M[REF][s]["auroc"][x] - M[MAIN][s]["auroc"][x] for x in sds]
    exc = [cur[i] - base[i] for i in range(len(sds))]
    m, lo, hi, sd = t_ci(exc)
    v = decide(lo, hi, UNION_EPS, "<")
    A0[s] = {"mean": m, "ci": [lo, hi], "sd": sd, "verdict": v}
    run.log(f"  {s:<7}초과 {m:>+8.4f} [{lo:+.4f}, {hi:+.4f}] (시드SD {sd:.4f}) "
            f"vs {UNION_EPS} → {MARK[v]}")
A0_IMI = A0["IMI"]["verdict"]
run.log(f"\n  → IMI 재채점 {MARK[A0_IMI]}"
        + ("  실험17 의 '기각' 은 **정정된다**" if A0_IMI is not False else
           "  실험17 의 기각이 유지된다"))

run.log("\n" + "=" * 100)
run.log("【A1】 부위별 유도 라우팅 천장 — 5전극으로 잡은 신호를 부위마다 다르게 쓰면")
run.log("=" * 100)
run.log("  ★ 중첩 선택: 겹 k 의 구성은 **다른 4겹**에서 고른다(테스트에 대고 고르지 않는다)")
POOL = ["I+II", MAIN, FIVE_A]           # 5전극 기기가 실제로 낼 수 있는 부분집합
run.log(f"  후보 {POOL}")
A1, route = {}, {}
_sd0 = SEEDS_OLD
for j, s in enumerate(SITES):
    y = Ymul[:, j].astype(bool)
    picked, sel = np.zeros(len(EID), "float32"), []
    for k in range(K_FOLD):
        te, other = np.where(CV == k)[0], np.where(CV != k)[0]
        best, bs = None, -9
        for c in POOL:                   # 다른 4겹에서만 고른다
            v = np.mean([_auc(y[other], OOF[c][x][other, j]) for x in _sd0])
            if v > bs:
                bs, best = v, c
        sel.append(best)
        picked[te] = np.mean([OOF[best][x][te, j] for x in _sd0], axis=0)
    au = float(_auc(y, picked))
    ref = float(np.mean([M[REF][s]["auroc"][x] for x in _sd0]))
    A1[s] = {"auroc": au, "D": ref - au, "picked": sel}
    route[s] = max(set(sel), key=sel.count)
    run.log(f"  {s:<7}{SITE_PLANE[s]:<7}선택 {route[s]:<12} AUROC {au:.4f} "
            f"D {ref - au:+.4f}   (겹별 {sel})")
A1_WORST = max(v["D"] for v in A1.values())
run.log(f"\n  → 라우팅 천장: 최악 D = {A1_WORST:+.4f} · "
        f"중앙값 {np.median([v['D'] for v in A1.values()]):+.4f}")
run.log(f"  고정 {FIVE_A} 최악 D = {max(np.mean(Dvals(FIVE_A, s)) for s in SITES):+.4f}")
run.log("  " + ("→ 라우팅 이득이 있다. 실험19(평면 분리 헤드) 착수 근거가 된다"
                if A1_WORST < max(np.mean(Dvals(FIVE_A, s)) for s in SITES) - 0.005
                else "→ 라우팅 이득이 작다. 구조를 바꿔도 못 얻는다 — 외부검증으로 간다"))
````

````python
# CELL 6 — 사전등록 채점 (전부 시드 t-CI · 점추정 채점 금지)
run.log("\n" + "=" * 100)
run.log(f"【사전등록 채점】 시드 {len(WANT[MAIN])}개 · t-CI(df={len(WANT[MAIN])-1})")
run.log("=" * 100)
V = {}

# ── P-1 ★★ 배포 관문 (6시드)
run.log(f"\n  P-1 배포 관문 — {MAIN}({N_ELEC[MAIN]}전극) 의 D 가 전 부위 < {D_THR}")
p1, worst = [], (None, -9)
for s in SITES:
    m, lo, hi, sd = t_ci(Dvals(MAIN, s))
    v = decide(lo, hi, D_THR, "<"); p1.append(v)
    if m > worst[1]:
        worst = (s, m)
    run.log(f"      {s:<7}{SITE_PLANE[s]:<7}D = {m:>+8.4f} [{lo:+.4f}, {hi:+.4f}] "
            f"(SD {sd:.4f}) → {MARK[v]}")
V["P-1"] = True if all(x is True for x in p1) else \
    False if any(x is False for x in p1) else None
run.log(f"  P-1 → {MARK[V['P-1']]}  최악 {worst[0]} D={worst[1]:+.4f}")

# ── P-2 ★ 경보 관문
run.log(f"\n  P-2 경보 관문 — 배수 중앙값 t-CI 상한 < {ALARM_R}")
sds = [x for x in WANT[MAIN] if x in WANT[REF]]
def _ratio(sd_):
    v = [M[MAIN][s]["alarm"][sd_] / M[REF][s]["alarm"][sd_] for s in SITES
         if M[REF][s]["alarm"][sd_] and np.isfinite(M[MAIN][s]["alarm"][sd_])]
    return float(np.median(v)) if v else float("nan")
rr = [_ratio(x) for x in sds]
for s in SITES:
    a = np.nanmean([M[MAIN][s]["alarm"][x] for x in sds])
    b = np.nanmean([M[REF][s]["alarm"][x] for x in sds])
    run.log(f"      {s:<7}{a:>8.3f}{b:>8.3f}{a/b if b else float('nan'):>8.2f}배")
m2, lo2, hi2, sd2 = t_ci(rr)
V["P-2"] = decide(lo2, hi2, ALARM_R, "<")
run.log(f"  P-2 → 중앙값 {m2:.3f} [{lo2:.3f}, {hi2:.3f}] (SD {sd2:.3f}) → {MARK[V['P-2']]}")

# ── P-3 V2 (주가설 = ASMI 단일 검정)
run.log(f"\n  P-3 전벽 커버 유도 — **{P3_SITE} 단일 검정** D(I+II+V2) < D({MAIN})")
sds3 = [x for x in WANT["I+II+V2"] if x in WANT[MAIN]]
d3 = [(M[REF][P3_SITE]["auroc"][x] - M["I+II+V2"][P3_SITE]["auroc"][x])
      - (M[REF][P3_SITE]["auroc"][x] - M[MAIN][P3_SITE]["auroc"][x]) for x in sds3]
m3, lo3, hi3, sd3 = t_ci(d3)
V["P-3"] = decide(lo3, hi3, 0.0, "<")
run.log(f"  P-3 → ΔD(V2−V1) = {m3:+.4f} [{lo3:+.4f}, {hi3:+.4f}] (SD {sd3:.4f}) "
        f"→ {MARK[V['P-3']]}" + ("   전벽은 V2 로" if V["P-3"] else "   V1 유지"))
run.log("      (탐색) 다른 부위: " + " ".join(
    f"{s} {np.mean(Dvals('I+II+V2', s)) - np.mean(Dvals(MAIN, s)):+.4f}"
    for s in SITES if s != P3_SITE))

# ── P-4 V5 (주가설 = ILMI 단일 검정)
run.log(f"\n  P-4 5번째 전극 — **{P4_SITE} 단일 검정** D({FIVE_A}) < D({MAIN})")
sds4 = [x for x in WANT[FIVE_A] if x in WANT[MAIN]]
d4 = [(M[REF][P4_SITE]["auroc"][x] - M[FIVE_A][P4_SITE]["auroc"][x])
      - (M[REF][P4_SITE]["auroc"][x] - M[MAIN][P4_SITE]["auroc"][x]) for x in sds4]
m4, lo4, hi4, sd4 = t_ci(d4)
V["P-4"] = decide(lo4, hi4, 0.0, "<")
run.log(f"  P-4 → ΔD(+V5) = {m4:+.4f} [{lo4:+.4f}, {hi4:+.4f}] (SD {sd4:.4f}) "
        f"→ {MARK[V['P-4']]}" + ("   5번째 전극 값어치 있음" if V["P-4"] else ""))
run.log("      (탐색) 다른 부위: " + " ".join(
    f"{s} {np.mean(Dvals(FIVE_A, s)) - np.mean(Dvals(MAIN, s)):+.4f}"
    for s in SITES if s != P4_SITE))

# ── P-5 5전극 최종 사양: V2+V5 vs V1+V5 (최악 D 로 비교)
run.log(f"\n  P-5 5전극 사양 — {FIVE_B} 의 최악 D < {FIVE_A} 의 최악 D")
sds5 = [x for x in WANT[FIVE_A] if x in WANT[FIVE_B] and x in WANT[REF]]
def _worst(c, x):
    return max(M[REF][s]["auroc"][x] - M[c][s]["auroc"][x] for s in SITES)
d5 = [_worst(FIVE_B, x) - _worst(FIVE_A, x) for x in sds5]
m5, lo5, hi5, sd5 = t_ci(d5)
V["P-5"] = decide(lo5, hi5, 0.0, "<")
run.log(f"      시드별 최악 D — {FIVE_A}: "
        + " ".join(f"{_worst(FIVE_A, x):+.4f}" for x in sds5))
run.log(f"      시드별 최악 D — {FIVE_B}: "
        + " ".join(f"{_worst(FIVE_B, x):+.4f}" for x in sds5))
run.log(f"  P-5 → Δ최악 = {m5:+.4f} [{lo5:+.4f}, {hi5:+.4f}] → {MARK[V['P-5']]}")

run.log("\n  【효과 vs 시드 잡음】 규약 ②")
for nm, mm, ss in (("P-3 V2−V1", m3, sd3), ("P-4 +V5", m4, sd4), ("P-5 최악", m5, sd5)):
    r = abs(mm) / ss if ss > 0 else float("inf")
    run.log(f"      {nm:<12} |효과| {abs(mm):.4f} / SD {ss:.4f} = {r:.2f}배"
            + ("" if r >= 1 else "  ⚠️ 잡음 이하 — 검출 불가로 종결"))

run.log("\n" + "=" * 100)
for k in ("P-1", "P-2", "P-3", "P-4", "P-5"):
    run.log(f"  {k}: {MARK[V.get(k)]}")
run.log(f"  A0 실험17 P-1 재채점: {MARK[A0_IMI]}  ·  A1 라우팅 천장 최악 D {A1_WORST:+.4f}")
if FATAL:
    run.log("  ⛔ 붕괴 과반 — 판정 무효")
run.log("=" * 100)

# ── 배포 권고: 관문 결과를 사양으로 번역한다 (판정 아님 — 결론 정리)
run.log("\n" + "=" * 100)
run.log("【배포 권고】 관문 결과 → 사양")
run.log("=" * 100)
undecided = [SITES[i] for i, x in enumerate(p1) if x is not True]
best5 = FIVE_B if V.get("P-5") is True else FIVE_A
w5 = {s: np.mean(Dvals(best5, s)) for s in SITES}
run.log(f"  4전극 {MAIN}: 7부위 중 {sum(1 for x in p1 if x is True)}개 확정"
        + (f" · 미확정 {undecided}" if undecided else ""))
run.log(f"  5전극 {best5}: 최악 D {max(w5.values()):+.4f}({max(w5, key=w5.get)}) · "
        f"중앙값 {np.median(list(w5.values())):+.4f}")
if V.get("P-1") is True:
    run.log("  → **4전극으로 확정.** 외부검증(Chapman/Ningbo)으로 넘어간다")
elif undecided and max(np.mean(Dvals(MAIN, s)) for s in undecided) < D_THR:
    run.log(f"  → 4전극의 점추정은 전부 문턱 아래인데 {undecided} 의 CI 가 걸친다.")
    run.log(f"     {undecided} 는 5전극에서 D "
            + " ".join(f"{s} {w5[s]:+.4f}" for s in undecided)
            + " 로 해소되는지 함께 본다.")
    run.log("     → **경계 부위는 5전극을 권고**하고, 4전극은 '점추정 기준 충족'으로 보고한다")
else:
    run.log("  → 4전극으로는 부족하다. 5전극 사양을 주 후보로 올린다")
# 시드를 더 늘리면 갈리는가 (규약 ②)
run.log("\n  【남은 미결에 시드가 답인가】")
for s in undecided:
    m_, lo_, hi_, sd_ = t_ci(Dvals(MAIN, s))
    if m_ >= D_THR:
        run.log(f"      {s:<7}D {m_:+.4f} — 점추정이 문턱 위다. 시드로 안 풀린다")
        continue
    n_ = next((n for n in range(len(WANT[MAIN]) + 1, 61)
               if m_ + stats.t.ppf(.975, n - 1) * sd_ / np.sqrt(n) < D_THR), None)
    run.log(f"      {s:<7}D {m_:+.4f} (SD {sd_:.4f}) → "
            + (f"시드 {n_}개면 확정 (추가 {(n_ - len(WANT[MAIN])) * 2 * K_FOLD}회)"
               if n_ else "60시드로도 확정 불가 — 문턱에 너무 가깝다"))
run.log("=" * 100)
````
