## 실험13이 남긴 두 가지 숙제

**① 지표가 틀렸다.** 비중(비율) `Δ(I+II)/Δ(12)` 는 분모가 0에 가까우면 폭발하거나
정의불가가 되는데, **분모가 0이 되는 경우가 하필 가장 강한 양성 사례**다.
STTC 하벽 허혈이 정확히 그랬다 — `{II}` AUC **0.919**(측정된 소집단 중 최고)인데
`Δ(12) = −0.005` 라 비중 정의불가, 그래서 STTC 패널 전체가 판정불가로 끝났다.
**지표가 결론을 못 내는 자리가 결론이 가장 선명한 자리였다.**

**② 표가 두 소견뿐이다.** 차별점 A("무엇을 못 보는가")를 규제·논문에 쓰려면
소견 두 개짜리 표로는 안 된다.

## 새 주지표 — **잔여 결손**

$$D = \mathrm{AUC}(\{12\}) - \mathrm{AUC}(\{\mathrm{I,II}\})$$

- **항상 정의된다** (분모가 없다)
- **바로 읽힌다**: "2유도 웨어러블이 12유도 대비 잃는 AUC"
- 포화 사례에서도 값이 나온다: 단일유도가 이미 다 보면 `D ≈ 0`

같이 낸다: `D₁ = AUC({12}) − AUC({II})` (단일유도 웨어러블) ·
`D_tele = AUC({12}) − AUC({II,V1})` (텔레메트리) · 네 구성의 절대 AUC.

## 사전등록 — 평면 배정을 **해부학만 보고** 먼저 고정한다

수치를 보기 전에 각 소견을 "그 소견을 읽는 유도가 어느 평면인가"로만 분류한다.

| 소견 | 읽는 유도 | 배정 |
|---|---|---|
| 하벽경색 IMI | II · III · aVF | **전두면** |
| 하벽허혈 ISCI | II · III · aVF | **전두면** |
| 속가지차단 LAFB/LPFB | 축편위 = I · aVL · II · III · aVF | **전두면** |
| 전중격/전벽경색 ASMI·AMI | V1–V4 | **횡단면** |
| 전벽허혈 ISCA | V1–V4 | **횡단면** |
| 우각차단 CRBBB·IRBBB | **V1의 rsR′** | **횡단면** |
| 좌각차단 CLBBB·ILBBB | V1 QS + V6 넓은 R | **횡단면** |
| 후벽경색 PMI 계열 | V1–V3 상반변화 | **횡단면** |
| 하측벽 ILMI · 측벽 LMI/ALMI | 하벽/측벽 유도 혼합 | 혼합(비채점) |
| **LVH** | Sokolow(흉부) + Cornell(aVL) | **혼합**(비채점 — 실험10″ 이후의 정직한 입장) |
| 심방확대 LAO/LAE·RAO/RAE | P축(II) + V1 종말음성력 | 혼합(비채점) |
| 1도 AV차단 · IVCD | 간격·폭 — 평면과 무관 | 무관(비채점) |
| 비특이적 ST-T (NST_·STTC·ISC_) | 광범위 | 혼합(비채점) |

**혼합·무관은 채점에서 뺀다.** 넣으면 어느 쪽으로든 끌어다 붙일 수 있어서다.
표에는 전부 싣되 **판정은 전두면 대 횡단면으로만** 한다.

| | 예측 | 성격 |
|---|---|---|
| **G0** | 순수 소집단 n ≥ 50 인 것만 채점 | 실험13과 동일 |
| **P-1 ★주가설** | 평균 `D`(횡단면) − 평균 `D`(전두면) **> 0**, CI가 0 제외 | 소견 3개 이상씩을 묶은 전체 대조 |
| **P-2** | 같은 비교의 **소견 단위 검정**(순위검정 + 평면 라벨 완전순열검정)도 한쪽으로 | 분포 가정 없는 복제. **3분한다** — `p<0.05` 지지 / `p>0.95` 기각(역방향 유의) / 나머지 미결. **검정력이 구조적으로 약하다**: 소견 8개면 달성 가능한 최소 p가 0.018이고, 소집단 하나가 순서를 어기면 바로 0.07로 밀린다. **미결은 반증이 아니다** |
| **P-3 CD 복제** | `D`(속가지차단) < `D`(각차단) | **MI·STTC와 독립인 세 번째 패널** — 같은 superclass 안에서 축(전두면) 대 형태(횡단면) |
| **P-4 포화 복구** | 하벽허혈 ISCI 의 `D` **CI가 0을 포함**(= 결손을 보일 수 없다)하고, 전벽허혈 ISCA 보다 작다 | 실험13이 판정 못 한 자리를 새 지표가 판정하는가. **임의 컷(`\|D\|<0.02`)이 아니라 CI로** 정의한다 — 절대 컷은 n에 따라 난이도가 달라져 불공정하다 |

**P-1이 실패하면** 실험13의 통과가 MI 한 쌍의 우연이었다는 뜻이므로,
차별점 A는 다시 '미검증' 딱지를 단다.

## 이 표가 최종 산출물이다

여기서 나오는 표가 곧 차별점 A — 규제 문서의 "이 기기가 무엇을 못 보는가" 항목이고,
논문의 Table 1이다. 그래서 **소견 하나하나에 n·CI·평면 배정 근거를 붙여** 낸다.


```python
# CELL 1 — 설정 + 평면 배정(해부학만 보고 사전 고정)
!pip -q install wfdb

import os, sys, json, time, ast, numpy as np
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
    DRIVE_ROOT = "/content/drive/MyDrive"
except Exception as e:
    print("⚠️ Colab 아님:", e); DRIVE_ROOT = "/content"
PROJECT = os.path.join(DRIVE_ROOT, "MedKOS", "ecg-model")
sys.path.insert(0, os.path.join(PROJECT, "lib"))
from medkos_run import MedKOSRun

CLASSES = ["NORM", "CD", "STTC", "MI", "HYP"]
CONFIGS = {"II": [1], "I+II": [0, 1], "II+V1": [1, 6], "12": list(range(12))}
K_FOLD, SEED0, BOOT, NMIN = 5, 20260801, 4000, 50

# ── 소견 정의: (superclass, subclass 코드들, 평면, 근거)
#    ★ 평면은 '그 소견을 읽는 유도가 어느 평면인가'만 보고 정했다. 수치를 보기 전이다.
FINDINGS = [
    # ── 전두면 (채점 대상)
    ("하벽경색 IMI",        "MI",   ["IMI"],                   "전두면", "II·III·aVF"),
    ("하벽허혈 ISCI",       "STTC", ["ISCI"],                  "전두면", "II·III·aVF"),
    ("속가지차단 LAFB/LPFB","CD",   ["LAFB/LPFB"],             "전두면", "축편위 = 사지유도"),
    # ── 횡단면 (채점 대상)
    ("전중격·전벽경색",     "MI",   ["ASMI", "AMI"],           "횡단면", "V1–V4"),
    ("전벽허혈 ISCA",       "STTC", ["ISCA"],                  "횡단면", "V1–V4"),
    ("우각차단 CRBBB",      "CD",   ["CRBBB"],                 "횡단면", "V1의 rsR′"),
    ("불완전우각차단 IRBBB","CD",   ["IRBBB"],                 "횡단면", "V1의 rsR′"),
    ("좌각차단 CLBBB",      "CD",   ["CLBBB"],                 "횡단면", "V1 QS + V6 넓은 R"),
    ("불완전좌각차단 ILBBB","CD",   ["ILBBB"],                 "횡단면", "V1·V6 형태"),
    ("후벽경색 PMI계열",    "MI",   ["PMI", "IPMI", "IPLMI"],  "횡단면", "V1–V3 상반변화"),
    # ── 혼합·무관 (표에는 싣되 채점 제외)
    ("하측벽경색 ILMI",     "MI",   ["ILMI"],                  "혼합",   "하벽+측벽"),
    ("측벽경색 LMI/ALMI",   "MI",   ["LMI", "ALMI"],           "혼합",   "I·aVL + V5·V6"),
    ("좌심실비대 LVH",      "HYP",  ["LVH"],                   "혼합",   "Sokolow(흉부)+Cornell(aVL)"),
    ("심방확대",            "HYP",  ["LAO/LAE", "RAO/RAE"],    "혼합",   "P축(II)+V1 종말음성력"),
    ("비특이적 ST-T NST_",  "STTC", ["NST_"],                  "혼합",   "광범위"),
    ("비특이적 ST-T STTC",  "STTC", ["STTC"],                  "혼합",   "광범위"),
    ("비특이적 허혈 ISC_",  "STTC", ["ISC_"],                  "혼합",   "광범위"),
    ("1도 AV차단 _AVB",     "CD",   ["_AVB"],                  "무관",   "PR 간격 — 평면 무관"),
    ("실내전도지연 IVCD",   "CD",   ["IVCD"],                  "무관",   "QRS 폭 — 평면 무관"),
    ("WPW",                 "CD",   ["WPW"],                   "혼합",   "델타파 광범위"),
]
SCORED = ("전두면", "횡단면")

CONFIG = dict(exp="exp13b_deficit_table", quest="ailab-2026-0015",
              parent_exp="exp10p_lead_cv", training="없음(재분석)",
              purpose="차별점 A의 표를 잔여 결손 지표로 재작성 + 평면 대조 확대 검정",
              primary_metric="잔여 결손 D = AUC({12}) - AUC({I,II})",
              why_new_metric=("비중(비율)은 분모≈0에서 폭발/정의불가인데 그 경우가 하필 "
                              "가장 강한 양성(단일유도 포화)이다 — 실험13 STTC가 그랬다"),
              predictions={"G0": f"순수 소집단 n>={NMIN}만 채점",
                           "P-1": "평균 D(횡단면) - 평균 D(전두면) > 0, CI가 0 제외 [주가설]",
                           "P-2": "소견 단위 순위검정 + 완전순열검정 (3분: p<0.05 지지 / p>0.95 기각 / 나머지 미결)",
                           "P-3": "D(속가지차단) < D(각차단) — CD 패널 독립 복제",
                           "P-4": "D(ISCI)의 CI가 0을 포함하고 D(ISCA)보다 작다 — 포화 복구"},
              plane_assignment={f[0]: {"plane": f[3], "basis": f[4]} for f in FINDINGS},
              scored_planes=list(SCORED), k_fold=K_FOLD, boot=BOOT, seed0=SEED0, nmin=NMIN)
np.random.seed(SEED0)
try:
    import matplotlib, matplotlib.font_manager as fm, subprocess
    if not any("Nanum" in f.name for f in fm.fontManager.ttflist):
        subprocess.run(["apt-get", "-qq", "install", "-y", "fonts-nanum"], capture_output=True)
        fm.fontManager.addfont("/usr/share/fonts/truetype/nanum/NanumGothic.ttf")
    matplotlib.rc("font", family="NanumGothic")
    matplotlib.rcParams["axes.unicode_minus"] = False
except Exception as e:
    print("한글 폰트 설정 생략:", e)

run = MedKOSRun("exp13b_deficit_table", CONFIG, project=PROJECT)

REG = os.path.join(PROJECT, "registry.jsonl")
prev = None
for line in (open(REG) if os.path.exists(REG) else []):
    try:
        r = json.loads(line)
    except Exception:
        continue
    if r.get("exp_id") == "exp10p_lead_cv" and os.path.isdir(r.get("dir", "")):
        prev = r
if prev is None:
    raise RuntimeError("registry.jsonl에서 exp10p_lead_cv를 못 찾았습니다")
PREV = prev["dir"]
run.log(f"실험10′ 산출물: {PREV}")

def prev_arm(name):
    p = os.path.join(PREV, "arms", name, "probs.npy")
    return np.load(p) if os.path.exists(p) else None

missing = [f"fixed_{c}_f{k}" for c in CONFIGS for k in range(K_FOLD)
           if prev_arm(f"fixed_{c}_f{k}") is None]
if missing:
    raise RuntimeError(f"실험10′ arm 없음: {missing[:5]} … 총 {len(missing)}개")
run.log(f"✅ 유도고정 arm 확인 · 소견 {len(FINDINGS)}개 · "
        f"채점 대상 평면 {SCORED} · 학습 없음")
```

```python
# CELL 2 — 라벨만 (X 미사용) + OOF 복원
import pandas as pd, subprocess

PTB = "/content/ptbxl"
os.makedirs(PTB, exist_ok=True)
BASE = "https://physionet.org/files/ptb-xl/1.0.3"
for f in ("ptbxl_database.csv", "scp_statements.csv"):
    d = os.path.join(PTB, f)
    if not (os.path.exists(d) and os.path.getsize(d) > 0):
        subprocess.run(["wget", "-q", "-O", d, f"{BASE}/{f}"])
df  = pd.read_csv(os.path.join(PTB, "ptbxl_database.csv"), index_col="ecg_id")
scp = pd.read_csv(os.path.join(PTB, "scp_statements.csv"), index_col=0)

agg    = scp[scp.diagnostic == 1].diagnostic_class.to_dict()
sub_of = scp[scp.diagnostic == 1].diagnostic_subclass.to_dict()
df["sc"] = df.scp_codes.apply(
    lambda s: sorted({agg[k] for k in ast.literal_eval(s) if k in agg}))
sub = df[df.sc.apply(lambda s: len(s) == 1 and s[0] in CLASSES)].copy()

CACHE = run.data("ptbxl_12lead_full.npz")
if not os.path.exists(CACHE):
    raise RuntimeError(f"캐시가 없습니다: {CACHE}")
z = np.load(CACHE, allow_pickle=True)
Y, FOLD10, EID = z["y"], z["fold"], z["eid"]       # ★ z["X"] 미사용
if len(Y) != len(sub) or not np.array_equal(np.sort(EID), np.sort(sub.index.values)):
    raise RuntimeError("캐시가 지금 sub와 다르다")
sub = sub.loc[EID]
sub["subcls"] = sub.scp_codes.apply(
    lambda s: sorted({sub_of[k] for k in ast.literal_eval(s) if k in sub_of}))
run.log(f"레코드 {len(Y):,} · 클래스 {np.bincount(Y, minlength=5).tolist()} ({CLASSES})")

OOF = {}
for c in CONFIGS:
    arr = np.zeros((len(Y), len(CLASSES)))
    for k in range(K_FOLD):
        arr[np.where((FOLD10 - 1) % K_FOLD == k)[0]] = prev_arm(f"fixed_{c}_f{k}")
    OOF[c] = arr
norm_idx = np.where(Y == 0)[0]
run.log(f"OOF 복원 완료 · NORM 대조 {len(norm_idx):,}건")

def alpha_for(prob, target):
    lo, hi = 0.02, 50.0
    for _ in range(40):
        mid = (lo * hi) ** 0.5
        p = prob.copy(); p[:, 0] *= mid
        if float((p.argmax(1)[Y == 0] != 0).mean()) > target: lo = mid
        else: hi = mid
    return hi

ref_fa = float((OOF["II"].argmax(1)[Y == 0] != 0).mean())
PRED = {}
for c in CONFIGS:
    p = OOF[c].copy(); p[:, 0] *= alpha_for(OOF[c], ref_fa)
    PRED[c] = p.argmax(1)
run.log(f"기준 오경보율 = 유도고정 {{II}}의 {ref_fa:.3f} (실험10′·10″·13과 동일 절차)")
```

### CELL 3 — 【G0】 소견별 인구조사

**AUC를 보기 전에 센다** (실험10″의 교훈). `순수`는 그 소견 코드만 가진 레코드,
`포함`은 다른 소견을 겸한 것까지. **주분석은 `순수`** — 겸하면 어느 기전인지 못 가른다.


```python
# CELL 3 — 소견별 인구조사
MEMB, CENSUS = {}, {}
run.log(f"  {'소견':<24}{'평면':<7}{'순수':>7}{'포함':>7}   채점")
for name, cls, codes, plane, basis in FINDINGS:
    idx = np.where(Y == CLASSES.index(cls))[0]
    inc  = np.array([bool(set(sub.subcls.iloc[i]) & set(codes)) for i in idx]) \
           if len(idx) else np.zeros(0, bool)
    pure = (np.array([len(sub.subcls.iloc[i]) > 0
                      and set(sub.subcls.iloc[i]) <= set(codes) for i in idx])
            & inc) if len(idx) else np.zeros(0, bool)
    MEMB[name] = idx[pure] if len(idx) else np.zeros(0, int)
    ok = int(pure.sum()) >= NMIN and plane in SCORED
    CENSUS[name] = {"superclass": cls, "plane": plane, "basis": basis,
                    "순수": int(pure.sum()), "포함": int(inc.sum()), "채점": bool(ok)}
    mark = "✅" if ok else ("— 비채점(혼합·무관)" if plane not in SCORED
                           else f"⚠️ n<{NMIN}")
    run.log(f"  {name:<24}{plane:<7}{pure.sum():>7}{inc.sum():>7}   {mark}")

SC = {p: [n for n in MEMB if CENSUS[n]["plane"] == p and CENSUS[n]["채점"]] for p in SCORED}
run.log(f"\n채점 대상 — 전두면 {len(SC['전두면'])}개 {SC['전두면']}")
run.log(f"           횡단면 {len(SC['횡단면'])}개 {SC['횡단면']}")
if len(SC["전두면"]) < 2 or len(SC["횡단면"]) < 2:
    run.log("⛔ 한쪽 평면의 소견이 2개 미만 — P-1/P-2 판정 불가")
```

```python
# CELL 4 — 잔여 결손 D 측정
from sklearn.metrics import roc_auc_score
from scipy.stats import rankdata, mannwhitneyu

def auc_fast(sp, sn):
    r = rankdata(np.concatenate([sp, sn]), method="average")
    a, b = len(sp), len(sn)
    return float((r[:a].sum() - a * (a + 1) / 2) / (a * b))

_rs = np.random.RandomState(0); _a = _rs.rand(200); _b = _rs.rand(2000)
_a, _b = _rs.choice(_a, 200), _rs.choice(_b, 2000)
_d = abs(auc_fast(_a, _b) - roc_auc_score(np.r_[np.ones(200), np.zeros(2000)], np.r_[_a, _b]))
assert _d < 1e-12, f"auc_fast가 sklearn과 다르다 ({_d:.2e})"
run.log(f"auc_fast 자체검산 통과 (오차 {_d:.1e})")

def auc_of(ci, cfg, g, ni=None):
    n = norm_idx if ni is None else norm_idx[ni]
    return auc_fast(OOF[cfg][g, ci], OOF[cfg][n, ci])

q = lambda v, p: float(np.percentile(v, p)) if len(v) else float("nan")

def measure(name):
    """잔여 결손 D 와 그 부트스트랩 표본.
    ★ NORM 재표본 축은 모든 소견이 공유한다(RNG 분리) → 소견 간 비교가 짝지어진다."""
    cls = CENSUS[name]["superclass"]; ci = CLASSES.index(cls); g = MEMB[name]
    a = {c: auc_of(ci, c, g) for c in CONFIGS}
    rs_g = np.random.RandomState(SEED0)
    rs_n = np.random.RandomState(SEED0 + 1)
    bs = np.empty(BOOT)
    for t in range(BOOT):
        gi = rs_g.randint(0, len(g), len(g))
        ni = rs_n.randint(0, len(norm_idx), len(norm_idx))
        gg = g[gi]
        bs[t] = auc_of(ci, "12", gg, ni) - auc_of(ci, "I+II", gg, ni)
    return {"n": int(len(g)), "auc": a,
            "D": a["12"] - a["I+II"], "D_ci": [q(bs, 2.5), q(bs, 97.5)],
            "D1": a["12"] - a["II"], "D_tele": a["12"] - a["II+V1"],
            "_bs": bs,
            "recall_matched": {c: float((PRED[c][g] == ci).mean()) for c in CONFIGS}}

R, t0 = {}, time.time()
for name in MEMB:
    if len(MEMB[name]) < NMIN:
        continue
    R[name] = measure(name)
run.log(f"측정 완료 {time.time()-t0:.0f}s · 소견 {len(R)}개")

run.log("\n" + "=" * 118)
run.log("【차별점 A】 소견 평면별 잔여 결손 — 2유도 웨어러블이 12유도 대비 잃는 AUC")
run.log("=" * 118)
run.log(f"  {'소견':<24}{'평면':<7}{'n':>6}"
        + "".join(f"{c:>8}" for c in CONFIGS)
        + f"{'D={12}-{I,II}':>15}{'95% CI':>20}{'D1({II})':>10}")
run.log("-" * 118)
for name in sorted(R, key=lambda n: R[n]["D"]):
    m, c_ = R[name], CENSUS[name]
    run.log(f"  {name:<24}{c_['plane']:<7}{m['n']:>6}"
            + "".join(f"{m['auc'][c]:>8.3f}" for c in CONFIGS)
            + f"{m['D']:>+15.3f}"
            + f"   [{m['D_ci'][0]:+.3f}, {m['D_ci'][1]:+.3f}]"
            + f"{m['D1']:>10.3f}"
            + ("" if c_["채점"] else "  (비채점)"))
run.log("-" * 118)
run.log("  D가 작을수록 웨어러블로 충분하다 · D1은 단일유도 {II}의 결손")
```

```python
# CELL 5 — 사전등록 채점
MARK = {True: "✅ 지지", False: "❌ 기각", None: "⚠️ 미결"}
def decide(lo, hi, thr, d):
    if d == ">":
        if lo > thr: return True
        if hi < thr: return False
    else:
        if hi < thr: return True
        if lo > thr: return False
    return None

SCORE = {}
run.log("\n" + "=" * 118)
run.log("【사전등록 채점】")
run.log("=" * 118)

F, T = SC["전두면"], SC["횡단면"]
if len(F) >= 2 and len(T) >= 2:
    # P-1 평균 D 차 (소견 집합은 고정, 소견 내부 표본오차만 전파)
    dif = np.mean([R[n]["_bs"] for n in T], axis=0) - np.mean([R[n]["_bs"] for n in F], axis=0)
    lo, hi = q(dif, 2.5), q(dif, 97.5)
    P1 = decide(lo, hi, 0.0, ">")
    run.log(f"  P-1 ★ 평균 D(횡단면) − 평균 D(전두면) > 0 → {MARK[P1]}")
    run.log(f"        Δ={np.mean(dif):+.4f} [{lo:+.4f}, {hi:+.4f}]  "
            f"(전두면 {np.mean([R[n]['D'] for n in F]):+.3f} vs "
            f"횡단면 {np.mean([R[n]['D'] for n in T]):+.3f})")
    # P-2 소견을 표본으로 보는 두 가지 검정 (부트스트랩과 독립인 축)
    #     ① Mann-Whitney 순위검정  ② 평면 라벨 완전순열검정(정확검정)
    u, p_mw = mannwhitneyu([R[n]["D"] for n in T], [R[n]["D"] for n in F],
                           alternative="greater")
    from itertools import combinations
    allD = [R[n]["D"] for n in F] + [R[n]["D"] for n in T]
    nF, nAll = len(F), len(F) + len(T)
    obs = np.mean(allD[nF:]) - np.mean(allD[:nF])
    perms = [np.mean([allD[i] for i in range(nAll) if i not in set(c)])
             - np.mean([allD[i] for i in c])
             for c in combinations(range(nAll), nF)]
    p_perm = float(np.mean(np.array(perms) >= obs - 1e-12))
    p_min = 1.0 / len(perms)
    # ★ 다른 예측과 같이 3분한다. p=0.07은 '상위 7% 배열'이라 약한 지지이지 반증이 아니다.
    #   기각은 **역방향이 유의할 때**(p > 0.95)만 찍는다.
    P2 = True if p_perm < 0.05 else (False if p_perm > 0.95 else None)
    run.log(f"  P-2  소견 단위 검정 → {MARK[P2]}")
    run.log(f"        순위검정 U={u:.0f}, p={p_mw:.4f}  ·  "
            f"완전순열검정 p={p_perm:.4f} (전조합 {len(perms)}개, 달성가능 최소 p={p_min:.4f})")
    if P2 is not True:
        if p_min > 0.05:
            run.log(f"        ⚠️ 소견 수가 적어 **어떤 결과가 나와도 p<0.05가 불가능**하다 "
                    f"— 이 검정은 미결로만 읽는다")
        else:
            run.log(f"        ⚠️ P-2는 소견 하나하나를 동등하게 세므로 **표본이 작은 소견의 "
                    f"잡음에 순서가 쉽게 뒤집힌다**(n≈100이면 D의 분해능이 ±0.02 수준). "
                    f"P-1이 주가설인 이유가 이것이다 — P-2 미결은 반증이 아니다")
else:
    P1 = P2 = None
    run.log(f"  P-1/P-2 → ⚠️ 판정불가 (전두면 {len(F)}개 · 횡단면 {len(T)}개)")

# P-3 CD 패널 독립 복제
cd_f = [n for n in F if CENSUS[n]["superclass"] == "CD"]
cd_t = [n for n in T if CENSUS[n]["superclass"] == "CD"]
if cd_f and cd_t:
    dif3 = np.mean([R[n]["_bs"] for n in cd_t], axis=0) - np.mean([R[n]["_bs"] for n in cd_f], axis=0)
    l3, h3 = q(dif3, 2.5), q(dif3, 97.5)
    P3 = decide(l3, h3, 0.0, ">")
    run.log(f"  P-3  CD 독립 복제: D(각차단 {cd_t}) > D(속가지차단 {cd_f}) → {MARK[P3]}")
    run.log(f"        Δ={np.mean(dif3):+.4f} [{l3:+.4f}, {h3:+.4f}]")
else:
    P3 = None
    run.log(f"  P-3 → ⚠️ 판정불가 (CD 전두면 {cd_f} · 횡단면 {cd_t})")

# P-4 포화 복구
if "하벽허혈 ISCI" in R and "전벽허혈 ISCA" in R:
    di, da = R["하벽허혈 ISCI"], R["전벽허혈 ISCA"]
    # ★ '포화'는 절대 임계값이 아니라 **결손을 보일 수 없다**로 정의한다.
    #   임의의 |D|<0.02 컷은 n에 따라 달성 난이도가 달라져 공정하지 않다.
    near0 = bool(di["D_ci"][0] <= 0 <= di["D_ci"][1])
    dif4 = da["_bs"] - di["_bs"]
    l4, h4 = q(dif4, 2.5), q(dif4, 97.5)
    smaller = decide(l4, h4, 0.0, ">")
    P4 = None if smaller is None else bool(near0 and smaller)
    run.log(f"  P-4  포화 복구 → {MARK[P4]}")
    run.log(f"        하벽허혈 D={di['D']:+.4f} [{di['D_ci'][0]:+.4f}, {di['D_ci'][1]:+.4f}]"
            f" → 결손 {'없음(CI가 0 포함)' if near0 else '있음(CI가 0 제외)'}")
    run.log(f"        ISCA − ISCI Δ={np.mean(dif4):+.4f} [{l4:+.4f}, {h4:+.4f}] → {MARK[smaller]}")
    run.log(f"        ※ 실험13에서 비중 지표는 이 자리를 '정의불가'로 버렸다 — "
            f"새 지표는 값과 CI를 냈다(그 자체가 지표 교체의 근거)")
else:
    P4 = None
    run.log("  P-4 → ⚠️ 판정불가 (ISCI 또는 ISCA 표본 부족)")

SCORE = {"P-1": P1, "P-2": P2, "P-3": P3, "P-4": P4,
         "frontal": F, "transverse": T,
         "mean_D_frontal": float(np.mean([R[n]["D"] for n in F])) if F else None,
         "mean_D_transverse": float(np.mean([R[n]["D"] for n in T])) if T else None}

run.log("\n" + "=" * 118)
if P1 is True and P3 is True:
    verdict = ("✅ 확증(이중) — 소견 평면 구분이 MI·STTC를 넘어 **CD에서도 독립 복제**됐다. "
               "차별점 A의 표는 검증된 계측 위에 선다")
elif P1 is True:
    verdict = ("✅ 확증 — 전두면 소견의 잔여 결손이 횡단면 소견보다 유의하게 작다. "
               "실험13이 MI 한 쌍에서 본 것이 소견 여러 개로 일반화된다")
elif P1 is False:
    verdict = ("🚨 기각 — 전두면·횡단면 구분이 소견을 넓히면 사라진다. 실험13의 통과는 "
               "MI 한 쌍의 특성이었을 수 있다. **차별점 A는 다시 '미검증' 딱지를 단다**")
else:
    verdict = "⚠️ 미결 — 방향은 맞을 수 있으나 CI가 0을 걸친다. 표는 서술용으로만 쓴다"
run.log(f"▶ {verdict}")
run.log("=" * 118)
```

## 이 표를 어디에 쓰나

`evaluation/differentiator_A_table.json` 이 산출물이다. 그대로 두 곳에 들어간다:

1. **규제·임상 도입 문서의 "이 기기가 무엇을 못 보는가"** — 소견별로 "2유도로 충분 / V1 필요 /
   12유도 필요"를 `D` 와 CI로 말한다. 평균 하나로 뭉개지 않는다.
2. **논문 Table 1** — 축소유도 논문들이 평균만 보고하는 자리에 소견 평면별 분해를 넣는다.

## 결과 읽는 법

| P-1 | P-3 | 뜻 |
|---|---|---|
| ✅ | ✅ | 평면 구분이 **세 superclass에서 독립 복제**됐다. 차별점 A 확정 |
| ✅ | ⚠️ | 일반화는 됐고 CD 복제는 표본에 걸렸다 |
| ❌ | — | 실험13의 통과가 MI 한 쌍의 특성이었다. **차별점 A 미검증으로 되돌린다** |

## 한계 — 그대로 표에 붙여 낸다

- **모델은 소견(subclass)을 배운 적이 없다.** 5-superclass 분류기이므로 `D`는 "그 소견을 가진
  환자에서 superclass 확률이 유도구성에 따라 얼마나 달라지나"이다. **소견 진단 성능이 아니다.**
  표의 제목을 "소견별 진단 성능"으로 쓰면 거짓이 된다.
- **`D`는 이 모델·이 데이터의 값이다.** 용량을 키우거나 소견별 헤드를 두면 달라진다.
  특히 실험13에서 본 **클래스 내부 재배분**(12유도가 한 소집단을 희생해 다른 소집단을
  올리는 현상) 때문에 `D < 0` 이 나올 수 있는데, 이는 정보량이 아니라 최적화 배분의 문제다.
- **혼합·무관 소견은 채점에서 뺐다.** 표에는 싣지만 이들로 결론을 만들지 않는다.
- **PTB-XL 라벨은 자동판독기 유래**라 소견 라벨이 유도 규칙으로 붙었을 가능성이 있다.
