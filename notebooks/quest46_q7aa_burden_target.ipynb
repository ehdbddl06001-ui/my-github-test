## 사전등록 — 표적 모집단 정의

★★ **문턱은 성적표가 아니라 분포로 정한다**(R34 ②). Q7-X 가 이미 사전등록한 규칙을
그대로 쓴다:

```
표적 모집단 = 「그 코호트의 SVEB 부담 **중앙값 이상**」
```

- 부담은 라벨에서 나오지만, 이건 **예측이 아니라 모집단 정의**다. 임상에서 「PAC 부담이
  높은 환자」는 홀터 판독으로 **사전에** 알 수 있는 정보다
- **문턱을 성적표로 고르지 않는다** — 중앙값 하나뿐이고, 다른 문턱을 훑지 않는다
- ★★ **그래서 영점이 필수다.** 부담으로 자르면 쌍이 많아져 추정이 안정되므로, 그 자체가
  값을 올릴 수 있다. **라벨을 층 안에서 치환한 뒤 같은 절차를 밟으면 0.5** 여야 한다

### 판정표 (R29 ②) — ★ 이 표가 갈래의 운명을 정한다

- **AA3 영점이 0.5 를 뜬다** → 부담 절단 자체가 값을 만든다. **AA1 을 읽지 않는다**
- **AA1 ✅ 이고 AA2 는 아니다** → ★ **부담 집중이 실재한다.** 표적 모집단을 확정하고
  **AA5 의 W1** 로 딥러닝 여부를 판단한다
- **AA1 ✅ 인데 AA2 도 ✅** → 부담이 아니라 **추정 안정성**이 올린 것이다. 종결
- **AA1 ⚠️/❌** → ★★ **형태 갈래를 종결한다.** 아래 문장을 최종으로 쓴다:

> 2리드 홀터에서 SVEB 검출 시, RR 기반 리듬 특징을 통제한 뒤 P 파 형태의 증분 효용은
> **ΔAUPRC ≤ +0.014**(상한)다. 이 값은 **측정 오차 보정으로 부풀릴 수 없다** —
> 검출기 잡음을 절반으로 줄여도(λ 0.26→0.49) 판별력은 변하지 않았다(Q7-Z).

⚠️ **단발이다.** 이 런의 결과로 문턱을 바꿔 다시 돌리지 않는다.

````python
# CELL 0 — 공용 사전점검
import numpy as np

def decide(lo, hi, thr, direction):
    if direction not in (">", "<"):
        raise ValueError("direction 은 '>' 또는 '<'")
    if direction == ">":
        if lo > thr: return "✅ 지지"
        if hi < thr: return "❌ 기각"
    else:
        if hi < thr: return "✅ 지지"
        if lo > thr: return "❌ 기각"
    return "⚠️ 미결"

def mde(lo, hi):
    return (hi - lo) / 2.0 if np.isfinite(lo) and np.isfinite(hi) else float("nan")

def boot_mean(v, seed, nb=3000, q=2.5):
    d = np.asarray(v, float); d = d[np.isfinite(d)]
    if len(d) < 3:
        return float("nan"), float("nan"), float("nan"), len(d)
    rng = np.random.RandomState(seed)
    b = [d[rng.randint(0, len(d), len(d))].mean() for _ in range(nb)]
    return (float(d.mean()), float(np.percentile(b, q)),
            float(np.percentile(b, 100 - q)), len(d))

def boot_pair(a, b, fn, seed, nb=2000, q=2.5):
    a = np.asarray(a, float); b = np.asarray(b, float)
    m = np.isfinite(a) & np.isfinite(b); a, b = a[m], b[m]
    if len(a) < 5:
        return float("nan"), float("nan"), float("nan"), len(a)
    rng = np.random.RandomState(seed); out = []
    for _ in range(nb):
        j = rng.randint(0, len(a), len(a))
        try:
            out.append(float(fn(a[j], b[j])))
        except Exception:
            pass
    if len(out) < 50:
        return float(fn(a, b)), float("nan"), float("nan"), len(a)
    return (float(fn(a, b)), float(np.percentile(out, q)),
            float(np.percentile(out, 100 - q)), len(a))

def boot_two(a, b, seed, nb=2000, q=2.5):
    """★ **독립 두 군**의 평균 차(b − a). 짝이 아니므로 **각 군을 따로** 재표집한다.
    ⚠️ `boot_pair` 를 쓰면 길이가 달라 브로드캐스트가 깨진다(1판이 그렇게 죽었다)."""
    a = np.asarray(a, float); a = a[np.isfinite(a)]
    b = np.asarray(b, float); b = b[np.isfinite(b)]
    if len(a) < 3 or len(b) < 3:
        return float("nan"), float("nan"), float("nan"), min(len(a), len(b))
    rng = np.random.RandomState(seed)
    d = [b[rng.randint(0, len(b), len(b))].mean() - a[rng.randint(0, len(a), len(a))].mean()
         for _ in range(nb)]
    return (float(b.mean() - a.mean()), float(np.percentile(d, q)),
            float(np.percentile(d, 100 - q)), min(len(a), len(b)))

def spearman(a, b):
    ra = np.asarray(a, float).argsort().argsort().astype(float)
    rb = np.asarray(b, float).argsort().argsort().astype(float)
    if np.std(ra) < 1e-12 or np.std(rb) < 1e-12:
        return float("nan")
    return float(np.corrcoef(ra, rb)[0, 1])

def need_super(n, half, eff, p80=False):
    if not np.isfinite(half) or abs(eff) < 1e-9 or n < 1:
        return float("nan")
    r = float(n) * (half / abs(eff)) ** 2
    return r * 2.04 if p80 else r

class AssetError(RuntimeError): pass
print("CELL 0 ✅")
````

````python
# CELL 2 — 【AA-0】 자산 · 특징 · AA0 재현
import pandas as pd
run.log("\n" + "=" * 100)
run.log("【AA-0】 SVDB 특징 재구성 · AA0 — Q7-Z 의 S3 재현")
run.log("=" * 100)
VERD, DIFF = {}, {}
def g_(k, v, d):
    VERD[k] = v; run.log(f"  {k:<5}{v}  {d}")

for p_, why in ((SV5, "svdb_labels.py build"), (PDEL, "Q7-P0")):
    if not os.path.exists(p_):
        raise AssetError(f"{p_} 없음 — {why}(R16)")
D5 = np.load(SV5, allow_pickle=True); PD = np.load(PDEL, allow_pickle=True)
PID = np.asarray(D5["pid"]).astype(int); SYM = np.asarray(D5["sym"]).astype(str)
Y3 = np.asarray(D5["y3"]).astype(int)
PRE = np.asarray(D5["pre_rr"], float); POST = np.asarray(D5["post_rr"], float)
if int((np.asarray(PD["pid"]).astype(int) != PID).sum()) or \
   int((np.asarray(PD["sym"]).astype(str) != SYM).sum()):
    raise AssetError("정합 깨짐 — Q7-P0 를 다시 돌린다")
P_IDX = np.asarray(PD["p_idx"]).astype(int); P_SC = np.asarray(PD["p_score"], float)
K = np.where(Y3 >= 0)[0]
RID = PID[K]; Y = Y3[K]; TT = (Y == IDX_S)
pre = PRE[K].astype(float); post = POST[K].astype(float)
pidx0 = P_IDX[K].copy(); psc_asset = P_SC[K].copy()
RS = np.array(sorted(set(RID.tolist())))
XB = np.ascontiguousarray(np.asarray(D5["beat"])[K][:, 0, :]).astype(float)

_S = pd.Series(pre); _G = _S.groupby(pd.Series(RID))
def local_base(k):
    r = np.asarray(_G.apply(lambda x: x.shift(1).rolling(k, min_periods=1).median())).astype(float)
    return np.where(np.isfinite(r), r, pre)
_med = _G.transform("median").to_numpy()
_std = _G.transform("std").to_numpy(); _mean = _G.transform("mean").to_numpy()
f1 = _med - pre
f2 = {k: 1.0 - pre / (local_base(k) + 1e-9) for k in FULL_K}
f3 = post - pre
f4 = np.nan_to_num(_std / (_mean + 1e-9))
RHY = np.c_[f1, np.column_stack([f2[k] for k in RHY_K]), f3, f4,
            np.log1p(pre), np.log1p(post)]

# ── Q7-Z 와 **같은** 비트배열 점수(w=0). 절단 규약도 같다
HW = int(round(100.0 * FS / 1000.0))
def beat_score0(pos, rows):
    X = XB[rows]; p = np.clip(np.asarray(pos, int), 0, X.shape[1] - 1)
    out = np.zeros(len(p))
    for i in range(len(p)):
        q = int(p[i]); a, b = max(q - HW, 0), min(q + HW + 1, X.shape[1])
        seg = X[i, a:b]
        if len(seg) < 5:
            continue
        t = np.arange(len(seg), dtype=float); tc = t - t.mean()
        sl = float((seg * tc).sum() / max((tc ** 2).sum(), 1e-12))
        seg = seg - (sl * tc + seg.mean())
        den = float(np.median(np.abs(seg - np.median(seg)))) + 1e-12
        out[i] = abs(seg[q - a]) / den
    return out

FIRE = pidx0 >= 0
T0 = time.time()
psc = np.zeros(len(K)); psc[FIRE] = beat_score0(pidx0[FIRE], FIRE)
run.log(f"  ({time.time()-T0:.0f}초) 비트 {len(K):,} · 레코드 {len(RS)}")

# ── P 정렬 5차원 (Q7-S′ 와 동일 — AA5 의 W1 에 쓴다)
MISS = pidx0 < 0
pr_ms = np.where(MISS, np.nan, (RPRE - pidx0) / FS * 1000.0)
pr_q8 = np.round(pr_ms / QUANT_MS) * QUANT_MS
nmask = (Y == 0)
pr_ref = np.full(len(pr_q8), np.nan); sc_ref = np.full(len(psc), np.nan)
sc_mad = np.full(len(psc), np.nan)
for u in np.unique(RID):
    m = RID == u; mn = m & nmask
    a = pr_q8[mn][np.isfinite(pr_q8[mn])]; b = psc[mn]
    pr_ref[m] = np.median(a) if len(a) >= 5 else np.nan
    sc_ref[m] = np.median(b) if len(b) >= 5 else np.nan
    sc_mad[m] = (np.median(np.abs(b - np.median(b))) + 1e-9) if len(b) >= 5 else np.nan
PAL = np.c_[psc, np.nan_to_num(pr_q8, nan=0.0), np.nan_to_num(pr_q8 - pr_ref, nan=0.0),
            np.nan_to_num((psc - sc_ref) / sc_mad, nan=0.0), MISS.astype(float)]
_rng = np.random.RandomState(SEED0 + 99)
NOISE = _rng.normal(size=(len(K), N_DIM))

def basis_ext(idx):
    return np.c_[np.ones(len(idx)), f1[idx], f3[idx], f4[idx],
                 np.column_stack([f2[k][idx] for k in FULL_K])]

def resid(v, idx):
    X = basis_ext(idx); y = v[idx]
    okm = np.isfinite(y)
    if okm.sum() < X.shape[1] + 5:
        return np.full(len(idx), np.nan)
    b = np.linalg.lstsq(X[okm], y[okm], rcond=None)[0]
    return y - X @ b

def matched_auc(vsub, idx, perm=None, rng=None):
    tt = TT[idx]; key = np.round(f1[idx]).astype(int)
    if perm == "stratum":
        tt = tt.copy()
        for kk in np.unique(key):
            m = np.where(key == kk)[0]
            tt[m] = tt[m][rng.permutation(len(m))]
    win = tie = tot = 0.0
    for kk in np.unique(key):
        m = np.where(key == kk)[0]
        a = vsub[m[tt[m]]]; b = vsub[m[~tt[m]]]
        a = a[np.isfinite(a)]; b = b[np.isfinite(b)]
        if not len(a) or not len(b):
            continue
        d = a[:, None] - b[None, :]
        win += float((d > 0).sum()); tie += float((d == 0).sum()); tot += float(d.size)
    return ((win + 0.5 * tie) / tot, int(tot)) if tot >= MIN_PAIR else (float("nan"), int(tot))

REC_OK = [r for r in RS
          if (lambda i: TT[i].sum() >= MIN_S and (~TT[i]).sum() >= MIN_N)(np.where(RID == r)[0])]
IDXS = {r: np.where(RID == r)[0] for r in REC_OK}
PER, NPR, RID_OK = [], [], []
for r in REC_OK:
    idx = IDXS[r]
    a, np_ = matched_auc(resid(psc, idx), idx)
    if np.isfinite(a):
        PER.append(a); NPR.append(np_); RID_OK.append(r)
m0, l0, h0, n0 = boot_mean(PER, SEED0 + 51)
d_ref = abs(m0 - REF["s3_all"])
run.log(f"  판정 후보 {len(REC_OK)} · 매칭 성립 {n0}")
run.log(f"  AA0 — S3 전체 {m0:.4f} [{l0:.4f}, {h0:.4f}] vs Q7-Z {REF['s3_all']:.4f} · "
        f"|Δ| {d_ref:.6f}")
if d_ref > TOL_REPRO:
    raise AssetError(f"AA0 실패(|Δ| {d_ref:.4f} > {TOL_REPRO}) — Q7-Z 와 같은 계산이 "
                     "아니면 부분군 비교도 무의미하다")
g_("AA0", "✅ 지지", "Q7-Z 와 같은 계산이다 — 부분군 비교를 얹을 수 있다")
CONFIG["AA0"] = dict(s3=m0, lo=l0, hi=h0, n=int(n0), n_ok=len(REC_OK))
run.save_json("config", CONFIG)
````

````python
# CELL 3 — 【AA-A】 ★★★ AA1 고부담 · AA2 저부담 · AA3 영점
run.log("\n" + "=" * 100)
run.log("【AA-A】 AA1(주) 고부담 · AA2 저부담(대조) · AA3 영점")
run.log("=" * 100)
run.log("  ▸ 표적 모집단 = **코호트 SVEB 부담 중앙값 이상**. 문턱은 이것 하나뿐이고")
run.log("    다른 문턱을 훑지 않는다(R34 ② · 단발)")
run.log("  ▸ ★★ 부담으로 자르면 쌍이 늘어 추정이 안정되므로 **영점이 필수**다")

# ★★ 문턱은 **판정 코호트 전체**(REC_OK)의 중앙값이다 — S3(매칭 가능 34)와
#    W1(판정 56)이 **같은 문턱**을 써야 두 숫자를 나란히 놓을 수 있다.
BURD_ALL = {r: float(TT[IDXS[r]].mean()) for r in REC_OK}
THR = float(np.median(list(BURD_ALL.values())))    # ★ 사전등록: 중앙값 하나
BURD = np.array([BURD_ALL[r] for r in RID_OK])     # S3 쪽(매칭 가능)
HI = BURD >= THR; LO = ~HI
PERa = np.asarray(PER, float); NPRa = np.asarray(NPR, float)
_ba = np.array(list(BURD_ALL.values()))
run.log(f"\n  부담 분포(판정 코호트 {len(REC_OK)}개) — 중앙값 **{THR:.4f}** · "
        f"범위 {_ba.min():.4f}~{_ba.max():.4f}")
run.log(f"  ★ 이 문턱 하나를 S3(매칭 가능 {len(RID_OK)})와 W1(판정 {len(REC_OK)})에 "
        "**똑같이** 적용한다")
run.log(f"  고부담 {int(HI.sum())}개(부담 중앙 {np.median(BURD[HI]):.4f} · "
        f"쌍 중앙 {np.median(NPRa[HI]):,.0f}) · "
        f"저부담 {int(LO.sum())}개({np.median(BURD[LO]):.4f} · {np.median(NPRa[LO]):,.0f})")

# ── AA3 영점 — **같은 절단·같은 추정**을 라벨 치환 뒤에
T1_ = time.time()
NUL = []
for r in RID_OK:
    idx = IDXS[r]; vr = resid(psc, idx); got = []
    for s in range(N_PERM):
        rng = np.random.RandomState(SEED0 + 700 + s)
        a, _ = matched_auc(vr, idx, perm="stratum", rng=rng)
        if np.isfinite(a):
            got.append(a)
    NUL.append(float(np.mean(got)) if got else np.nan)
NULa = np.asarray(NUL, float)
run.log(f"\n  ({time.time()-T1_:.0f}초) AA3 — 영점 (라벨 층내 치환 · reps={N_PERM})")
AA3 = {}
for nm, m_ in (("전체", np.ones(len(PERa), bool)), ("고부담", HI), ("저부담", LO)):
    v_, lo_, hi_, n_ = boot_mean(NULa[m_], SEED0 + 61)
    AA3[nm] = dict(null=v_, lo=lo_, hi=hi_, n=n_, dev=float(abs(v_ - 0.5)))
    run.log(f"    {nm:<6} null **{v_:.4f}** [{lo_:.4f}, {hi_:.4f}] · n={n_} · "
            f"0.5 를 {'덮음' if (np.isfinite(lo_) and lo_ <= 0.5 <= hi_) else '**벗어남**'}")
# ★ 판정은 **CI 가 0.5 를 덮는가** — 절대 편차 문턱은 부분군 크기에 따라 불공평하다
z_ok = all(np.isfinite(AA3[k]["lo"]) and AA3[k]["lo"] <= 0.5 <= AA3[k]["hi"]
           for k in ("고부담", "저부담"))
g_("AA3", "✅ 지지" if z_ok else "❌ 기각",
   ("두 부분군 다 영점 CI 가 **0.5 를 덮는다** — 부담 절단 자체는 값을 만들지 않는다" if z_ok else
    "★ **부담 절단이 영점을 움직인다** — AA1 을 부담의 증거로 읽을 수 없다"))
NULL_HI = AA3["고부담"]["null"] if np.isfinite(AA3["고부담"]["null"]) else 0.5
NULL_LO = AA3["저부담"]["null"] if np.isfinite(AA3["저부담"]["null"]) else 0.5

# ── AA1 · AA2 — **각 부분군의 측정된 null 기준**으로 판정
run.log("\n  ★★★ AA1(주) 고부담 · AA2 저부담 — 판정은 **측정된 null 기준**")
AA12 = {}
for nm, m_, nl in (("고부담", HI, NULL_HI), ("저부담", LO, NULL_LO)):
    v_, lo_, hi_, n_ = boot_mean(PERa[m_], SEED0 + 71)
    AA12[nm] = dict(auc=v_, lo=lo_, hi=hi_, n=n_, null=nl, excess=float(v_ - nl),
                    mde=float(mde(lo_, hi_)))
    run.log(f"    {nm:<6} AUROC **{v_:.4f}** [{lo_:.4f}, {hi_:.4f}] · null {nl:.4f} · "
            f"초과 **{v_-nl:+.4f}** · n={n_} · MDE {mde(lo_,hi_):.4f}")
d1, d2 = AA12["고부담"], AA12["저부담"]
if not z_ok:
    g_("AA1", "⛔ 측정 불가", "영점이 깨져 읽지 않는다(R29 ②)")
    g_("AA2", "⛔ 측정 불가", "동일")
else:
    g_("AA1", decide(d1["lo"], d1["hi"], NULL_HI, ">"),
       f"★★★ 고부담 초과 **{d1['excess']:+.4f}** · CI [{d1['lo']:.4f}, {d1['hi']:.4f}] vs "
       f"null {NULL_HI:.4f} · n={d1['n']}")
    g_("AA2", decide(d2["lo"], d2["hi"], NULL_LO, ">"),
       f"저부담 초과 **{d2['excess']:+.4f}** · CI [{d2['lo']:.4f}, {d2['hi']:.4f}] vs "
       f"null {NULL_LO:.4f} · n={d2['n']}")
    # ★ 성분 병기(R36 ⑤) — 부분군 차이는 두 값과 **함께만**
    dm_, dlo_, dhi_, dn_ = boot_two(PERa[LO], PERa[HI], SEED0 + 81)   # ★ 독립 두 군
    run.log(f"    ▸ 부분군 차(고−저) **{dm_:+.4f}** [{dlo_:+.4f}, {dhi_:+.4f}] "
            f"(성분 고 {d1['auc']:.4f} n={d1['n']} · 저 {d2['auc']:.4f} n={d2['n']} · R36 ⑤)")
    run.log("      ▸ **짝이 아니라 독립 두 군**이므로 각 군을 따로 재표집했다")
    run.log("      ⚠️ 저부담은 n 이 작아 **MDE 가 크다** — 「저부담엔 없다」가 아니라")
    run.log("        **「저부담에선 못 잰다」**일 수 있다. AA2 의 판정 문구를 그대로 읽어라")
    DIFF["AA_sub"] = dict(diff=dm_, lo=dlo_, hi=dhi_)
CONFIG["AA1"] = d1; CONFIG["AA2"] = d2; CONFIG["AA3"] = AA3
CONFIG["burden"] = dict(thr=THR, n_hi=int(HI.sum()), n_lo=int(LO.sum()),
                        med_hi=float(np.median(BURD[HI])), med_lo=float(np.median(BURD[LO])))
run.save_json("config", CONFIG)
````

````python
# CELL 4 — 【AA-B】 AA4 연속 회귀 · ★★ AA5 결정 숫자(부분군 W1)
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score
run.log("\n" + "=" * 100)
run.log("【AA-B】 AA4 — 부담 연속 회귀 · AA5 — **결정 숫자**(부분군 W1)")
run.log("=" * 100)

# ── AA4 — Q7-Z 의 ρ(쌍 수) 재현 + 부담 ρ, 둘 다 영점 대비
run.log(f"  {'공변량':<12}{'관측 ρ':>10}{'영점 ρ':>10}{'영점 CI':>22}  판정")
AA4 = {}
rho_nul = [spearman(NPRa, NULa)]
nm_, nlo, nhi, _ = boot_mean(
    [spearman(NPRa[np.random.RandomState(SEED0 + 900 + s).permutation(len(NPRa))], NULa)
     for s in range(60)], SEED0 + 91)
for nm, cv in (("SVEB 부담", BURD), ("쌍 수(log)", np.log1p(NPRa))):
    rho = spearman(cv, PERa)
    rn = [spearman(cv, NULa[np.random.RandomState(SEED0 + 950 + s).permutation(len(NULa))])
          for s in range(60)]
    zm, zlo, zhi, _ = boot_mean(rn, SEED0 + 92)
    AA4[nm] = dict(rho=float(rho), null=float(zm), lo=float(zlo), hi=float(zhi))
    v = "✅ 영점 초과" if np.isfinite(zhi) and rho > zhi else "⚠️ 영점 안"
    run.log(f"  {nm:<12}{rho:>+10.4f}{zm:>+10.4f}  [{zlo:>+7.4f},{zhi:>+7.4f}]  {v}")
    run.log(f"    (Q7-Z 관측 {REF['rho_pairs']:+.4f})" if "쌍" in nm else "")
g_("AA4", "✅ 지지" if any(AA4[k]["rho"] > AA4[k]["hi"] for k in AA4) else "⚠️ 미결",
   "부담·쌍 수 중 하나 이상이 **영점을 넘는다**"
   if any(AA4[k]["rho"] > AA4[k]["hi"] for k in AA4) else
   "둘 다 영점 범위 안이다 — Q7-Z 의 연관이 재현되지 않는다")

# ── ★★ AA5 — 결정 숫자. **S3 는 기전, W1 이 효용**이다
run.log("\n  ★★ AA5 — **결정 숫자**: 부분군의 W1 (교차환자 ΔAUPRC · `palign − shuf5`)")
run.log("     ▸ Q7-Z 이후 이 값은 **탈감쇠로 부풀릴 수 없다** — 있는 그대로다")

def fit_eval(Xtr, ytr, Xte, yte):
    mu, sd = Xtr.mean(0), Xtr.std(0) + 1e-9
    lr = LogisticRegression(max_iter=3000, C=1.0)
    lr.fit((Xtr - mu) / sd, ytr)
    return average_precision_score(yte, lr.decision_function((Xte - mu) / sd))

def feats(arm, tr, te):
    if arm == "noise5":
        return NOISE[tr], NOISE[te]
    Ztr, Zte = PAL[tr].copy(), PAL[te].copy()
    if arm == "shuf5":
        rr = np.random.RandomState(SEED0 + 7)
        for Z_, m_ in ((Ztr, tr), (Zte, te)):
            for u in np.unique(RID[m_]):
                sel = np.where(RID[m_] == u)[0]
                Z_[sel] = Z_[sel][rr.permutation(len(sel))]
    return Ztr, Zte

def w1_of(recs, tag):
    """★ LORO 를 **그 모집단 안에서** 돈다 — 표적 모집단의 모형이므로."""
    d = {a: [] for a in ("palign", "shuf5", "noise5")}
    pool = np.concatenate([IDXS[r] for r in recs])
    for r in recs:
        te = IDXS[r]; tr = np.setdiff1d(pool, te)
        if TT[te].sum() < 3 or TT[tr].sum() < 10:
            continue
        base = fit_eval(RHY[tr], TT[tr].astype(int), RHY[te], TT[te].astype(int))
        for a in d:
            Ztr, Zte = feats(a, tr, te)
            d[a].append(fit_eval(np.c_[RHY[tr], Ztr], TT[tr].astype(int),
                                 np.c_[RHY[te], Zte], TT[te].astype(int)) - base)
    if len(d["palign"]) < 5:
        run.log(f"    ⚠️ {tag} — LORO 가 선 레코드가 {len(d['palign'])}개뿐이다(R17)")
    out = {}
    for a in d:
        m_, lo_, hi_, n_ = boot_mean(d[a], SEED0 + 101)
        out[a] = dict(mean=m_, lo=lo_, hi=hi_, n=n_)
    pm, plo, phi, pn = boot_pair(np.asarray(d["shuf5"]), np.asarray(d["palign"]),
                                 lambda a, b: float(b.mean() - a.mean()), SEED0 + 111)
    out["W1"] = dict(mean=pm, lo=plo, hi=phi, n=pn, mde=float(mde(plo, phi)))
    run.log(f"    {tag:<8} W1 **{pm:+.4f}** [{plo:+.4f}, {phi:+.4f}] · n={pn} · "
            f"(성분 palign {out['palign']['mean']:+.4f} · shuf5 {out['shuf5']['mean']:+.4f} · "
            f"noise5 {out['noise5']['mean']:+.4f})")
    return out

T2_ = time.time()
# ★ W1 은 **판정 코호트 전체**(REC_OK)에서 — 매칭 가능 여부는 S3 의 조건이지 W1 의 조건이 아니다
HI_R = [r for r in REC_OK if BURD_ALL[r] >= THR]
LO_R = [r for r in REC_OK if BURD_ALL[r] < THR]
run.log(f"    (W1 코호트 — 고부담 {len(HI_R)}개 · 저부담 {len(LO_R)}개 · "
        f"S3 는 각각 {int(HI.sum())}·{int(LO.sum())})")
AA5 = {"고부담": w1_of(HI_R, "고부담"), "저부담": w1_of(LO_R, "저부담")}
run.log(f"    ({time.time()-T2_:.0f}초) 전체 코호트(Q7-S″) W1 "
        f"{REF['w1']['mean']:+.4f} [{REF['w1']['lo']:+.4f}, {REF['w1']['hi']:+.4f}]")
wh = AA5["고부담"]["W1"]
g_("AA5", decide(wh["lo"], wh["hi"], 0.0, ">"),
   f"★★ **고부담 W1 {wh['mean']:+.4f}** [{wh['lo']:+.4f}, {wh['hi']:+.4f}] — "
   f"리듬 기저 {REF['base_auprc']} 대비 상한 "
   f"**{100*wh['hi']/REF['base_auprc']:+.1f}% 상대**")
run.log("     ⚠️ **CI 와 함께만 인용한다**(R36 ①) — 점추정만 쓰면 사실로 굳는다")
CONFIG["AA4"] = AA4
CONFIG["AA5"] = {k: {kk: vv for kk, vv in v.items()} for k, v in AA5.items()}
run.save_json("config", CONFIG)
````
