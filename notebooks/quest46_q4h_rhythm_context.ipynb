## Q4-G 가 내 가설 둘을 반증했다

### ① 「AUROC 는 상위 꼬리를 대변하지 못한다」 → **철회**

```
ρ(AUROC, 달성률)  = +0.8869      ← 강하게 **양수**
ρ(lift,  달성률)  = +0.4388      ← 오히려 낮다
```

AUROC 가 달성률을 **잘 예측한다**. 예외는 4/56(AUROC>0.85·달성<50%)뿐이다. 그리고
「PR-AUC lift 가 상위 꼬리를 더 잘 대변한다」도 틀렸다. **둘 다 철회한다.**
⇒ 달성률을 올리려면 **AUROC 를 올려야** 하고, 그건 **층①(표현)** 이다.

### ② 「예산을 부담에 비례시키면 민감도가 오른다」 → **정반대였다**

```
균등(기록당 300)   민감도 0.7459   FN 5896
적응(k ∝ π̂)       민감도 0.3639   FN 8108     Δ **−0.3821** ❌
오라클(탐욕)        민감도 0.8250   FN 4431     Δ **+0.0791**
영점               −0.3373
```

**비례 배분이 민감도를 반토막 냈다.** 영점이 −0.3373 이므로 대부분이 **구조적**이다 —
평균 민감도는 레코드를 **동등 가중**하므로, S 가 작은 레코드는 **적은 예산으로도 높은
민감도**에 도달한다. 비례 배분은 그들을 굶긴다. 탐욕 오라클(+0.0791)은 **반대 방향**으로
간다. ⇒ 예산은 부담에 **비례하면 안 된다**.

### ③ K2 는 미결이지만 **기준선이 빠졌다**

```
reg     MAE 0.0591  중앙 0.0258  상대 52.2%  ρ +0.1635 [−0.1156, +0.5589]
mean_p  MAE 0.0724  중앙 0.0400  상대 64.5%  ρ −0.0684
```

MAE·중앙·상대오차 **전부 개선**됐고 ρ 도 **유일하게 양수**지만 문턱(+0.0608)을 못 넘었다.
⚠️ 그런데 **상수 예측기**(학습 평균 유병률)를 안 쟀다 — 그게 없으면 MAE 개선을
읽을 수 없다. L1 이 그걸 채운다.

## ★★★ 이번에 넣는 것 — 교과서적 판별자 둘

### (a) 보상성 휴지기 비율 `(pre + post) / (2 × baseline)`

**심방조기박동(PAC)은 불완전 대상성 휴지**(비율 < 2), **심실조기박동은 완전 대상성**(≈ 2).
현재 특징엔 `f3 = post − pre` 만 있는데, **차이지 비율이 아니고 기저선 정규화가 없다**.
합성 확인: PAC 0.940 vs 정상 0.994, PVC 1.000 — `f3` 로는 PAC·PVC 가 둘 다 큰 양수라
못 가른다.

### (b) 리듬 문맥 (±k 박동)

**이 퀘스트가 이미 관측한 것**: Q7-K6 에서 고립 **+0.0101 ✅** · 혼합 +0.0125 ⚠️ ·
**런 −0.2739 ❌**. **연속·이단맥 문맥에서 모델이 무너진다**는 걸 봐 놓고 **그 문맥을
특징으로 넣은 적이 없다.** 직전/직후 상대 RR, ±4 교대지수, 국소 이소성 밀도를 넣는다.

⚠️ **기저선 분위수 교체는 짓기 전에 버렸다** — 레코드 안에서 기저선이 거의 상수면
분위수를 바꿔도 **레코드 내 순위가 안 바뀐다**(합성 확인: 분리도 +1.981 로 동일).

````python
# CELL 0 — 공용 사전점검
import numpy as np

def decide(lo, hi, thr, direction):
    if direction not in (">", "<"):
        raise ValueError("direction 은 '>' 또는 '<'")
    if not (np.isfinite(lo) and np.isfinite(hi) and np.isfinite(thr)):
        return "⚠️ 미결"
    if direction == ">":
        if lo > thr: return "✅ 지지"
        if hi < thr: return "❌ 기각"
    else:
        if hi < thr: return "✅ 지지"
        if lo > thr: return "❌ 기각"
    return "⚠️ 미결"

def mde(lo, hi):
    return (hi - lo) / 2.0 if np.isfinite(lo) and np.isfinite(hi) else float("nan")

def boot_mean(v, seed, nb=3000, q=2.5):
    d = np.asarray(v, float); d = d[np.isfinite(d)]
    if len(d) < 3:
        return float("nan"), float("nan"), float("nan"), len(d)
    rng = np.random.RandomState(seed)
    b = [d[rng.randint(0, len(d), len(d))].mean() for _ in range(nb)]
    return (float(d.mean()), float(np.percentile(b, q)),
            float(np.percentile(b, 100 - q)), len(d))

def boot_pair(a, b, seed, nb=3000, q=2.5):
    a = np.asarray(a, float); b = np.asarray(b, float)
    m = np.isfinite(a) & np.isfinite(b); a, b = a[m], b[m]
    if len(a) < 3:
        return float("nan"), float("nan"), float("nan"), len(a)
    rng = np.random.RandomState(seed)
    d = [(b[j] - a[j]).mean() for j in (rng.randint(0, len(a), len(a)) for _ in range(nb))]
    return (float((b - a).mean()), float(np.percentile(d, q)),
            float(np.percentile(d, 100 - q)), len(a))

def need_super(n, half, eff, p80=False):
    if not np.isfinite(half) or not np.isfinite(eff) or abs(eff) < 1e-9 or n < 1:
        return float("nan")
    r = float(n) * (half / abs(eff)) ** 2
    return r * 2.04 if p80 else r

class AssetError(RuntimeError): pass
print("CELL 0 ✅")
````

````python
# CELL 2 — 【L-0】 코호트 · ★ 특징 네 벌
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score, roc_auc_score

run.log("\n" + "=" * 100)
run.log("【L-0】 코호트 · 특징 네 벌(base · comp · ctx · both)")
run.log("=" * 100)
VERD, NOTE = {}, {}
def g_(k, v, d):
    VERD[k] = v; NOTE[k] = d; run.log(f"  {k:<5}{v}  {d}")

if not os.path.exists(SV5):
    raise AssetError(f"{SV5} 없음(R16)")
D5 = np.load(SV5, allow_pickle=True)
PID = np.asarray(D5["pid"]).astype(int); Y3 = np.asarray(D5["y3"]).astype(int)
PRE = np.asarray(D5["pre_rr"], float); POST = np.asarray(D5["post_rr"], float)
K = np.where(Y3 >= 0)[0]
RID = PID[K]; TT_ = (Y3[K] == IDX_S)
pre = PRE[K].astype(float); post = POST[K].astype(float)
RS = np.array(sorted(set(RID.tolist())))

_S = pd.Series(pre); _G = _S.groupby(pd.Series(RID))
def local_base(k):
    r = np.asarray(_G.apply(lambda x: x.shift(1).rolling(k, min_periods=1).median())).astype(float)
    return np.where(np.isfinite(r), r, pre)
_med = _G.transform("median").to_numpy()
_std = _G.transform("std").to_numpy(); _mean = _G.transform("mean").to_numpy()
BASE12 = local_base(12)

# ── 현행 특징(base)
F_BASE = np.nan_to_num(np.c_[_med - pre,
                             np.column_stack([1.0 - pre / (local_base(k) + 1e-9) for k in RHY_K]),
                             post - pre, np.nan_to_num(_std / (_mean + 1e-9)),
                             np.log1p(np.clip(pre, 0, None)), np.log1p(np.clip(post, 0, None))],
                       nan=0.0, posinf=0.0, neginf=0.0)

# ── ★★★ (a) 보상성 휴지기 비율 — PAC 은 불완전(<2) · PVC 는 완전(≈2)
COMP = (pre + post) / (2.0 * BASE12 + 1e-9)
F_COMP = np.nan_to_num(np.c_[COMP, COMP - 1.0, np.abs(COMP - 1.0),
                             post / (BASE12 + 1e-9), pre / (BASE12 + 1e-9)],
                       nan=0.0, posinf=0.0, neginf=0.0)

# ── ★★★ (b) 리듬 문맥 — Q7-K6 이 「런 −0.2739」로 본 그 문맥을 특징으로 넣는다
def ctx_feats():
    rel = pre / (BASE12 + 1e-9)
    cols = []
    s = pd.Series(rel); grp = s.groupby(pd.Series(RID))
    for lag in range(1, CTX + 1):                       # 직전/직후 상대 RR
        cols.append(np.nan_to_num(grp.shift(lag).to_numpy(), nan=1.0))
        cols.append(np.nan_to_num(grp.shift(-lag).to_numpy(), nan=1.0))
    short = (rel < 0.90).astype(float)                  # 국소 이소성 밀도
    sg = pd.Series(short).groupby(pd.Series(RID))
    cols.append(np.nan_to_num(
        sg.transform(lambda x: x.rolling(2 * CTX + 1, center=True, min_periods=1).mean())
        .to_numpy(), nan=0.0))
    alt = np.abs(np.nan_to_num(grp.shift(1).to_numpy(), nan=1.0) - rel)   # 교대 강도
    cols.append(alt)
    cols.append(np.abs(np.nan_to_num(grp.shift(-1).to_numpy(), nan=1.0) - rel))
    return np.nan_to_num(np.column_stack(cols), nan=0.0, posinf=0.0, neginf=0.0)

F_CTX = ctx_feats()
FEAT = {"base": F_BASE, "comp": np.c_[F_BASE, F_COMP], "ctx": np.c_[F_BASE, F_CTX],
        "both": np.c_[F_BASE, F_COMP, F_CTX]}
run.log(f"  특징 차원 — " + " · ".join(f"{a} {FEAT[a].shape[1]}" for a in ARMS))
run.log(f"  ★★★ `comp` = 보상성 휴지기 비율 5개 — PAC 불완전(<2) vs PVC 완전(≈2)")
run.log(f"  ★★★ `ctx` = ±{CTX} 박동 문맥 {F_CTX.shape[1]}개 — Q7-K6 의 "
        f"「런 {REF['q7k6']['run']:+.4f} ❌」를 특징으로 넣는다")

IDXS = {int(r): np.where(RID == r)[0] for r in RS}
REC_OK = [int(r) for r in RS
          if TT_[IDXS[int(r)]].sum() >= MIN_S and (~TT_[IDXS[int(r)]]).sum() >= MIN_N]
BURD = {r: float(TT_[IDXS[r]].mean()) for r in REC_OK}
NS_ = {r: int(TT_[IDXS[r]].sum()) for r in REC_OK}
NRE = len(REC_OK); MEAN_PREV = float(np.mean([BURD[r] for r in REC_OK]))
DOM_REC = int(max(REC_OK, key=lambda r: NS_[r]))
run.log(f"  레코드 {len(RS)} · 채점 가능 **{NRE}** · 평균 유병률 {MEAN_PREV:.4f}")

SLOPES = []
def make_cal(s, y):
    lr = LogisticRegression(max_iter=3000, C=1e6).fit(np.asarray(s).reshape(-1, 1),
                                                       np.asarray(y).astype(int))
    a, b = float(lr.coef_[0, 0]), float(lr.intercept_[0])
    SLOPES.append(a)
    return lambda v: a * np.asarray(v, float) + b

def split_rest(held):
    rest = sorted([r for r in REC_OK if r != held], key=lambda r: (BURD[r], r))
    dv = [r for i, r in enumerate(rest) if i % DEV_EVERY == 0]
    return [r for r in rest if r not in set(dv)], dv

def loro(X, y_override=None):
    out = np.full(len(K), np.nan)
    for held in REC_OK:
        tr_r, dv_r = split_rest(held)
        tr = np.concatenate([IDXS[r] for r in tr_r]); dv = np.concatenate([IDXS[r] for r in dv_r])
        te = IDXS[held]
        Ftr = X[tr]; fmu, fsd = Ftr.mean(0), Ftr.std(0) + 1e-9
        ytr = TT_[tr].astype(int) if y_override is None else y_override[held]
        lr = LogisticRegression(max_iter=3000, C=1.0).fit((Ftr - fmu) / fsd, ytr)
        sc = lambda ii: lr.decision_function((X[ii] - fmu) / fsd)
        cl = make_cal(sc(dv), TT_[dv])
        out[te] = cl(sc(te))
    return out

def per_auc(L):
    return {r: float(roc_auc_score(TT_[IDXS[r]].astype(int), L[IDXS[r]])) for r in REC_OK}
def per_ap(L):
    return {r: float(average_precision_score(TT_[IDXS[r]].astype(int), L[IDXS[r]]))
            for r in REC_OK}
def at_k(L, r, k):
    idx = IDXS[r]; sc = L[idx]; yy = TT_[idx]
    k = int(min(max(1, k), len(idx)))
    fl = sc >= np.partition(sc, -k)[-k]
    tp = int((fl & yy).sum()); ceil = min(1.0, k / max(1, NS_[r]))
    return dict(sens=tp / max(1, NS_[r]), ppv=tp / max(1, int(fl.sum())), ceil=ceil,
                ach=(tp / max(1, NS_[r])) / ceil if ceil > 0 else np.nan,
                tp=tp, fn=NS_[r] - tp)
CONFIG["cohort"] = dict(n_ok=NRE, mean_prev=MEAN_PREV, dom_rec=DOM_REC,
                        dims={a: int(FEAT[a].shape[1]) for a in ARMS})
run.save_json("config", CONFIG)
````

````python
# CELL 4 — 【L-B】 ★★ L2 리듬 문맥 진단
run.log("\n" + "=" * 100)
run.log("【L-B】 ★★ L2 — 리듬 문맥(이단맥·불규칙성)이 실패를 설명하는가")
run.log("=" * 100)
run.log("  SVDB 는 **심방세동·심방이단맥·조동·방실접합부율동**을 포함한다(PhysioNet 문서)")
BIG, IRR = {}, {}
for r in REC_OK:
    idx = IDXS[r]; p = pre[idx]; med = float(np.median(p))
    sgn = np.sign(p - med)
    BIG[r] = float(np.mean(sgn[:-1] * sgn[1:] < 0))          # 교대 비율(이단맥 대리)
    IRR[r] = float(np.sqrt(np.mean(np.diff(p) ** 2)) / (med + 1e-9))   # RMSSD/중앙(AF 대리)
a0 = "base"
ac = np.array([R300[a0][r]["ach"] for r in REC_OK])
au = np.array([AUC[a0][r] for r in REC_OK])
bg = np.array([BIG[r] for r in REC_OK]); ir = np.array([IRR[r] for r in REC_OK])
pv = np.array([BURD[r] for r in REC_OK])
run.log(f"\n  {'':<16}{'ρ(·, 달성률)':>14}{'ρ(·, AUROC)':>14}{'ρ(·, 유병률)':>14}")
for nm, v in (("이단맥 지수", bg), ("불규칙성(AF 대리)", ir)):
    run.log(f"  {nm:<16}{np.corrcoef(v, ac)[0,1]:>+14.4f}{np.corrcoef(v, au)[0,1]:>+14.4f}"
            f"{np.corrcoef(v, pv)[0,1]:>+14.4f}")
qs = np.quantile(bg, [0, .5, 1.0])
hi_b = [r for r in REC_OK if BIG[r] >= qs[1]]; lo_b = [r for r in REC_OK if BIG[r] < qs[1]]
run.log(f"\n  이단맥 상위 절반({len(hi_b)}) vs 하위({len(lo_b)})")
run.log(f"    달성률 {np.mean([R300[a0][r]['ach'] for r in hi_b]):.4f} vs "
        f"{np.mean([R300[a0][r]['ach'] for r in lo_b]):.4f} · "
        f"AUROC {np.mean([AUC[a0][r] for r in hi_b]):.4f} vs "
        f"{np.mean([AUC[a0][r] for r in lo_b]):.4f} · "
        f"유병률 {np.mean([BURD[r] for r in hi_b]):.4f} vs {np.mean([BURD[r] for r in lo_b]):.4f}")
g_("L2", "(관문 아님)",
   f"이단맥 지수~달성률 ρ {np.corrcoef(bg, ac)[0,1]:+.4f} · 불규칙성~달성률 "
   f"{np.corrcoef(ir, ac)[0,1]:+.4f} · 이단맥~유병률 {np.corrcoef(bg, pv)[0,1]:+.4f}")
CONFIG["L2"] = dict(rho_big_ach=float(np.corrcoef(bg, ac)[0, 1]),
                    rho_big_auc=float(np.corrcoef(bg, au)[0, 1]),
                    rho_big_prev=float(np.corrcoef(bg, pv)[0, 1]),
                    rho_irr_ach=float(np.corrcoef(ir, ac)[0, 1]),
                    rho_irr_auc=float(np.corrcoef(ir, au)[0, 1]),
                    big={str(r): BIG[r] for r in REC_OK},
                    irr={str(r): IRR[r] for r in REC_OK})
run.save_json("config", CONFIG)
````

````python
# CELL 5 — 【L-C】 ★★★ L3 주 관문 · L4 동작점
run.log("\n" + "=" * 100)
run.log("【L-C】 ★★★ L3 — 보상성 휴지기 + 리듬 문맥이 **매크로 AUROC** 를 올리는가")
run.log("=" * 100)
L3 = {}
for a in ARMS[1:]:
    m_, lo_, hi_, n_ = boot_pair([AUC["base"][r] for r in REC_OK],
                                 [AUC[a][r] for r in REC_OK], SEED0 + 41 + len(a), NB_BOOT)
    L3[a] = dict(mean=m_, lo=lo_, hi=hi_, n=int(n_), mde=float(mde(lo_, hi_)))
    run.log(f"  {a:<8} 매크로 AUROC Δ **{m_:+.4f}** [{lo_:+.4f}, {hi_:+.4f}] · MDE "
            f"{L3[a]['mde']:.4f}")

run.log(f"\n  ★★ **영점** — 학습 라벨 치환 뒤 같은 대비 (reps={N_PERM} · `{MAIN_ARM}` 기준)")
nul = {}
for s_ in range(N_PERM):
    rr = np.random.RandomState(SEED0 + 400 + s_)
    yov = {}
    for held in REC_OK:
        tr_r, _ = split_rest(held)
        tr = np.concatenate([IDXS[r] for r in tr_r])
        yov[held] = TT_[tr].astype(int)[rr.permutation(len(tr))]
    la, lb = loro(FEAT["base"], yov), loro(FEAT[MAIN_ARM], yov)
    pa, pb = per_auc(la), per_auc(lb)
    for r in REC_OK:
        nul.setdefault(r, []).append(pb[r] - pa[r])
    run.log(f"    ({time.time()-T0:>5.0f}초) 영점 rep {s_+1}/{N_PERM}")
NS = boot_mean([float(np.mean(v)) for v in nul.values()], SEED0 + 61, NB_BOOT)
run.log(f"    영점 **{NS[0]:+.4f}** [{NS[1]:+.4f}, {NS[2]:+.4f}] (레코드 {NS[3]})")
L3_THR = max(0.0, NS[2]) if np.isfinite(NS[2]) else float("nan")
run.log(f"    ▸ 문턱 = max(0, 영점 상단) **{L3_THR:+.4f}**")
l3v = decide(L3[MAIN_ARM]["lo"], L3[MAIN_ARM]["hi"], L3_THR, ">")
best = max(ARMS[1:], key=lambda a: L3[a]["mean"])
g_("L3", l3v,
   f"`{MAIN_ARM}` 매크로 AUROC {L3[MAIN_ARM]['mean']:+.4f} "
   f"[{L3[MAIN_ARM]['lo']:+.4f}, {L3[MAIN_ARM]['hi']:+.4f}] · 문턱 {L3_THR:+.4f} "
   f"(최선 팔 `{best}` {L3[best]['mean']:+.4f}) — "
   + ("**교과서적 특징이 층①을 실제로 올린다**" if l3v.startswith("✅") else
      ("추가 특징이 오히려 해롭다" if l3v.startswith("❌") else "가르지 못했다(R33 ①)")))

run.log("\n  L4 — 같은 비교를 **동작점**으로")
run.log(f"  {'팔':<8}{'민감도@300 Δ':>18}{'달성률 Δ':>18}{'PPV@300':>10}")
L4 = {}
for a in ARMS[1:]:
    ms, lo_, hi_, _ = boot_pair([R300["base"][r]["sens"] for r in REC_OK],
                                [R300[a][r]["sens"] for r in REC_OK], SEED0 + 51 + len(a),
                                NB_BOOT)
    ma, alo, ahi, _ = boot_pair([R300["base"][r]["ach"] for r in REC_OK],
                                [R300[a][r]["ach"] for r in REC_OK], SEED0 + 52 + len(a),
                                NB_BOOT)
    L4[a] = dict(sens=dict(mean=ms, lo=lo_, hi=hi_), ach=dict(mean=ma, lo=alo, hi=ahi),
                 ppv=float(np.mean([R300[a][r]["ppv"] for r in REC_OK])))
    run.log(f"  {a:<8}{ms:>+9.4f} [{lo_:+.4f},{hi_:+.4f}]{ma:>+9.4f} [{alo:+.4f},{ahi:+.4f}]"
            f"{L4[a]['ppv']:>10.4f}")
run.log(f"  (base PPV@300 {np.mean([R300['base'][r]['ppv'] for r in REC_OK]):.4f})")
run.log(f"\n  ▸ 문헌 대조 — de Chazal 2004 S민감도 {REF['lit'][0][1]:.1%}/+P {REF['lit'][0][2]:.1%} · "
        f"Llamedo 2011 {REF['lit'][1][1]:.0%}/~{REF['lit'][1][2]:.0%}")
run.log(f"    본 런 `{MAIN_ARM}` — 민감도@300 "
        f"{np.mean([R300[MAIN_ARM][r]['sens'] for r in REC_OK]):.1%} · PPV "
        f"{L4[MAIN_ARM]['ppv']:.1%}")
g_("L4", "(관문 아님)",
   f"`{MAIN_ARM}` 민감도@300 Δ {L4[MAIN_ARM]['sens']['mean']:+.4f} · 달성률 Δ "
   f"{L4[MAIN_ARM]['ach']['mean']:+.4f} · PPV {L4[MAIN_ARM]['ppv']:.4f}")
CONFIG["L3"] = dict(diff=L3, thr=float(L3_THR), best=best,
                    null=dict(mean=NS[0], lo=NS[1], hi=NS[2], n=int(NS[3])))
CONFIG["L4"] = L4
run.save_json("config", CONFIG)
````
